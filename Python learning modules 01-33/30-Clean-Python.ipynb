# 3. Naming Conventions

Common Python naming conventions:

| Item | Convention | Example |
|---|---|---|
| Variable | `snake_case` | `customer_name` |
| Function | `snake_case` | `calculate_total()` |
| Class | `PascalCase` | `SalesReport` |
| Constant | `UPPER_SNAKE_CASE` | `MAX_RETRIES` |
| Internal name | leading underscore | `_validate_data` |

Names should communicate purpose.

````python
MAX_RETRIES = 3

customer_name = "Kaif"

def calculate_total(values):
    return sum(values)

class SalesReport:
    pass

print(MAX_RETRIES)
print(customer_name)
print(calculate_total([100, 200]))
print(SalesReport.__name__)
````

# 9. DRY — Don't Repeat Yourself

If the same business logic is copied in multiple places, a future change can
require multiple edits.

Instead:

```text
DUPLICATED LOGIC
      ↓
COMMON FUNCTION
      ↓
REUSE
```

DRY does not mean forcing every similar-looking line into one abstraction.
Abstract repeated **meaningful behavior**.

````python
def calculate_tax(amount: float, rate: float = 0.18) -> float:
    return amount * rate

print(calculate_tax(1000.0))
print(calculate_tax(2500.0))
````

# 10. Before and After DRY

### Repetition

```python
tax_a = 1000 * 0.18
tax_b = 2500 * 0.18
```

### Reusable logic

```python
def calculate_tax(amount, rate=0.18):
    return amount * rate
```

Now the business rule has one clear location.

````python
def calculate_tax(amount: float, rate: float = 0.18) -> float:
    return amount * rate

amounts = [1000.0, 2500.0]

for amount in amounts:
    print(calculate_tax(amount))
````

# 11. Comments

Comments should explain **why**, not simply repeat what the code already says.

Weak:

```python

```

Better:

```python

```

````python
processed_count = 0

for transaction in ["A", "B", "C"]:
    # Count only transactions that successfully reached processing.
    processed_count += 1

print("Processed:", processed_count)
````

# 12. Docstrings

A docstring documents a module, function, class, or method.

Example:

```python
def calculate_total(values):
    """Return the sum of numeric values."""
```

Docstrings are useful for developers, documentation tools, and IDEs.

````python
def calculate_total(values: list[float]) -> float:
    """Return the total of a sequence of numeric values."""
    return sum(values)

print(calculate_total.__doc__)
print(calculate_total([100.0, 200.0]))
````

# 13. Useful Function Docstrings

For public functions, explain:

- purpose
- important parameters
- return value
- important exceptions
- assumptions when necessary

Keep the docstring proportional to the complexity of the function.

````python
def calculate_discount(
    price: float,
    rate: float,
) -> float:
    """Return price after applying a fractional discount rate.

    `rate=0.10` represents a 10% discount.
    """
    return price * (1 - rate)

print(calculate_discount(1000.0, 0.10))
````

# 14. Type Hints

Type hints make interfaces clearer.

```python
def add(a: int, b: int) -> int:
    return a + b
```

They improve:

- readability
- IDE support
- static analysis
- maintainability

Type hints do not automatically enforce types at runtime.

````python
def add(a: int, b: int) -> int:
    return a + b

def average(values: list[float]) -> float:
    if not values:
        return 0.0
    return sum(values) / len(values)

print(add(10, 20))
print(average([10.0, 20.0, 30.0]))
````

# 15. Readability

# 17. Error Handling

Clean code handles expected errors deliberately.

```python
try:
    ...
except ValueError:
    ...
```

Avoid hiding all errors with a broad, silent exception handler.

````python
def parse_amount(value: str) -> float:
    try:
        return float(value)
    except ValueError as error:
        raise ValueError("Amount must be numeric") from error

print(parse_amount("2500.50"))
````

# 18. Validate at Boundaries

External input should be validated before entering core business logic.

```text
EXTERNAL INPUT
      ↓
VALIDATION
      ↓
CLEAN INTERNAL DATA
      ↓
BUSINESS LOGIC
```

This keeps the core logic easier to reason about.

````python
def validate_amount(amount: float) -> float:
    if amount < 0:
        raise ValueError("Amount cannot be negative")
    return amount

amount = validate_amount(2500.0)

print("Validated:", amount)
````

# 19. Avoid Silent Failures

This is usually dangerous:

```python
try:
    process()
except Exception:
    pass
```

It can hide real defects.

If an error is expected, catch an appropriate exception and handle it
explicitly.

````python
def parse_integer(value: str) -> int:
    try:
        return int(value)
    except ValueError:
        return 0

print(parse_integer("100"))
print(parse_integer("invalid"))
````

Instead of:

# 🚀 Project 1 — 01 — Clean Sales Calculator

## Business Problem

Refactor sales calculations into small, typed, reusable functions.

## Architecture

```text
INPUT
 ↓
VALIDATION
 ↓
SMALL FUNCTIONS
 ↓
MODULAR BUSINESS LOGIC
 ↓
ERROR HANDLING
 ↓
REPORT
 ↓
TESTABLE OUTPUT
```

## Implementation

````python
from typing import Iterable

DEFAULT_TAX_RATE = 0.18

def calculate_total(values: Iterable[float]) -> float:
    """Return the total of sales values."""
    return sum(values)

def calculate_tax(
    amount: float,
    rate: float = DEFAULT_TAX_RATE,
) -> float:
    """Return tax calculated from an amount and rate."""
    if amount < 0:
        raise ValueError("Amount cannot be negative")
    return amount * rate

def calculate_final_amount(
    amount: float,
    rate: float = DEFAULT_TAX_RATE,
) -> float:
    """Return amount including calculated tax."""
    return amount + calculate_tax(amount, rate)

sales = [1200.0, 2500.0, 1800.0]

total = calculate_total(sales)
final_total = calculate_final_amount(total)

print("Total:", total)
print("Final with tax:", final_total)
````

# 🚀 Project 2 — 02 — Clean Customer Validator

## Business Problem

Build a readable validation layer for customer records.

## Architecture

```text
INPUT
 ↓

 ↓
SMALL FUNCTIONS
 ↓
MODULAR BUSINESS LOGIC
 ↓
ERROR HANDLING
 ↓
REPORT
 ↓

```

## Implementation

````python
from typing import TypedDict

class Customer(TypedDict):
    name: str
    age: int
    email: str

def validate_customer(customer: Customer) -> None:
    """Raise ValueError when required customer data is invalid."""
    if not customer["name"].strip():
        raise ValueError("Customer name is required")

    if customer["age"] < 18:
        raise ValueError("Customer must be at least 18")

    if "@" not in customer["email"]:
        raise ValueError("Customer email is invalid")

customer: Customer = {
    "name": "Kaif",
    "age": 22,
    "email": "kaif@example.com",
}

validate_customer(customer)

print("Customer is valid:", customer)
````

# 🚀 Project 4 — 04 — Modular Report Builder

## Business Problem

Separate data preparation, metric calculation, and report formatting.

## Architecture

```text
INPUT
 ↓
VALIDATION
 ↓
SMALL FUNCTIONS
 ↓
MODULAR BUSINESS LOGIC
 ↓
ERROR HANDLING
 ↓
REPORT
 ↓
TESTABLE OUTPUT
```

## Implementation

````python
def clean_values(values: list[float]) -> list[float]:
    """Return non-negative sales values."""
    return [value for value in values if value >= 0]

def calculate_metrics(values: list[float]) -> dict[str, float]:
    """Return basic sales metrics."""
    total = sum(values)
    average = total / len(values) if values else 0.0

    return {
        "total": total,
        "average": average,
    }

def format_report(metrics: dict[str, float]) -> str:
    """Format sales metrics for display."""
    return (
        f"Total: {metrics['total']:.2f}\n"
        f"Average: {metrics['average']:.2f}"
    )

raw_sales = [1200.0, 2500.0, -100.0, 1800.0]

sales = clean_values(raw_sales)
metrics = calculate_metrics(sales)

print(format_report(metrics))
````

# 🚀 Project 5 — 05 — Clean Inventory Service

## Business Problem

Create a small inventory service with clear responsibilities and error handling.

## Architecture

```text
INPUT
 ↓
VALIDATION
 ↓
SMALL FUNCTIONS
 ↓
MODULAR BUSINESS LOGIC
 ↓
ERROR HANDLING
 ↓
REPORT
 ↓
TESTABLE OUTPUT
```

## Implementation

````python
class Inventory:
    """Manage product quantities for a small inventory."""

    def __init__(self, stock: dict[str, int]):
        self.stock = stock

    def available(self, product: str) -> bool:
        """Return whether a product has positive stock."""
        return self.stock.get(product, 0) > 0

    def remove(self, product: str, quantity: int) -> None:
        """Remove quantity from stock or raise ValueError."""
        if quantity <= 0:
            raise ValueError("Quantity must be positive")

        available_quantity = self.stock.get(product, 0)

        if available_quantity < quantity:
            raise ValueError("Insufficient stock")

        self.stock[product] -= quantity

inventory = Inventory({
    "Laptop": 5,
    "Mouse": 10,
})

inventory.remove("Laptop", 2)

print("Laptop available:", inventory.available("Laptop"))
print("Stock:", inventory.stock)
````

# 🚀 Project 6 — 06 — Clean File Processing

## Business Problem

Separate file I/O from parsing logic.

## Architecture

```text
INPUT
 ↓
VALIDATION
 ↓
SMALL FUNCTIONS
 ↓
MODULAR BUSINESS LOGIC
 ↓
ERROR HANDLING
 ↓
REPORT
 ↓
TESTABLE OUTPUT
```

## Implementation

````python
from pathlib import Path

def read_lines(path: Path) -> list[str]:
    """Read a UTF-8 text file and return its lines."""
    return path.read_text(encoding="utf-8").splitlines()

def parse_sales(lines: list[str]) -> list[float]:
    """Convert non-empty lines to sales amounts."""
    values: list[float] = []

    for line in lines:
        line = line.strip()

        if not line:
            continue

        try:
            values.append(float(line))
        except ValueError as error:
            raise ValueError(f"Invalid sales value: {line}") from error

    return values

path = Path("sales_clean.txt")
path.write_text(
    "1200\n2500\n1800\n",
    encoding="utf-8",
)

lines = read_lines(path)
sales = parse_sales(lines)

print("Sales:", sales)
print("Total:", sum(sales))
````

# 🚀 Project 7 — 07 — Clean Order Service

## Business Problem

Use small functions, type hints, validation, and a clear business workflow.

## Architecture

```text
INPUT
 ↓
VALIDATION
 ↓
SMALL FUNCTIONS
 ↓
MODULAR BUSINESS LOGIC
 ↓
ERROR HANDLING
 ↓
REPORT
 ↓
TESTABLE OUTPUT
```

## Implementation

````python
from typing import TypedDict

class Order(TypedDict):
    order_id: int
    customer: str
    quantity: int
    unit_price: float

def validate_order(order: Order) -> None:
    """Validate the required order business rules."""
    if order["order_id"] <= 0:
        raise ValueError("Order ID must be positive")
    if not order["customer"].strip():
        raise ValueError("Customer is required")
    if order["quantity"] <= 0:
        raise ValueError("Quantity must be positive")
    if order["unit_price"] < 0:
        raise ValueError("Unit price cannot be negative")

def order_total(order: Order) -> float:
    """Return the total value of an order."""
    validate_order(order)
    return order["quantity"] * order["unit_price"]

order: Order = {
    "order_id": 1001,
    "customer": "Kaif",
    "quantity": 2,
    "unit_price": 65000.0,
}

print("Order total:", order_total(order))
````

# 🚀 Project 8 — 08 — Clean Analytics Pipeline

## Business Problem

Build a reusable pipeline with clear stages.

## Architecture

```text
INPUT
 ↓
VALIDATION
 ↓
SMALL FUNCTIONS
 ↓
MODULAR BUSINESS LOGIC
 ↓
ERROR HANDLING
 ↓
REPORT
 ↓
TESTABLE OUTPUT
```

## Implementation

````python
from collections.abc import Callable

def filter_positive(
    values: list[float],
) -> list[float]:
    """Keep only non-negative values."""
    return [value for value in values if value >= 0]

def transform(
    values: list[float],
    operation: Callable[[float], float],
) -> list[float]:
    """Apply an operation to each value."""
    return [operation(value) for value in values]

def summarize(values: list[float]) -> dict[str, float | int]:
    """Return count, total, and average."""
    total = sum(values)
    count = len(values)

    return {
        "count": count,
        "total": total,
        "average": total / count if count else 0.0,
    }

raw_sales = [1000.0, 2500.0, -300.0, 1800.0]

clean_sales = filter_positive(raw_sales)
taxed_sales = transform(clean_sales, lambda value: value * 1.18)
report = summarize(taxed_sales)

print(report)
````

# 🚀 Project 10 — 10 — Production-Style Clean Python Project

## Business Problem

Combine naming, modular functions, typed records, validation, error handling, reporting, and testable business logic.

## Architecture

```text
INPUT
 ↓
VALIDATION
 ↓
SMALL FUNCTIONS
 ↓
MODULAR BUSINESS LOGIC
 ↓
ERROR HANDLING
 ↓
REPORT
 ↓
TESTABLE OUTPUT
```

## Implementation

````python
from typing import TypedDict

DEFAULT_TAX_RATE = 0.18

class Sale(TypedDict):
    sale_id: int
    customer: str
    amount: float

def validate_sale(sale: Sale) -> None:
    """Validate a sale record."""
    if sale["sale_id"] <= 0:
        raise ValueError("Sale ID must be positive")

    if not sale["customer"].strip():
        raise ValueError("Customer is required")

    if sale["amount"] < 0:
        raise ValueError("Sale amount cannot be negative")

def clean_sales(sales: list[Sale]) -> list[Sale]:
    """Return valid sales and skip invalid records with diagnostics."""
    valid_sales: list[Sale] = []

    for sale in sales:
        try:
            validate_sale(sale)
        except ValueError as error:
            print("Skipped sale:", sale, "|", error)
            continue

        valid_sales.append(sale)

    return valid_sales

def calculate_report(
    sales: list[Sale],
) -> dict[str, float | int]:
    """Build core sales KPIs."""
    total = sum(sale["amount"] for sale in sales)
    count = len(sales)
    average = total / count if count else 0.0
    tax = total * DEFAULT_TAX_RATE

    return {
        "count": count,
        "total": total,
        "average": average,
        "tax": tax,
        "total_with_tax": total + tax,
    }

def main() -> None:
    """Run the clean sales analytics workflow."""
    sales: list[Sale] = [
        {"sale_id": 1, "customer": "Kaif", "amount": 65000.0},
        {"sale_id": 2, "customer": "Sara", "amount": 18000.0},
        {"sale_id": 3, "customer": "Ali", "amount": -500.0},
        {"sale_id": 4, "customer": "Maya", "amount": 42000.0},
    ]

    valid_sales = clean_sales(sales)
    report = calculate_report(valid_sales)

    print("\nFINAL REPORT")
    for key, value in report.items():
        print(f"{key}: {value}")

if __name__ == "__main__":
    main()
````
