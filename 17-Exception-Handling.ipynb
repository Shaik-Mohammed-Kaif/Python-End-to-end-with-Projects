# 1. What Is an Error?

An error means something has gone wrong in a program.

Python programs commonly involve three broad categories:

### 1. Syntax Error

Python cannot parse the code.

```python
if True
    print("Hello")
```

The missing `:` causes a `SyntaxError`.

### 2. Runtime Error / Exception

The code is syntactically valid, but something goes wrong while it runs.

Examples:

```python
10 / 0
```

→ `ZeroDivisionError`

```python
int("abc")
```

→ `ValueError`

### 3. Logical Error

The program runs but produces the wrong result.

Example:

```python
price = 100
discount = 20

final_price = price + discount
```

No exception is necessarily raised, but the business logic is incorrect.

**Important:** exception handling mainly deals with runtime exceptions. Syntax
and logical errors require different fixes.

````python
# Syntax errors are normally discovered before execution.
# This example demonstrates a runtime exception.

number = 10
divisor = 0

try:
    result = number / divisor
except ZeroDivisionError:
    result = None

print("Result:", result)
````

# 2. What Is an Exception?

An exception is an object representing an abnormal condition during program
execution.

When Python encounters an unhandled exception:

```text
Program starts
     ↓
Exception occurs
     ↓
No matching handler
     ↓
Traceback
     ↓
Program stops
```

Exception handling changes the flow:

```text
Program starts
     ↓
Risky operation
     ↓
Exception?
   ↙       ↘
 No        Yes
 ↓          ↓
continue   except handler
             ↓
          recover/report
```

# 3. Basic `try` / `except`

### Syntax

```python
try:
    risky_code()
except SomeException:
    recovery_code()
```

Put only the operations that can reasonably fail inside the `try` block.

````python
try:
    number = int("100")
    print(10 / number)
except ValueError:
    print("The value is not a valid integer.")
except ZeroDivisionError:
    print("Cannot divide by zero.")
````

# 4. Catching a Specific Exception

Prefer specific exception types.

Good:

```python
except ValueError:
```

Less useful:

```python
except Exception:
```

A broad handler can be appropriate at a program boundary for logging or
controlled reporting, but it should not hide programming bugs.

````python
try:
    age = int("twenty")
except ValueError:
    print("Please provide a numeric age.")
````

# 5. Multiple `except` Blocks

Different exceptions can have different recovery behavior.

````python
value = "abc"

try:
    number = int(value)
    result = 100 / number
except ValueError:
    print("Invalid numeric value.")
except ZeroDivisionError:
    print("Division by zero is not allowed.")
````

# 6. Catching Multiple Exceptions Together

When the handling is identical, multiple exception types can be grouped:

```python
except (TypeError, ValueError):
```

This keeps the handler concise.

````python
values = ["10", "20", "abc"]

for value in values:
    try:
        number = int(value)
        print(number)
    except (TypeError, ValueError):
        print(f"Invalid value: {value}")
````

# 7. `else`

The `else` block runs only when the `try` block completes without raising an
exception.

### Structure

```python
try:
    ...
except SomeError:
    ...
else:
    ...
```

A useful pattern is:

```text
try   → risky operation
except → failure handling
else   → success handling
```

````python
try:
    number = int("25")
except ValueError:
    print("Invalid number.")
else:
    print("Conversion successful:", number)
````

# 8. `finally`

The `finally` block runs whether an exception occurs or not.

Typical use cases include cleanup operations.

```python
try:
    ...
except:
    ...
finally:
    cleanup()
```

`finally` is especially useful for resources that must be released.

````python
try:
    print("Processing...")
except Exception:
    print("An error occurred.")
finally:
    print("Cleanup/finalization step executed.")
````

# 9. Complete `try` / `except` / `else` / `finally`

```text
try
 ↓
Success? ─────────→ else
 ↓
Exception
 ↓
except
 ↓
finally
```

`finally` is executed regardless of whether the protected operation succeeds
or fails, subject to Python's normal control-flow behavior.

````python
try:
    value = int("50")
    result = 100 / value
except ValueError:
    print("Invalid integer.")
except ZeroDivisionError:
    print("Cannot divide by zero.")
else:
    print("Result:", result)
finally:
    print("Operation finished.")
````

# 10. `raise`

Use `raise` when your program should explicitly signal an exception.

### Syntax

```python
raise ExceptionType("message")
```

This is useful for enforcing business rules and validating inputs.

````python
def withdraw(balance, amount):
    if amount < 0:
        raise ValueError("Amount cannot be negative.")

    if amount > balance:
        raise ValueError("Insufficient balance.")

    return balance - amount

try:
    print(withdraw(5000, 6000))
except ValueError as error:
    print("Transaction rejected:", error)
````

# 11. Raising a Specific Built-in Exception

Choose an exception that communicates the problem clearly.

Examples:

```python
raise ValueError(...)
raise TypeError(...)
raise KeyError(...)
```

Do not raise a random exception merely to make an error disappear.

````python
def calculate_discount(price, rate):
    if not isinstance(price, (int, float)):
        raise TypeError("price must be numeric")

    if not 0 <= rate <= 1:
        raise ValueError("rate must be between 0 and 1")

    return price * (1 - rate)

print(calculate_discount(1000, 0.20))
````

# 12. `as` — Accessing the Exception Object

You can store the exception object:

```python
except ValueError as error:
```

This gives access to the exception message and other exception information.

````python
try:
    int("Python")
except ValueError as error:
    print("Exception type:", type(error).__name__)
    print("Message:", error)
````

This catches almost everything that inherits from `BaseException`.

```python

    ...

    ...
```

Avoid bare `except` in normal application code because it can catch things
you usually should not suppress, including `KeyboardInterrupt` and
`SystemExit`.

Prefer specific exceptions or, when genuinely necessary at a boundary:

```python
except Exception:
```


````python
try:
    result = int("abc")
except ValueError:
    print("Handled specifically instead of using bare except.")
````

# 14. `Exception` vs `BaseException`

Most application errors inherit from `Exception`.

`BaseException` is higher in the hierarchy and includes control-flow-related
exceptions such as `KeyboardInterrupt`, `SystemExit`, and `GeneratorExit`.

### Practical rule

```text
Catch Exception
when you intentionally need a broad application-level boundary.

Do not normally catch BaseException.
```

````python
try:
    raise ValueError("Example application error")
except Exception as error:
    print(type(error).__name__, "-", error)
````

# 25. Exception Hierarchy — Big Picture

Python exceptions form an inheritance hierarchy.

A simplified view:

```text
BaseException
├── KeyboardInterrupt
├── SystemExit
├── GeneratorExit
└── Exception
    ├── ValueError
    ├── TypeError
    ├── NameError
    ├── LookupError
    │   ├── IndexError
    │   └── KeyError
    ├── OSError
    │   ├── FileNotFoundError
    │   ├── PermissionError
    │   └── ...
    └── ImportError
        └── ModuleNotFoundError
```

The hierarchy matters because a handler for a parent exception can also catch
instances of its child exceptions.

````python
try:
    numbers = [1, 2]
    print(numbers[10])
except LookupError as error:
    print("Lookup-related problem:", type(error).__name__)
````

# 26. Exception Ordering

Put more specific handlers before broader handlers.

Good:

```python
except FileNotFoundError:
    ...
except OSError:
    ...
```

If `OSError` comes first, it can catch `FileNotFoundError` before the specific
handler gets a chance.

````python
from pathlib import Path

try:
    Path("missing.txt").read_text(encoding="utf-8")
except FileNotFoundError:
    print("Specific: file was not found.")
except OSError:
    print("General OS-level problem.")
````

# 27. Nested `try` Blocks

A `try` block can appear inside another `try` block, but nesting should have a
clear reason.

Prefer small, focused error boundaries rather than deeply nested exception
handling.

````python
try:
    value = "10"

    try:
        number = int(value)
        print(100 / number)
    except ZeroDivisionError:
        print("Inner handler: zero divisor.")

except ValueError:
    print("Outer handler: invalid integer.")
````

# 28. Re-Raising an Exception

Inside an exception handler, plain `raise` re-raises the currently handled
exception.

This is useful when a lower-level function logs/contextualizes a problem but
the caller still needs to handle it.

````python
def process():
    try:
        int("not-a-number")
    except ValueError:
        print("Logging validation failure...")
        raise

try:
    process()
except ValueError:
    print("Caller handled the re-raised exception.")
````

# 29. Exception Chaining — `raise ... from ...`

When converting one exception into another, `raise ... from ...` preserves the
relationship between the original and new exceptions.

This is valuable when a low-level implementation detail should be translated
into a domain-specific error.

````python
def parse_customer_age(value):
    try:
        return int(value)
    except ValueError as error:
        raise ValueError("Customer age must be numeric.") from error

try:
    parse_customer_age("unknown")
except ValueError as error:
    print(error)
````

# 30. Custom Exceptions

Create a custom exception by inheriting from `Exception`.

```python
class InvalidCustomerError(Exception):
    pass
```

Use custom exceptions when a domain-specific failure deserves a meaningful
name.

````python
class InvalidCustomerError(Exception):
    pass

def validate_customer(age):
    if age < 18:
        raise InvalidCustomerError("Customer must be at least 18.")

try:
    validate_customer(16)
except InvalidCustomerError as error:
    print("Validation error:", error)
````

# 31. Custom Exception With Additional Data

A custom exception can store useful context.

Keep the design simple and purposeful.

````python
class InsufficientBalanceError(Exception):
    def __init__(self, balance, requested):
        self.balance = balance
        self.requested = requested
        super().__init__(
            f"Balance {balance} is insufficient for {requested}."
        )

def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientBalanceError(balance, amount)
    return balance - amount

try:
    withdraw(5000, 7000)
except InsufficientBalanceError as error:
    print(error)
    print("Available:", error.balance)
    print("Requested:", error.requested)
````

# 32. Exception Handling With Functions

A strong design often separates:

```text
Business logic
      ↓
raise meaningful exception
      ↓
Application boundary
      ↓
Handle / report / recover
```

Do not automatically catch an exception inside the same function that could
instead communicate failure clearly to its caller.

````python
class InvalidAmountError(ValueError):
    pass

def calculate_total(price, quantity):
    if quantity < 0:
        raise InvalidAmountError("Quantity cannot be negative.")
    return price * quantity

try:
    print(calculate_total(100, 3))
except InvalidAmountError as error:
    print("Could not calculate:", error)
````

# 33. `assert` vs `raise`

`assert` is mainly for internal assumptions and debugging.

```python
assert total >= 0
```

It should not be used as the primary mechanism for validating user input or
business rules because assertions can be disabled.

For user/business validation, explicitly raise an appropriate exception.

````python
def calculate_total(price, quantity):
    if quantity < 0:
        raise ValueError("quantity cannot be negative")

    return price * quantity

print(calculate_total(100, 5))
````

# 34. File Handling + Exception Handling

File operations are a common real-world place for exceptions.

Good practice:

```text
Open/read
  ↓
Handle expected file errors
  ↓
Process
  ↓
Cleanup automatically with `with`
```

Use a context manager for files; it handles normal resource cleanup for you.

````python
from pathlib import Path

path = Path("sample.txt")

try:
    with path.open("r", encoding="utf-8") as file:
        content = file.read()
except FileNotFoundError:
    print("File does not exist.")
else:
    print("Characters:", len(content))
````

# 35. Exception Handling Best Practices

### DO

- Catch specific exceptions.
- Keep `try` blocks small.
- Provide useful recovery or reporting.
- Use `raise` for invalid business states.
- Preserve useful exception context.
- Use `finally` or context managers for cleanup.
- Create custom exceptions for meaningful domain errors.
- Log exceptions at appropriate application boundaries.

### DON'T

- Use `except: pass` to hide failures.
- Catch `Exception` everywhere.
- Use exceptions for ordinary control flow when a normal condition check is
  clearer.
- Return meaningless values simply to avoid handling errors.
- Expose sensitive internal error details to end users.

# 36. Common Mistakes

### Mistake 1 — Catching everything

```python
except:
    pass
```

This hides real problems.

### Mistake 2 — Huge `try` blocks

If ten unrelated operations are inside one `try`, it becomes unclear which
operation failed.

### Mistake 3 — Wrong exception type

Catch the exception that the operation actually raises.

### Mistake 4 — Incorrect handler order

Specific exceptions should generally precede broader parent exceptions.

### Mistake 5 — Losing context

When translating an exception, use:

```python
raise NewError(...) from error
```

when preserving the original cause is useful.

### Mistake 6 — Using `assert` for user validation

Use explicit validation and `raise` instead.

````python
# Better than `except: pass`
try:
    value = int("abc")
except ValueError:
    value = 0

print("Fallback value:", value)
````

# 🚀 Project 01 — Robust Calculator

## Business Problem

Build a calculator that safely handles invalid numeric input, division by zero, and unsupported operations.

## Workflow

```text
INPUT
 ↓
VALIDATE
 ↓
PROCESS
 ↓
raise on invalid state
 ↓
APPLICATION HANDLER
 ↓
SUCCESS / RECOVERY / REPORT
```

## Solution

````python
def calculate(a, b, operation):
    if operation == "add":
        return a + b
    if operation == "subtract":
        return a - b
    if operation == "multiply":
        return a * b
    if operation == "divide":
        if b == 0:
            raise ZeroDivisionError("Cannot divide by zero.")
        return a / b

    raise ValueError(f"Unsupported operation: {operation}")

tests = [
    (10, 5, "add"),
    (10, 0, "divide"),
    (10, 5, "unknown")
]

for a, b, operation in tests:
    try:
        print(operation, "=", calculate(a, b, operation))
    except (ValueError, ZeroDivisionError) as error:
        print(operation, "→ Error:", error)
````

# 🚀 Project 02 — Customer Data Validator

## Business Problem

Validate customer records and raise a custom exception when required fields or values are invalid.

## Workflow

```text
INPUT
 ↓
VALIDATE
 ↓
PROCESS
 ↓
raise on invalid state
 ↓
APPLICATION HANDLER
 ↓
SUCCESS / RECOVERY / REPORT
```

## Solution

````python
class InvalidCustomerError(Exception):
    pass

def validate_customer(customer):
    required = ("name", "age", "email")

    for field in required:
        if field not in customer:
            raise InvalidCustomerError(
                f"Missing required field: {field}"
            )

    if not isinstance(customer["age"], int):
        raise TypeError("age must be an integer")

    if not 18 <= customer["age"] <= 100:
        raise ValueError("age must be between 18 and 100")

    if "@" not in customer["email"]:
        raise ValueError("Invalid email")

    return True

customers = [
    {"name": "Kaif", "age": 22, "email": "kaif@example.com"},
    {"name": "Ali", "age": 15, "email": "ali@example.com"},
    {"name": "Sara", "age": 27}
]

for customer in customers:
    try:
        validate_customer(customer)
        print(customer["name"], "→ Valid")
    except (InvalidCustomerError, TypeError, ValueError) as error:
        print("Invalid customer:", error)
````

# 🚀 Project 03 — File Report Reader

## Business Problem

Read a text report safely and distinguish a missing file from other operating-system errors.

## Workflow

```text
INPUT
 ↓
VALIDATE
 ↓
PROCESS
 ↓
raise on invalid state
 ↓
APPLICATION HANDLER
 ↓
SUCCESS / RECOVERY / REPORT
```

## Solution

````python
from pathlib import Path

def read_report(filename):
    path = Path(filename)

    try:
        with path.open("r", encoding="utf-8") as file:
            return file.read()
    except FileNotFoundError as error:
        raise FileNotFoundError(
            f"Report not found: {filename}"
        ) from error
    except OSError as error:
        raise OSError(
            f"Could not read report: {filename}"
        ) from error

try:
    report = read_report("sales_report.txt")
except FileNotFoundError as error:
    print(error)
except OSError as error:
    print("System error:", error)
else:
    print("Report loaded:", len(report), "characters")
````

# 🚀 Project 04 — Banking Transaction System

## Business Problem

Create safe deposit and withdrawal operations using domain-specific exceptions.

## Workflow

```text
INPUT
 ↓
VALIDATE
 ↓
PROCESS
 ↓
raise on invalid state
 ↓
APPLICATION HANDLER
 ↓
SUCCESS / RECOVERY / REPORT
```

## Solution

````python
class InsufficientFundsError(Exception):
    pass

class InvalidTransactionError(Exception):
    pass

def deposit(balance, amount):
    if amount <= 0:
        raise InvalidTransactionError(
            "Deposit must be greater than zero."
        )
    return balance + amount

def withdraw(balance, amount):
    if amount <= 0:
        raise InvalidTransactionError(
            "Withdrawal must be greater than zero."
        )

    if amount > balance:
        raise InsufficientFundsError(
            "Insufficient account balance."
        )

    return balance - amount

balance = 10000

try:
    balance = deposit(balance, 2500)
    balance = withdraw(balance, 3000)
    print("Final Balance:", balance)
except (InvalidTransactionError, InsufficientFundsError) as error:
    print("Transaction failed:", error)
````

# 🚀 Project 05 — Inventory Processing

## Business Problem

Process inventory records while handling missing keys, invalid values, and unavailable products.

## Workflow

```text
INPUT
 ↓
VALIDATE
 ↓
PROCESS
 ↓
raise on invalid state
 ↓
APPLICATION HANDLER
 ↓
SUCCESS / RECOVERY / REPORT
```

## Solution

````python
class InventoryError(Exception):
    pass

def calculate_item_value(item):
    try:
        name = item["name"]
        price = item["price"]
        stock = item["stock"]
    except KeyError as error:
        raise InventoryError(
            f"Missing inventory field: {error.args[0]}"
        ) from error

    if not isinstance(price, (int, float)):
        raise TypeError("price must be numeric")

    if not isinstance(stock, int):
        raise TypeError("stock must be an integer")

    if price < 0 or stock < 0:
        raise ValueError("price and stock cannot be negative")

    return name, price * stock

inventory = [
    {"name": "Laptop", "price": 65000, "stock": 5},
    {"name": "Mouse", "price": 1200, "stock": 20},
    {"name": "Broken Record", "price": -100, "stock": 2},
    {"name": "Missing Stock", "price": 1000}
]

for item in inventory:
    try:
        print(calculate_item_value(item))
    except (InventoryError, TypeError, ValueError) as error:
        print("Inventory error:", error)
````

# 🚀 Project 06 — Exam Result Processor

## Business Problem

Safely convert marks, validate ranges, and calculate grades for multiple students.

## Workflow

```text
INPUT
 ↓
VALIDATE
 ↓
PROCESS
 ↓
raise on invalid state
 ↓
APPLICATION HANDLER
 ↓
SUCCESS / RECOVERY / REPORT
```

## Solution

````python
def grade_student(name, raw_mark):
    try:
        mark = float(raw_mark)
    except (TypeError, ValueError) as error:
        raise ValueError(
            f"{name}: mark must be numeric."
        ) from error

    if not 0 <= mark <= 100:
        raise ValueError(
            f"{name}: mark must be between 0 and 100."
        )

    grade = (
        "A" if mark >= 80
        else "B" if mark >= 60
        else "C" if mark >= 40
        else "F"
    )

    return name, mark, grade

students = [
    ("Kaif", "88"),
    ("Sara", "95"),
    ("Ali", "unknown"),
    ("Omar", "105")
]

for name, mark in students:
    try:
        print(grade_student(name, mark))
    except ValueError as error:
        print("Result error:", error)
````

# 🚀 Project 07 — Data Import Validation

## Business Problem

Validate rows from a list of dictionaries and separate valid records from rejected records.

## Workflow

```text
INPUT
 ↓
VALIDATE
 ↓
PROCESS
 ↓
raise on invalid state
 ↓
APPLICATION HANDLER
 ↓
SUCCESS / RECOVERY / REPORT
```

## Solution

````python
def validate_row(row):
    if "id" not in row:
        raise KeyError("id")

    if "amount" not in row:
        raise KeyError("amount")

    if not isinstance(row["id"], int):
        raise TypeError("id must be an integer")

    if not isinstance(row["amount"], (int, float)):
        raise TypeError("amount must be numeric")

    if row["amount"] < 0:
        raise ValueError("amount cannot be negative")

    return True

rows = [
    {"id": 101, "amount": 500},
    {"id": 102, "amount": 750},
    {"id": "103", "amount": 900},
    {"id": 104, "amount": -20},
    {"amount": 100}
]

valid = []
rejected = []

for row in rows:
    try:
        validate_row(row)
        valid.append(row)
    except (KeyError, TypeError, ValueError) as error:
        rejected.append({
            "row": row,
            "error": str(error)
        })

print("Valid:", valid)
print("Rejected:", rejected)
````

# 🚀 Project 08 — API-Style Service Layer

## Business Problem

Simulate a service function that translates low-level failures into a domain-specific exception.

## Workflow

```text
INPUT
 ↓
VALIDATE
 ↓
PROCESS
 ↓
raise on invalid state
 ↓
APPLICATION HANDLER
 ↓
SUCCESS / RECOVERY / REPORT
```

## Solution

````python
class CustomerServiceError(Exception):
    pass

def fetch_customer(data, customer_id):
    try:
        return data[customer_id]
    except KeyError as error:
        raise CustomerServiceError(
            f"Customer {customer_id} was not found."
        ) from error

customers = {
    101: {"name": "Kaif", "plan": "Premium"},
    102: {"name": "Sara", "plan": "Standard"}
}

for customer_id in (101, 999):
    try:
        customer = fetch_customer(customers, customer_id)
    except CustomerServiceError as error:
        print("Service error:", error)
    else:
        print("Customer:", customer)
````


## Solution

````python
def parse_expense(category, raw_amount):
    try:
        amount = float(raw_amount)
    except (TypeError, ValueError) as error:
        raise ValueError(
            f"Invalid amount for {category}: {raw_amount}"
        ) from error

    if amount < 0:
        raise ValueError(
            f"Expense cannot be negative: {amount}"
        )

    return {"category": category, "amount": amount}

raw_expenses = [
    ("Food", "250"),
    ("Transport", "450.50"),
    ("Shopping", "invalid"),
    ("Bills", "-100")
]

expenses = []

for category, amount in raw_expenses:
    try:
        expenses.append(parse_expense(category, amount))
    except ValueError as error:
        print("Rejected:", error)

print("Valid Expenses:", expenses)
print(
    "Total:",
    sum(item["amount"] for item in expenses)
)
````

# 🚀 Project 10 — Production-Style Validation Engine

## Business Problem

Combine custom exceptions, specific handlers, exception chaining, and a clean application boundary.

## Workflow

```text
INPUT
 ↓
VALIDATE
 ↓
PROCESS
 ↓
raise on invalid state
 ↓
APPLICATION HANDLER
 ↓
SUCCESS / RECOVERY / REPORT
```

## Solution

````python
class ValidationError(Exception):
    pass

class MissingFieldError(ValidationError):
    pass

class InvalidFieldError(ValidationError):
    pass

def validate_user(user):
    if "name" not in user:
        raise MissingFieldError("name is required")

    if "age" not in user:
        raise MissingFieldError("age is required")

    try:
        age = int(user["age"])
    except (TypeError, ValueError) as error:
        raise InvalidFieldError("age must be numeric") from error

    if not 18 <= age <= 100:
        raise InvalidFieldError("age must be between 18 and 100")

    return {
        "name": user["name"].strip(),
        "age": age
    }

users = [
    {"name": "Kaif", "age": "22"},
    {"name": "Sara", "age": "27"},
    {"name": "Ali", "age": "unknown"},
    {"name": "Omar", "age": 15},
    {"age": 30}
]

valid_users = []
errors = []

for user in users:
    try:
        valid_users.append(validate_user(user))
    except MissingFieldError as error:
        errors.append({"type": "missing", "message": str(error)})
    except InvalidFieldError as error:
        errors.append({"type": "invalid", "message": str(error)})

print("Valid Users:", valid_users)
print("Errors:", errors)
````

# 📚 Exception Handling Quick Reference

| Keyword / Exception | Purpose |
|---|---|
| `try` | Contains code that may raise an exception |
| `except` | Handles a matching exception |
| `else` | Runs when `try` succeeds |
| `finally` | Runs during finalization regardless of normal success/failure |
| `raise` | Explicitly raises an exception |
| `ValueError` | Invalid value |
| `TypeError` | Inappropriate type/operation |
| `NameError` | Name not found |
| `IndexError` | Sequence index out of range |
| `KeyError` | Dictionary key not found |
| `FileNotFoundError` | Requested file/directory not found |
| `ZeroDivisionError` | Division/modulo by zero |
| `ImportError` | Import operation failed |
| `ModuleNotFoundError` | Requested module not found |
| `Exception` | Common base for most application exceptions |
| `BaseException` | Root of Python's exception hierarchy |
| Custom exception | Domain-specific failure |

### Core Pattern

```python
try:
    risky_operation()
except SpecificError as error:
    handle_error(error)
else:
    handle_success()
finally:
    cleanup()
```

### Raise Pattern

```python
if invalid:
    raise ValueError("Meaningful message")
```

### Custom Exception

```python
class BusinessRuleError(Exception):
    pass
```
