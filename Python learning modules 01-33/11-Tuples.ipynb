# 1. What Is a Tuple?

A Python `tuple` is an ordered collection of objects.

The key difference from a list is that a tuple is **immutable**: once the
tuple has been created, its elements cannot be reassigned or removed from
the tuple itself.

### Basic Example

```python
coordinates = (10, 20)
```

Tuples are useful when the collection represents a fixed group of values.

Examples:

- Coordinates
- RGB values
- Database records
- Configuration values
- Fixed categories
- Function return values

````python
coordinates = (10, 20)

print(coordinates)
print(type(coordinates))
````

# 2. Creating Tuples

Tuples are commonly written with parentheses:

```python
numbers = (10, 20, 30)
```

Python also supports tuple packing without parentheses:

```python
numbers = 10, 20, 30
```

Both create tuples.

### Empty Tuple

```python
empty = ()
```

### Important: Single-Item Tuple

A comma is required:

```python
single = (10,)
```

This is **not** a tuple:

```python
not_tuple = (10)
```

That expression is simply the integer `10`.

````python
numbers = (10, 20, 30)

packed = 10, 20, 30

empty = ()

single = (10,)

not_tuple = (10)

print(numbers)
print(type(numbers))

print(packed)
print(type(packed))

print(empty)
print(single)
print(type(single))

print(not_tuple)
print(type(not_tuple))
````

# 3. Tuple Indexing

Tuple indexing works like list indexing.

Indexes start at `0`.

```text
Value:  "A"    "B"    "C"    "D"
Index:   0      1      2      3
```

Example:

```python
names = ("Kaif", "Ali", "Sara", "Omar")

print(names[0])
print(names[2])
```

````python
names = ("Kaif", "Ali", "Sara", "Omar")

print(names[0])
print(names[2])
print(names[3])
````

# 4. Negative Indexing

Negative indexes count from the end.

```text
Value:  "A"    "B"    "C"    "D"
Index:   0      1      2      3
        -4     -3     -2     -1
```

Therefore:

```python
names[-1]
```

returns the last element.

````python
names = ("Kaif", "Ali", "Sara", "Omar")

print(names[-1])
print(names[-2])
print(names[-4])
````

# 5. Tuple Slicing

Tuples support slicing.

### Syntax

```python
tuple[start:stop:step]
```

The `stop` index is excluded.

Examples:

```python
numbers[1:4]
numbers[:3]
numbers[2:]
numbers[::2]
numbers[::-1]
```

````python
numbers = (10, 20, 30, 40, 50)

print(numbers[1:4])
print(numbers[:3])
print(numbers[2:])
print(numbers[::2])
print(numbers[::-1])
````

# 6. Tuple Immutability

This is the most important tuple concept.

After creation, you cannot change an element:

```python
numbers = (10, 20, 30)

numbers[0] = 100
```

This raises:

```text
TypeError
```

because tuple item assignment is not supported.

### What You Can Do

You can create a **new tuple** from existing values.

```python
numbers = (10, 20, 30)

numbers = (100,) + numbers[1:]
```

````python
numbers = (10, 20, 30)

updated_numbers = (100,) + numbers[1:]

print("Original:", numbers)
print("New tuple:", updated_numbers)
````

# 7. Why Immutability Matters

Immutability is useful when values should remain fixed.

For example:

```python
rgb = (255, 255, 255)
```

A fixed RGB configuration can be represented naturally by a tuple.

### Important Detail

A tuple itself cannot be modified, but it can contain a mutable object.

For example, a tuple may contain a list:

```python
data = ("Python", [10, 20])
```

The tuple cannot replace its second element, but the nested list is still a
mutable object and can be changed.

So immutability applies to the tuple's references/elements, not to the
deep contents of arbitrary objects stored inside it.

````python
data = ("Python", [10, 20])

data[1].append(30)

print(data)
````

# 8. Tuple Unpacking

Tuple unpacking assigns tuple elements to multiple variables.

```python
person = ("Kaif", 25)

name, age = person
```

Now:

```text
name → "Kaif"
age  → 25
```

The number of variables must normally match the number of unpacked
elements.

````python
person = ("Kaif", 25)

name, age = person

print("Name:", name)
print("Age:", age)
````

# 9. Multiple Assignment

Python's multiple assignment uses the same unpacking idea.

```python
x, y = 10, 20
```

This is convenient for assigning several values at once.

````python
x, y = 10, 20

print("x:", x)
print("y:", y)
````

# 10. Swapping Variables

Tuple unpacking makes swapping values simple.

```python
a = 10
b = 20

a, b = b, a
```

No temporary variable is required.

````python
a = 10
b = 20

print("Before:", a, b)

a, b = b, a

print("After:", a, b)
````

# 11. Extended Unpacking

The `*` operator can capture multiple values during unpacking.

```python
numbers = (1, 2, 3, 4, 5)

first, *middle, last = numbers
```

Result:

```text
first  → 1
middle → [2, 3, 4]
last   → 5
```

Notice that the starred target receives a **list**.

````python
numbers = (1, 2, 3, 4, 5)

first, *middle, last = numbers

print("First:", first)
print("Middle:", middle)
print("Last:", last)
````

# 12. Nested Tuples

A tuple can contain other tuples.

Example:

```python
students = (
    ("Kaif", 85),
    ("Ali", 78),
    ("Sara", 92)
)
```

Access nested data with multiple indexes:

```python
students[0][0]
students[1][1]
```

````python
students = (
    ("Kaif", 85),
    ("Ali", 78),
    ("Sara", 92)
)

print(students[0])
print(students[0][0])
print(students[1][1])
````

# 13. Iterating Through Tuples

Use a `for` loop to process tuple elements.

```python
skills = ("Python", "SQL", "Excel")

for skill in skills:
    print(skill)
```

````python
skills = ("Python", "SQL", "Excel")

for skill in skills:
    print(skill)
````

# 14. Membership Testing

Use `in` and `not in` to search for a value.

```python
skills = ("Python", "SQL", "Excel")

"Python" in skills
"Java" not in skills
```

````python
skills = ("Python", "SQL", "Excel")

print("Python" in skills)
print("Java" in skills)
print("Java" not in skills)
````

# 15. `count()`

`count()` returns how many times a value occurs.

### Syntax

```python
tuple.count(value)
```

Example:

````python
numbers = (10, 20, 10, 30, 10)

print(numbers.count(10))
print(numbers.count(20))
````

# 16. `index()`

`index()` returns the position of the first matching value.

### Syntax

```python
tuple.index(value)
```

Example:

````python
names = ("Ali", "Kaif", "Sara", "Kaif")

print(names.index("Kaif"))
````

# 17. Safe Searching With `index()`

If the value does not exist, `index()` raises `ValueError`.

When absence is possible, check membership first.

```python
if "Python" in skills:
    position = skills.index("Python")
```

This makes the lookup safe and explicit.

````python
skills = ("Python", "SQL", "Excel")

search = "Python"

if search in skills:
    print("Position:", skills.index(search))
else:
    print("Not found")
````

# 18. Length of a Tuple

Use `len()` to count elements.

```python
subjects = ("Python", "SQL", "Statistics")

print(len(subjects))
```

````python
subjects = ("Python", "SQL", "Statistics")

print("Number of subjects:", len(subjects))
````

# 19. Minimum, Maximum, and Sum

For compatible numeric tuples, built-in functions can be used:

```python
numbers = (10, 20, 30, 40)

min(numbers)
max(numbers)
sum(numbers)
```

These functions do not modify the tuple.

````python
numbers = (10, 20, 30, 40)

print("Minimum:", min(numbers))
print("Maximum:", max(numbers))
print("Total:", sum(numbers))
````

# 20. Sorting a Tuple

A tuple cannot be sorted in place because it is immutable.

Use `sorted()` to create a new **list**:

```python
numbers = (30, 10, 20)

sorted_numbers = sorted(numbers)
```

Result:

```text
[10, 20, 30]
```

If you specifically need a sorted tuple, convert the result:

```python
sorted_tuple = tuple(sorted(numbers))
```

````python
numbers = (30, 10, 20)

sorted_list = sorted(numbers)
sorted_tuple = tuple(sorted(numbers))

print("Original tuple:", numbers)
print("Sorted list:", sorted_list)
print("Sorted tuple:", sorted_tuple)
````

Tuples can be combined with `+`.

```python
first = (1, 2)
second = (3, 4)

combined = first + second
```

A new tuple is created.

````python
first = (1, 2)
second = (3, 4)

combined = first + second

print(combined)
````

Result:

```text
('Python', 'Python', 'Python')
```

````python
values = ("Python",) * 3

print(values)
````

# 23. Converting List → Tuple

Use `tuple()` to create a tuple from an iterable.

```python
numbers_list = [10, 20, 30]

numbers_tuple = tuple(numbers_list)
```

````python
numbers_list = [10, 20, 30]

numbers_tuple = tuple(numbers_list)

print(numbers_list)
print(numbers_tuple)
print(type(numbers_tuple))
````

# 24. Converting Tuple → List

Use `list()` when you need a mutable list.

```python
numbers_tuple = (10, 20, 30)

numbers_list = list(numbers_tuple)
```

You can modify the new list without modifying the original tuple.

````python
numbers_tuple = (10, 20, 30)

numbers_list = list(numbers_tuple)

numbers_list.append(40)

print("Tuple:", numbers_tuple)
print("List:", numbers_list)
````

# 25. Tuple vs List

| Feature | List | Tuple |
|---|---|---|
| Syntax | `[ ]` | `( )` |
| Mutable | Yes | No |
| Ordered | Yes | Yes |
| Duplicate values | Yes | Yes |
| Indexing | Yes | Yes |
| Slicing | Yes | Yes |
| `append()` | Yes | No |
| `remove()` | Yes | No |
| `count()` | Yes | Yes |
| `index()` | Yes | Yes |
| Typical use | Changeable collection | Fixed collection |

### Simple Rule

Use a **list** when the collection needs to change.

Use a **tuple** when the collection represents a fixed group of values.

# 26. When Should You Use a Tuple?

Good tuple use cases include:

### Coordinates

```python
point = (10, 20)
```

### RGB Values

```python
rgb = (255, 128, 0)
```

### Fixed Configuration

```python
window_size = (1920, 1080)
```

```python
def get_user():
```

Tuples are especially useful when the structure is conceptually fixed.

````python
def get_user():
    return "Kaif", 25

user = get_user()

print(user)
print(type(user))
````

# 27. Common Tuple Mistakes

## Mistake 1 — Forgetting the comma

❌

```python
value = (10)
```

This is an integer.

✅

```python
value = (10,)
```

This is a tuple.

## Mistake 2 — Trying to modify a tuple

❌

```python
numbers[0] = 100
```

Tuples do not support item assignment.

## Mistake 3 — Assuming `sorted()` returns a tuple

```python
sorted(numbers)
```

returns a list.

Use:

```python
tuple(sorted(numbers))
```

if a tuple is required.

## Mistake 4 — Calling `remove()` or `append()`

Tuples do not provide list mutation methods such as `append()`,
`remove()`, or `pop()`.

# 🚀 Project 01 — Student Marks Record

## Problem

Represent each student's fixed record as a tuple:

```text
(Name, Roll Number, Marks)
```

Store multiple records in a tuple and calculate:

- Number of students
- Highest marks
- Lowest marks
- Average marks
- Top student

This demonstrates nested tuples and unpacking.

````python
# Project 01 — Student Marks Record

students = (
    ("Kaif", 101, 85),
    ("Ali", 102, 78),
    ("Sara", 103, 92),
    ("Omar", 104, 88)
)

marks = tuple(student[2] for student in students)

highest = max(marks)
lowest = min(marks)
average = sum(marks) / len(marks)

top_student = max(students, key=lambda student: student[2])

print("Students:", len(students))
print(f"Highest Marks: {highest}")
print(f"Lowest Marks: {lowest}")
print(f"Average Marks: {average:.2f}")
print(f"Top Student: {top_student[0]}")
print(f"Top Marks: {top_student[2]}")
````

# 🚀 Project 02 — Coordinate Analyzer

## Problem

Represent points as tuples:

```python
(x, y)
```

Analyze a collection of points and display:

- Number of points
- X coordinates
- Y coordinates
- Minimum X
- Maximum X
- Minimum Y
- Maximum Y

````python
# Project 02 — Coordinate Analyzer

points = (
    (10, 20),
    (15, 5),
    (3, 25),
    (18, 12)
)

x_values = tuple(point[0] for point in points)
y_values = tuple(point[1] for point in points)

print("Number of Points:", len(points))
print("X Coordinates:", x_values)
print("Y Coordinates:", y_values)
print("Minimum X:", min(x_values))
print("Maximum X:", max(x_values))
print("Minimum Y:", min(y_values))
print("Maximum Y:", max(y_values))
````

# 🚀 Project 03 — Product Catalog

## Problem

Represent fixed product records as tuples:

```text
(Product ID, Product Name, Price)
```

Store multiple products and:

- Search by product name
- Find the most expensive product
- Sort products by price
- Display the catalog

This demonstrates tuple records and `sorted()`.

````python
# Project 03 — Product Catalog

products = (
    (101, "Laptop", 65000),
    (102, "Mouse", 1200),
    (103, "Keyboard", 2500),
    (104, "Monitor", 18000)
)

search_name = input("Enter product name: ").strip().lower()

found = False

for product_id, name, price in products:
    if name.lower() == search_name:
        print(f"Found: {name} | ₹{price:.2f}")
        found = True
        break

if not found:
    print("Product not found.")

most_expensive = max(products, key=lambda product: product[2])

print(
    f"Most Expensive: {most_expensive[1]} "
    f"| ₹{most_expensive[2]:.2f}"
)

sorted_products = sorted(products, key=lambda product: product[2])

print("\nProducts by Price:")

for product_id, name, price in sorted_products:
    print(f"{product_id} | {name} | ₹{price:.2f}")
````

# 🚀 Project 04 — Employee Records

## Problem

Represent employee records as:

```text
(Employee ID, Name, Department, Salary)
```

Analyze:

- Number of employees
- Highest salary
- Lowest salary
- Average salary
- Employees in a department

This demonstrates nested tuples, unpacking, `max()`, `min()`, and `sum()`.

````python
# Project 04 — Employee Records

employees = (
    (1, "Kaif", "Data", 60000),
    (2, "Ali", "Engineering", 75000),
    (3, "Sara", "Data", 68000),
    (4, "Omar", "HR", 52000)
)

salaries = tuple(employee[3] for employee in employees)

print("Employee Count:", len(employees))
print(f"Highest Salary: ₹{max(salaries):.2f}")
print(f"Lowest Salary: ₹{min(salaries):.2f}")
print(f"Average Salary: ₹{sum(salaries) / len(salaries):.2f}")

department = input("Enter department: ").strip().lower()

print("\nEmployees in Department:")

found = False

for employee_id, name, dept, salary in employees:
    if dept.lower() == department:
        print(f"{employee_id} | {name} | ₹{salary:.2f}")
        found = True

if not found:
    print("No employees found.")
````

# 🚀 Project 05 — Sales Transaction Analyzer

## Problem

Represent each transaction as:

```text
(Transaction ID, Product, Amount)
```

Analyze:

- Total transactions
- Total sales
- Average transaction
- Highest transaction
- Lowest transaction
- Transactions above average
- Sorted transactions

This is a practical business-data example using tuples.

````python
# Project 05 — Sales Transaction Analyzer

transactions = (
    (1001, "Laptop", 65000),
    (1002, "Mouse", 1200),
    (1003, "Monitor", 18000),
    (1004, "Keyboard", 2500),
    (1005, "Laptop Bag", 3500)
)

amounts = tuple(transaction[2] for transaction in transactions)

total_sales = sum(amounts)
average_sales = total_sales / len(amounts)
highest_transaction = max(transactions, key=lambda transaction: transaction[2])
lowest_transaction = min(transactions, key=lambda transaction: transaction[2])

print("=" * 60)
print("          SALES TRANSACTION ANALYZER")
print("=" * 60)

print(f"Transaction Count : {len(transactions)}")
print(f"Total Sales       : ₹{total_sales:.2f}")
print(f"Average Sale      : ₹{average_sales:.2f}")
print(
    f"Highest Sale      : {highest_transaction[1]} "
    f"₹{highest_transaction[2]:.2f}"
)
print(
    f"Lowest Sale       : {lowest_transaction[1]} "
    f"₹{lowest_transaction[2]:.2f}"
)

print("\nTransactions Above Average:")

for transaction_id, product, amount in transactions:
    if amount > average_sales:
        print(f"{transaction_id} | {product} | ₹{amount:.2f}")

print("\nSorted Transactions:")

sorted_transactions = sorted(
    transactions,
    key=lambda transaction: transaction[2],
    reverse=True
)

for transaction_id, product, amount in sorted_transactions:
    print(f"{transaction_id} | {product} | ₹{amount:.2f}")
````

# 📚 Tuple Methods — Quick Reference

Python tuples have two main value-oriented methods:

| Method | Purpose |
|---|---|
| `count(value)` | Count occurrences |
| `index(value)` | Find first matching index |

### Useful Built-ins

| Function | Purpose |
|---|---|
| `len()` | Number of elements |
| `min()` | Minimum value |
| `max()` | Maximum value |
| `sum()` | Sum numeric values |
| `sorted()` | Return a new sorted list |
| `tuple()` | Create a tuple from an iterable |
| `list()` | Convert tuple to list |

### Core Pattern

```python
data = (10, 20, 10, 30)

print(data.count(10))

if 20 in data:
    print(data.index(20))
```

```text
            ↓
        count()

Need first index?
            ↓
        index()

Need to unpack?
            ↓
    a, b = tuple

Need multiple middle values?
            ↓
 first, *middle, last = tuple

Need sorting?
            ↓
    sorted(tuple)
            ↓
       returns LIST

            ↓
 tuple(sorted(tuple))
```
