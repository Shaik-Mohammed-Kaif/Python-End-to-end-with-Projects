# 1. What Is Regex?

Python's built-in `re` module provides regular-expression operations. A regex describes a text pattern.

Typical uses:
- pattern matching
- extraction
- validation
- replacement
- text cleaning

## Example

````python
import re

text = "Customer ID: 1024"
match = re.search(r"\d+", text)

if match:
    print("Found:", match.group())
````

# 2. Raw Strings

Regex patterns commonly use raw strings such as `r"\d+"`. This keeps backslashes readable.

## Example

````python
import re

pattern = r"\d+"
print(re.search(pattern, "Order 125").group())
````

# 3. Character Classes

Character classes match one character from a set.

```text
[abc]   → a, b, or c
[a-z]   → lowercase ASCII letters
[A-Z]   → uppercase ASCII letters
[0-9]   → digits
[^0-9]  → non-digit
```

## Example

````python
import re

text = "Python 3"
print(re.findall(r"[A-Z]", text))
print(re.findall(r"[a-z]", text))
print(re.findall(r"[0-9]", text))
````

# 4. Shorthand Classes

Important shorthand classes:

| Pattern | Meaning |
|---|---|
| `\d` | Unicode decimal digit |
| `\D` | Not a Unicode decimal digit |
| `\w` | Unicode word character |
| `\W` | Not a Unicode word character |
| `\s` | Unicode whitespace |
| `\S` | Not Unicode whitespace |

## Example

````python
import re

text = "Python 123"
print(re.findall(r"\d+", text))
print(re.findall(r"\w+", text))
print(re.findall(r"\s+", text))
````

# 5. Dot `.`

`.` matches any character except a newline by default.

```text
c.t
```

can match `cat`, `cut`, and `c9t`.

## Example

````python
import re

for word in ["cat", "cut", "c9t", "cart"]:
    print(word, bool(re.fullmatch(r"c.t", word)))
````

# 6. Quantifiers

Quantifiers control repetition.

```text
*       0 or more
+       1 or more
?       0 or 1
{m}     exactly m
{m,}    at least m
{m,n}   m through n
```

## Example

````python
import re

text = "7 42 2026 123456"
print(re.findall(r"\d+", text))
print(re.findall(r"\d{4}", text))
print(re.findall(r"\d{2,4}", text))
````

# 7. Greedy and Non-Greedy

Most repetition is greedy. Adding `?` after a quantifier makes it non-greedy.

```text
.*   greedy
.*?  non-greedy
```

## Example

````python
import re

text = "<b>Python</b><b>Regex</b>"
print(re.findall(r"<b>.*</b>", text))
print(re.findall(r"<b>.*?</b>", text))
````

# 8. Anchors

Important anchors:

```text
^   start position
$   end position
\A  start of string
\Z  end of string
```

For whole-input validation, `fullmatch()` is often clearer.

## Example

````python
import re

print(bool(re.search(r"^Python", "Python Regex")))
print(bool(re.search(r"Python$", "Learn Python")))
print(bool(re.fullmatch(r"\d{4}", "2026")))
````

# 9. Capturing Groups

Parentheses create capturing groups. Use `group(1)`, `group(2)`, and so on.

## Example

````python
import re

match = re.search(r"Order-(\d+)", "Order-1024")
print("Full:", match.group())
print("ID:", match.group(1))
````

# 10. Named Groups

Named groups use `(?P<name>pattern)` and can be accessed by name or through `groupdict()`.

## Example

````python
import re

match = re.search(
    r"(?P<type>[A-Za-z]+)-(?P<id>\d+)",
    "Customer-1024"
)

print(match.group("type"))
print(match.group("id"))
print(match.groupdict())
````

# 11. Alternation `|`

`|` means OR.

```python
r"Python|SQL"
```

matches either alternative.

## Example

````python
import re

print(re.findall(r"Python|SQL", "Python SQL Excel"))
````

# 12. Escaping Special Characters

Special regex characters can be escaped when you need their literal meaning.

Examples:

```text
\.  literal dot
\+  literal plus
\?  literal question mark
```

## Example

````python
import re

text = "Price: $10.50"
print(re.findall(r"\.", text))
print(re.findall(r"\$\d+\.\d+", text))
````

# 13. `re.search()`

`search()` scans through the string and returns the first matching `Match` object, or `None`.

## Example

````python
import re

text = "Customer email: kaif@example.com"
match = re.search(r"[\w.-]+@[\w.-]+\.[A-Za-z]{2,}", text)

if match:
    print(match.group())
````

# 14. `re.match()`

`match()` attempts a match at the beginning of the string.

## Example

````python
import re

print(bool(re.match(r"Python", "Python is useful")))
print(bool(re.match(r"Python", "Learn Python")))
````

# 15. `re.fullmatch()`

`fullmatch()` requires the entire string to match. It is especially useful for strict validation.

## Example

````python
import re

print(bool(re.fullmatch(r"\d{6}", "560001")))
print(bool(re.fullmatch(r"\d{6}", "560001A")))
````

# 16. Search vs Match vs Fullmatch

```text
search()     → anywhere
match()      → beginning
fullmatch()  → entire input
```

## Example

````python
import re

text = "Order-1024"
print("search:", bool(re.search(r"\d+", text)))
print("match:", bool(re.match(r"Order", text)))
print("fullmatch:", bool(re.fullmatch(r"Order-\d+", text)))
````

# 17. `re.findall()`

`findall()` returns all non-overlapping matches. When capturing groups are used, its result shape depends on those groups.

## Example

````python
import re

print(re.findall(r"\d+", "Order 101, Order 202, Order 303"))
print(re.findall(r"([A-Z])(\d)", "A1 B2 C3"))
````

# 18. `re.finditer()`

`finditer()` returns an iterator of `Match` objects, useful when you need values and positions.

## Example

````python
import re

for match in re.finditer(r"\d+", "Order 101 and Order 202"):
    print(match.group(), match.start(), match.end())
````

# 19. `re.sub()`

`sub()` replaces matching text. A count can limit the number of replacements.

## Example

````python
import re

text = "Python   is   powerful"
print(re.sub(r"\s+", " ", text).strip())
print(re.sub(r"\d", "*", "Account 12345"))
````

# 20. `re.split()`

`split()` divides text wherever the regex pattern matches.

## Example

````python
import re

text = "Python,SQL;Excel|Power BI"
print(re.split(r"[,;|]", text))
````

# 21. `re.compile()`

`compile()` creates a reusable regex pattern. This is useful when the same pattern is used repeatedly.

## Example

````python
import re

pattern = re.compile(r"\d+")
print(pattern.findall("A12 B34 C56"))
print(pattern.search("ID 999").group())
````

# 22. Regex Flags

Common flags include `re.IGNORECASE`, `re.MULTILINE`, `re.DOTALL`, and `re.VERBOSE`.

## Example

````python
import re

print(re.findall(r"python", "Python PYTHON python", flags=re.IGNORECASE))
````

# 23. Validation

For strict validation, describe the complete accepted input and use `fullmatch()`.

## Example

````python
import re

for value in ["560001", "110001", "12345", "560001A"]:
    valid = re.fullmatch(r"\d{6}", value) is not None
    print(value, "→", "Valid" if valid else "Invalid")
````

# 24. Email Validation

A simple project-level regex can check email-like structure. It does not prove that the address exists or can receive mail.

## Example

````python
import re

pattern = re.compile(r"^[\w.-]+@[\w.-]+\.[A-Za-z]{2,}$")

for email in ["kaif@example.com", "analyst@company.co", "invalid-email"]:
    print(email, "→", bool(pattern.fullmatch(email)))
````

# 25. Phone Validation

Phone formats vary by country and application. This example validates a simple 10-digit numeric format.

## Example

````python
import re

pattern = re.compile(r"\d{10}")

for phone in ["9876543210", "1234567890", "98765"]:
    print(phone, "→", bool(pattern.fullmatch(phone)))
````

# 26. Password Validation

Regex can check structural rules such as length, uppercase, lowercase, digit, and special character. Never store passwords as plain text.

## Example

````python
import re

pattern = re.compile(
    r"(?=.*[A-Z])"
    r"(?=.*[a-z])"
    r"(?=.*\d)"
    r"(?=.*[^A-Za-z0-9])"
    r".{8,}"
)

for password in ["Python@123", "password", "Kaif123"]:
    print("Valid" if pattern.fullmatch(password) else "Invalid")
````

# 27. URL Extraction

A simple regex can extract HTTP/HTTPS URL-like strings. Full URL parsing is more complicated.

## Example

````python
import re

text = (
    "Visit https://example.com and "
    "https://www.python.org for documentation."
)

urls = re.findall(r"https?://[^\s]+", text)

for url in urls:
    print(url)
````

# 28. Number Extraction

Regex can extract integer and decimal-looking numbers from unstructured text.

## Example

````python
import re

text = "Orders: 125, 250, 375. Revenue: 65000.50"
numbers = re.findall(r"\d+(?:\.\d+)?", text)

print(numbers)
print([float(value) for value in numbers])
````

# 29. Text Cleaning

Regex can normalize repeated whitespace. For simple cases, ordinary string methods may be clearer.

## Example

````python
import re

text = "   Python    is   powerful\n\nData Science   "
cleaned = re.sub(r"\s+", " ", text).strip()

print(cleaned)
````

# 30. Replacement With Groups

Captured groups can be reused in a replacement string.

## Example

````python
import re

text = "Kaif, Mohammed"
formatted = re.sub(
    r"([A-Za-z]+), ([A-Za-z]+)",
    r"\2 \1",
    text
)

print(formatted)
````

# 31. Positive Lookahead

A positive lookahead checks for another pattern without consuming it. It is useful for advanced validation rules.

## Example

````python
import re

pattern = re.compile(r"(?=.*\d).{6,}")

for value in ["Python", "Python1"]:
    print(value, "→", bool(pattern.fullmatch(value)))
````

# 32. Non-Capturing Groups

`(?:pattern)` groups alternatives without creating a capturing group.

## Example

````python
import re

print(re.findall(r"(?:Python|SQL)", "Python SQL Python"))
````

# 33. Common Mistakes

| Mistake | Better approach |
|---|---|
| Using `match()` for an anywhere search | use `search()` |
| Using `search()` for strict validation | use `fullmatch()` |
| Forgetting to escape special characters | use the correct escaped pattern |
| Overusing `.*` | use a narrower pattern |
| Assuming email regex is universal | validate only the required structure |
| Using regex for simple `replace()`/`split()` tasks | prefer normal string methods when clearer |
| Not testing edge cases | test empty, invalid, boundary, and unexpected input |
| Using regex as a security system | combine validation with proper application security |

# 34. Best Practices

- Prefer raw strings for regex patterns.
- Use `fullmatch()` for whole-input validation.
- Compile patterns that are reused.
- Use named groups for complex extraction.
- Keep patterns readable.
- Test edge cases.
- Avoid unnecessarily broad `.*`.
- Use `re.VERBOSE` for genuinely complex patterns.
- Do not claim that a regex proves external data is trustworthy.
- Use normal string methods when they are simpler.

# 🚀 Project 1 — 01 — Email Validator

## Business Problem

Validate the basic structure of email-like addresses.

## Workflow

```text
RAW INPUT
   ↓
REGEX PATTERN
   ↓
SEARCH / VALIDATE / EXTRACT / CLEAN
   ↓
RESULT
```

````python
import re

pattern = re.compile(r"^[\w.-]+@[\w.-]+\.[A-Za-z]{2,}$")
emails = [
    "kaif@example.com",
    "analyst@company.co",
    "invalid-email",
    "user@domain",
]

for email in emails:
    result = "Valid" if pattern.fullmatch(email) else "Invalid"
    print(f"{email:25} → {result}")
````

Check structural password requirements: length, upper/lower case, digit, and special character.

# 📚 Regex Quick Reference

| Pattern/API | Purpose |
|---|---|
| `.` | any character except newline by default |
| `^` | start position |
| `$` | end position |
| `\d` | digit |
| `\D` | non-digit |
| `\w` | word character |
| `\W` | non-word character |
| `\s` | whitespace |
| `\S` | non-whitespace |
| `[abc]` | one of a/b/c |
| `[a-z]` | lowercase ASCII range |
| `[^0-9]` | not a digit |
| `*` | zero or more |
| `+` | one or more |
| `?` | zero or one |
| `{m}` | exactly m |
| `{m,n}` | m through n |
| `(pattern)` | capturing group |
| `(?:pattern)` | non-capturing group |
| `(?P<name>pattern)` | named group |
| `A|B` | A or B |
| `(?=pattern)` | positive lookahead |

## Core API

```python
re.search()
re.match()
re.fullmatch()
re.findall()
re.finditer()
re.sub()
re.split()
re.compile()
```

## Quick Decision

```text
ANYWHERE       → search()
START          → match()
ENTIRE INPUT   → fullmatch()
ALL MATCHES    → findall() / finditer()
REPLACE        → sub()
SPLIT          → split()
REUSE PATTERN  → compile()
```
