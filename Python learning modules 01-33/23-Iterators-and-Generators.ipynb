# 1. Iterable

An **iterable** is an object that can provide its elements one at a time.

Common iterables include:

```text
list
tuple
string
set
dictionary
range
file objects
```

An iterable can normally be passed to `iter()`.

## Example

````python
numbers = [10, 20, 30]

iterator = iter(numbers)

print(type(numbers).__name__)
print(type(iterator).__name__)
````

# 2. Iterator

An **iterator** produces values one at a time.

The iterator protocol is based on:

```python
__iter__()
__next__()
```

`next()` requests the next available value.

## Example

````python
numbers = [10, 20, 30]
iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(next(iterator))
````

# 3. `iter()`

`iter(iterable)` returns an iterator for an iterable.

## Example

````python
items = ["Python", "SQL", "Excel"]

iterator = iter(items)

print(next(iterator))
print(next(iterator))
print(next(iterator))
````

# 4. `next()`

`next(iterator)` requests the next value. When the iterator is exhausted, Python raises `StopIteration`.

## Example

````python
numbers = iter([100, 200])

print(next(numbers))
print(next(numbers))

try:
    print(next(numbers))
except StopIteration:
    print("Iterator is exhausted")
````

# 5. Iterable vs Iterator

```text
Iterable
→ can provide an iterator

Iterator
→ produces the next value

iter(iterable)
→ obtains iterator

next(iterator)
→ obtains next value
```

## Example

````python
items = ["A", "B", "C"]

print(iter(items) is items)

iterator = iter(items)

print(next(iterator))
print(next(iterator))
print(next(iterator))
````

# 6. `for` Loop and Iterators

A `for` loop uses the iterator protocol internally: obtain an iterator, request values, and stop when `StopIteration` occurs.

## Example

````python
numbers = [10, 20, 30]

for number in numbers:
    print(number)
````

# 7. Manual Iteration

Manual iteration makes the iterator protocol visible.

## Example

````python
numbers = [10, 20, 30]
iterator = iter(numbers)

while True:
    try:
        value = next(iterator)
        print(value)
    except StopIteration:
        break
````

# 8. `StopIteration`

`StopIteration` signals that an iterator has no more values. Normal `for` loops handle it automatically.

## Example

````python
iterator = iter(["A", "B"])

print(next(iterator))
print(next(iterator))

try:
    next(iterator)
except StopIteration:
    print("No more values")
````

# 9. Custom Iterator

A custom iterator normally implements `__iter__()` and `__next__()`.

`__iter__()` returns the iterator object, while `__next__()` returns the next
value or raises `StopIteration`.

## Example

````python
class CountUp:
    def __init__(self, limit):
        self.current = 1
        self.limit = limit

    def __iter__(self):
        return self

    def __next__(self):
        if self.current > self.limit:
            raise StopIteration

        value = self.current
        self.current += 1
        return value

counter = CountUp(5)

for value in counter:
    print(value)
````

# 10. Iterator State

Iterators maintain their current position. Calling `next()` continues from that position.

## Example

````python
items = ["A", "B", "C"]
iterator = iter(items)

print(next(iterator))
print(next(iterator))
print(next(iterator))
````

# 11. Iterator Exhaustion

Many iterators are one-pass objects. Once consumed, they generally cannot produce the old values again.

## Example

````python
items = [1, 2, 3]
iterator = iter(items)

print(list(iterator))
print(list(iterator))
````

# 12. Generator Functions

A generator function contains `yield`. Calling it returns a generator object; its body executes as values are requested.

## Example

````python
def numbers():
    yield 1
    yield 2
    yield 3

generator = numbers()

print(type(generator).__name__)
print(next(generator))
print(next(generator))
print(next(generator))
````

# 13. `yield`

`yield` produces a value and pauses the generator. The function resumes after the previous `yield` when another value is requested.

## Example

````python
def simple_generator():
    print("Step 1")
    yield "A"

    print("Step 2")
    yield "B"

generator = simple_generator()

print(next(generator))
print(next(generator))
````

# 14. Generator vs Normal Function

A normal function generally returns a result with `return`. A generator produces a sequence lazily using `yield`.

## Example

````python
def normal_function():
    return [1, 2, 3]

def generator_function():
    yield 1
    yield 2
    yield 3

print(normal_function())
print(list(generator_function()))
````

# 15. Generator with a Loop

Generators are useful when a sequence can naturally be produced one item at a time.

## Example

````python
def count_up(limit):
    for number in range(1, limit + 1):
        yield number

for number in count_up(5):
    print(number)
````

# 16. Generator Expressions

A generator expression uses parentheses:

```python
(x ** 2 for x in range(10))
```

It produces values lazily.

## Example

````python
squares = (number ** 2 for number in range(5))

print(type(squares).__name__)

for square in squares:
    print(square)
````

# 17. List Comprehension vs Generator Expression

List comprehensions are eager and create the result collection immediately. Generator expressions are lazy and produce values when requested.

## Example

````python
numbers = range(5)

squares_list = [x ** 2 for x in numbers]
squares_generator = (x ** 2 for x in numbers)

print(squares_list)
print(list(squares_generator))
````

# 18. Lazy Evaluation

Lazy evaluation means a value is produced only when needed. Generators are a common Python mechanism for lazy processing.

## Example

````python
def sales_stream():
    for sale in [1000, 2500, 1800, 4200]:
        print("Producing:", sale)
        yield sale

sales = sales_stream()

print("Generator created")
print(next(sales))
print(next(sales))
````

# 19. Memory Efficiency

Generators can reduce memory usage because they do not need to store the entire generated sequence at once.

They are especially useful for large or streaming workloads.

They are **not automatically faster** than lists; the benefit depends on the workload.

## Example

````python
def numbers(limit):
    for number in range(limit):
        yield number

generator = numbers(5)

print(next(generator))
print(next(generator))
print(next(generator))
````

# 20. Lazy Transformation

A generator can transform data one item at a time.

## Example

````python
def sales_with_tax(sales, tax_rate):
    for sale in sales:
        yield sale * (1 + tax_rate)

sales = [1000, 2000, 3000]

for value in sales_with_tax(sales, 0.18):
    print(round(value, 2))
````

# 21. Generator Pipeline

Generators can be chained into lazy pipelines.

```text
RAW DATA
   ↓
FILTER
   ↓
TRANSFORM
   ↓
OUTPUT
```

## Example

````python
def positive_sales(values):
    for value in values:
        if value > 0:
            yield value

def add_tax(values, rate):
    for value in values:
        yield value * (1 + rate)

sales = [1000, -500, 2000, 3000]

pipeline = add_tax(positive_sales(sales), 0.18)

for value in pipeline:
    print(round(value, 2))
````

# 22. Generator `return`

A generator can use `return` to finish. Its return value becomes the `value` of the resulting `StopIteration` exception.

## Example

````python
def process():
    yield "Step 1"
    yield "Step 2"
    return "Completed"

generator = process()

try:
    while True:
        print(next(generator))
except StopIteration as error:
    print("Return value:", error.value)
````

# 23. Generator Expressions with Built-ins

Generator expressions work well with `sum()`, `any()`, `all()`, `min()`, and `max()` without first creating an intermediate list.

## Example

````python
sales = [100, 200, 300, 400]

total = sum(value for value in sales)
has_large_sale = any(value > 350 for value in sales)
all_positive = all(value > 0 for value in sales)

print(total)
print(has_large_sale)
print(all_positive)
````

# 24. `yield from`

`yield from` delegates iteration to another iterable or generator.

## Example

````python
def first():
    yield "A"
    yield "B"

def second():
    yield "C"
    yield "D"

def combined():
    yield from first()
    yield from second()

print(list(combined()))
````

# 25. Generator vs Iterator

A generator is a convenient type of iterator.

```text
Generator → iterator
Iterator  → broader concept

Every generator is an iterator.
Not every iterator is a generator.
```

## Example

````python
def generate_values():
    yield 1
    yield 2

generator = generate_values()

print(iter(generator) is generator)
print(next(generator))
print(next(generator))
````

# 26. Generator vs List

| Feature | List | Generator |
|---|---|---|
| Evaluation | eager | lazy |
| Storage | all results | produces on demand |
| Reusable | normally yes | generally one pass |
| Indexing | yes | no |
| Large streams | less suitable | often suitable |

## Example

````python
values = range(5)

data_list = [value * 2 for value in values]
data_generator = (value * 2 for value in values)

print(data_list)
print(list(data_generator))
````

# 27. File Objects Are Iterable

File objects can be iterated line by line. This is useful for processing large files without loading the complete file into memory.

## Example

````python
from pathlib import Path

path = Path("sales_lines.txt")

path.write_text(
    "Laptop,65000\nMouse,1200\nMonitor,18000\n",
    encoding="utf-8"
)

with path.open("r", encoding="utf-8") as file:
    for line in file:
        print(line.strip())
````

# 28. Generator for File Processing

A generator can add filtering or transformation while preserving lazy file processing.

## Example

````python
from pathlib import Path

path = Path("numbers.txt")

path.write_text(
    "10\n25\n40\n55\n",
    encoding="utf-8"
)

def read_numbers(file_path):
    with Path(file_path).open("r", encoding="utf-8") as file:
        for line in file:
            value = int(line.strip())
            if value >= 25:
                yield value

for number in read_numbers(path):
    print(number)
````

# 29. `iter(callable, sentinel)`

`iter(callable, sentinel)` repeatedly calls a zero-argument callable until the returned value equals the sentinel.

## Example

````python
from io import StringIO

source = StringIO("Python\nSQL\nExcel\n")

for line in iter(source.readline, ""):
    print(line.strip())
````

# 30. Infinite Generators

Generators can produce values indefinitely. Consume them with a bounded strategy.

## Example

````python
def infinite_numbers():
    number = 1

    while True:
        yield number
        number += 1

generator = infinite_numbers()

for _ in range(5):
    print(next(generator))
````

# 31. Memory-Efficient Analytics

A generator can validate, filter, and transform records one at a time instead of creating multiple intermediate lists.

## Example

````python
def clean_sales(values):
    for value in values:
        if value is None:
            continue

        value = float(value)

        if value > 0:
            yield value

raw_sales = ["1000", None, "-500", "2500", "3000"]

total = sum(clean_sales(raw_sales))

print("Total valid sales:", total)
````

# 32. Common Mistakes

### Mistake 1 — Passing an iterable directly to `next()`

```python
next([1, 2, 3])
```

Use:

```python
next(iter([1, 2, 3]))
```

### Mistake 2 — Reusing an exhausted generator

A consumed generator normally cannot be restarted.

### Mistake 3 — Converting a huge generator to a list unnecessarily

```python
list(huge_generator)
```

This materializes all values and can remove the memory advantage.

### Mistake 4 — Creating an infinite generator without a stopping strategy

Always consume it with a bounded condition when appropriate.

### Mistake 5 — Assuming generators are always faster

Generators primarily provide lazy/streaming behavior and can reduce memory
usage. Performance depends on the workload.

# 33. Best Practices

- Use normal `for` loops when simple iteration is all you need.
- Use generators for large, streaming, or one-pass sequences.
- Use generator expressions for simple lazy transformations.
- Avoid unnecessary `list(generator)`.
- Keep generator functions focused.
- Use `yield from` for clean generator composition.
- Document whether a function returns a collection or a one-pass iterator.
- Be aware of iterator exhaustion.
- Bound infinite generators.
- Prefer readable code over unnecessary iterator complexity.

# 🚀 Project 5 — 05 — Custom Counter Iterator

## Business Problem

Create an iterator that produces a bounded sequence.

## Architecture

```text
INPUT
 ↓
ITERABLE / ITERATOR
 ↓
LAZY PROCESSING
 ↓
FILTER / TRANSFORM
 ↓
OUTPUT
```

## Implementation

````python
class Counter:
    def __init__(self, start, stop):
        self.current = start
        self.stop = stop

    def __iter__(self):
        return self

    def __next__(self):
        if self.current > self.stop:
            raise StopIteration

        value = self.current
        self.current += 1
        return value

for value in Counter(1, 5):
    print(value)
````

# 🚀 Project 6 — 06 — Batch Generator

## Business Problem

Yield records in fixed-size batches.

## Architecture

```text
INPUT
 ↓
ITERABLE / ITERATOR
 ↓
LAZY PROCESSING
 ↓
FILTER / TRANSFORM
 ↓
OUTPUT
```

## Implementation

````python
def batches(values, size):
    if size <= 0:
        raise ValueError("Batch size must be positive")

    batch = []

    for value in values:
        batch.append(value)

        if len(batch) == size:
            yield batch
            batch = []

    if batch:
        yield batch

data = list(range(1, 11))

for batch in batches(data, 3):
    print(batch)
````

# 🚀 Project 9 — 09 — Generator Analytics Pipeline

## Business Problem

Validate, filter, and aggregate business values lazily.

## Architecture

```text
INPUT
 ↓
ITERABLE / ITERATOR
 ↓
LAZY PROCESSING
 ↓
FILTER / TRANSFORM
 ↓
OUTPUT
```

## Implementation

````python
def valid_amounts(values):
    for value in values:
        try:
            amount = float(value)
        except (TypeError, ValueError):
            continue

        if amount > 0:
            yield amount

def high_value(values, threshold):
    for value in values:
        if value >= threshold:
            yield value

raw = ["1000", "bad", "-200", "5000", "7500", None, "2500"]

pipeline = high_value(valid_amounts(raw), 2500)

print("High-value sales:", list(pipeline))
````

# 🚀 Project 10 — 10 — Streaming Analytics Mini Project

## Business Problem

Build a lazy analytics workflow that validates sales records, filters high-value sales, and aggregates the result.

## Architecture

```text
INPUT
 ↓
ITERABLE / ITERATOR
 ↓
LAZY PROCESSING
 ↓
FILTER / TRANSFORM
 ↓
OUTPUT
```

## Implementation

````python
def sales_records(rows):
    for row in rows:
        try:
            customer, amount = row.split(",")
            amount = float(amount)
        except (ValueError, AttributeError):
            continue

        if amount > 0:
            yield {
                "customer": customer.strip(),
                "amount": amount
            }

def high_value_sales(records, threshold):
    for record in records:
        if record["amount"] >= threshold:
            yield record

raw_rows = [
    "Kaif,65000",
    "Sara,18000",
    "Invalid",
    "Ali,-500",
    "Maya,42000",
]

records = sales_records(raw_rows)
high_value = high_value_sales(records, 40000)

count = 0
total = 0.0

for record in high_value:
    count += 1
    total += record["amount"]
    print(record)

print("High-value count:", count)
print("High-value total:", total)
````

# 📚 Iterator & Generator Quick Reference

| Concept | Meaning |
|---|---|
| Iterable | Object that can provide an iterator |
| Iterator | Object that produces values with `next()` |
| `iter()` | Obtains an iterator |
| `next()` | Requests the next value |
| `StopIteration` | Signals exhaustion |
| Generator | Lazy iterator produced by generator syntax |
| `yield` | Produces a value and pauses execution |
| Generator expression | Lazy expression using parentheses |
| `yield from` | Delegates to another iterable/generator |
| Lazy evaluation | Produce values only when requested |

## Core Syntax

```python
iterator = iter(iterable)
value = next(iterator)

def generator():
    yield value

expression = (x for x in iterable)
```

```text

████████████████
     ↓
  MEMORY


GENERATOR

 ↓

 ↓

 ↓

 ↓

 ↓
...
```

## Quick Decision

```text
Need all values repeatedly?
        ↓
       LIST

Need one-pass / large / streaming processing?
        ↓
     GENERATOR

        ↓

```
