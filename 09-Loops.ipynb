# 1. What Is a Loop?

A loop repeats a block of code.

Instead of writing:

```python
print(1)
print(2)
print(3)
print(4)
print(5)
```

we can write:

```python
for i in range(1, 6):
    print(i)
```

### Why Loops Matter

Loops are essential for:

- Processing multiple values
- Repeating calculations
- Searching data
- Generating patterns
- Building algorithms
- Processing records
- Automating repetitive tasks

````python
for i in range(1, 6):
    print(i)
````

# 2. `for` Loop

A `for` loop iterates over the items of an iterable.

### Basic Syntax

```python
for variable in iterable:
    statement
```

Example:

```python
for name in ["Kaif", "Ali", "Sara"]:
    print(name)
```

The variable receives each item one at a time.

````python
names = ["Kaif", "Ali", "Sara"]

for name in names:
    print(name)
````

# 3. `range()`

`range()` is commonly used with `for` loops.

### Forms

```python
range(stop)
range(start, stop)
range(start, stop, step)
```

The `stop` value is excluded.

### Examples

```python
range(5)
```

produces:

```text
0 1 2 3 4
```

```python
range(2, 7)
```

produces:

```text
2 3 4 5 6
```

```python
range(10, 0, -2)
```

produces:

```text
10 8 6 4 2
```

````python
print("range(5):")

for i in range(5):
    print(i)

print("\nrange(2, 7):")

for i in range(2, 7):
    print(i)

print("\nrange(10, 0, -2):")

for i in range(10, 0, -2):
    print(i)
````

# 4. `while` Loop

A `while` loop repeats while its condition remains true.

### Syntax

```python
while condition:
    statement
```

Example:

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

### Important

A `while` loop needs a condition that eventually becomes false, unless
you intentionally create an infinite loop.

````python
count = 1

while count <= 5:
    print(count)
    count += 1
````

# 5. `for` vs `while`

| Feature | `for` | `while` |
|---|---|---|
| Main idea | Iterate over an iterable | Repeat while condition is true |
| Common use | Known collection/range | Condition-controlled repetition |
| Counter needed | Not always | Often |
| Infinite loop risk | Lower | Higher |
| Example | Process each item | Keep asking until valid input |

### Simple Rule

Use `for` when you naturally have items to iterate over.

Use `while` when repetition depends on a condition.

# 6. Loop Counters and Accumulators

Loops frequently maintain a value.

### Counter

Counts occurrences:

```python
count = 0

for i in range(5):
    count += 1
```

### Accumulator

Builds a total:

```python
total = 0

for number in range(1, 6):
    total += number
```

These two patterns are fundamental for problem solving.

````python
total = 0

for number in range(1, 6):
    total += number

print("Total:", total)

count = 0

for number in range(1, 6):
    count += 1

print("Count:", count)
````

# 7. Nested Loops

A nested loop is a loop inside another loop.

### Structure

```python
for outer in range(...):
    for inner in range(...):
        ...
```

The inner loop completes its iterations for each iteration of the outer
loop.

### Example

```text
Outer 1 → Inner 1, 2, 3
Outer 2 → Inner 1, 2, 3
Outer 3 → Inner 1, 2, 3
```

````python
for row in range(1, 4):
    for column in range(1, 4):
        print(f"({row}, {column})")
````

# 8. Infinite Loops

An infinite loop never reaches a false condition.

Example:

```python
while True:
    print("Running")
```

This is sometimes intentional, such as in continuously running programs,
but it must have a safe exit mechanism when appropriate.

### Controlled Infinite Loop

```python
while True:
    command = input("Enter q to quit: ")

    if command == "q":
        break
```

Never create an accidental infinite loop by forgetting to update the
condition.

````python
while True:
    command = input("Enter q to quit: ").strip().lower()

    if command == "q":
        print("Loop stopped.")
        break

    print("Command received:", command)
````

# 9. `break`

`break` immediately terminates the nearest enclosing loop.

### Example

```python
for number in range(1, 11):
    if number == 6:
        break
    print(number)
```

Output stops before `6` is printed.

````python
for number in range(1, 11):
    if number == 6:
        break
    print(number)
````

# 10. `continue`

`continue` skips the rest of the current iteration and moves to the next
iteration.

### Example

Skip even numbers:

```python
for number in range(1, 11):
    if number % 2 == 0:
        continue
    print(number)
```

````python
for number in range(1, 11):
    if number % 2 == 0:
        continue
    print(number)
````

# 11. `pass`

`pass` does nothing.

It is useful as a placeholder when Python requires a statement but you
have not implemented the logic yet.

```python
for number in range(5):
    if number == 3:
        pass
    print(number)
```

Unlike `break`, `pass` does not stop the loop.

Unlike `continue`, `pass` does not skip the remaining statements.

````python
for number in range(5):
    if number == 3:
        pass

    print(number)
````

# 12. Loop `else`

Python loops can have an `else` block.

The loop `else` runs when the loop finishes normally, but not when it is
terminated by `break`.

Example:

````python
for number in range(1, 6):
    print(number)
else:
    print("Loop completed normally.")
````

# 13. Nested Loop + `break`

`break` exits only the nearest enclosing loop.

```python
for row in range(3):
    for column in range(3):
        if column == 1:
            break
        print(row, column)
```

Understanding which loop is nearest is important when working with nested
loops.

````python
for row in range(3):
    for column in range(3):
        if column == 1:
            break
        print(row, column)
````

# 16. Common Loop Mistakes

### Mistake 1 — Forgetting to update a `while` loop

❌

```python
count = 1

while count <= 5:
    print(count)
```

The condition never changes.

### Mistake 2 — Wrong indentation

Python uses indentation to define the loop body.

### Mistake 3 — Off-by-one error

Remember that:

```python
range(1, 6)
```

stops before `6`.

### Mistake 4 — Using `break` when you need `continue`

`break` stops the loop.

`continue` skips only the current iteration.

### Mistake 5 — Accidentally modifying loop logic

Keep counters, conditions, and updates easy to understand.

# 🚀 Project 04 — Sum of Numbers

## Problem

Ask the user for a positive integer `n` and calculate the sum from `1`
through `n`.

Example:

```text
n = 5

1 + 2 + 3 + 4 + 5 = 15
```

````python
# Project 04 — Sum of Numbers

n = int(input("Enter a positive integer: "))

if n < 1:
    print("Please enter a positive integer.")
else:
    total = 0

    for number in range(1, n + 1):
        total += number

    print(f"Sum from 1 to {n}: {total}")
````

# 🚀 Project 05 — Factorial

## Problem

Calculate the factorial of a non-negative integer.

Definition:

```text
0! = 1
n! = 1 × 2 × 3 × ... × n
```

Example:

```text
5! = 120
```

Use a loop.

````python
# Project 05 — Factorial

n = int(input("Enter a non-negative integer: "))

if n < 0:
    print("Factorial is not defined for negative integers.")
else:
    factorial = 1

    for number in range(2, n + 1):
        factorial *= number

    print(f"{n}! = {factorial}")
````

# 🚀 Project 06 — Prime Number Checker

## Problem

Check whether a number is prime.

A prime number is greater than 1 and has no positive divisors other than
1 and itself.

### Efficient beginner approach

Only test divisors up to the square root of the number.

For this project, use:

```python
number ** 0.5
```

to determine the upper limit.

````python
# Project 06 — Prime Number Checker

number = int(input("Enter an integer: "))

if number <= 1:
    print("Not prime")
else:
    is_prime = True

    for divisor in range(2, int(number ** 0.5) + 1):
        if number % divisor == 0:
            is_prime = False
            break

    if is_prime:
        print(f"{number} is prime")
    else:
        print(f"{number} is not prime")
````

# 🚀 Project 07 — Prime Numbers in a Range

## Problem

Ask for a limit and print all prime numbers from `2` through that limit.

This combines:

- Nested logic
- `for`
- `range()`
- `break`
- Boolean flags

````python
# Project 07 — Prime Numbers in a Range

limit = int(input("Enter the upper limit: "))

if limit < 2:
    print("There are no prime numbers in this range.")
else:
    print("Prime numbers:")

    for number in range(2, limit + 1):
        is_prime = True

        for divisor in range(2, int(number ** 0.5) + 1):
            if number % divisor == 0:
                is_prime = False
                break

        if is_prime:
            print(number, end=" ")

    print()
````

# 🚀 Project 08 — Fibonacci Sequence

## Problem

Generate the first `n` Fibonacci numbers.

The sequence begins:

```text
0, 1, 1, 2, 3, 5, 8, 13, ...
```

Each next value is the sum of the previous two values.

````python
# Project 08 — Fibonacci Sequence

n = int(input("How many Fibonacci numbers? "))

if n <= 0:
    print("Enter a positive integer.")
else:
    first = 0
    second = 1

    for _ in range(n):
        print(first, end=" ")

        first, second = second, first + second

    print()
````

# 🚀 Project 09 — Reverse a Number

## Problem

Reverse the digits of a positive integer.

Example:

```text
Input  : 12345
Output : 54321
```

Use arithmetic and a `while` loop.

### Logic

```text
number
  ↓
last digit = number % 10
  ↓
add digit to reversed number
  ↓
remove last digit using // 10
  ↓
repeat until number becomes 0
```

````python
# Project 09 — Reverse Number

number = int(input("Enter a positive integer: "))

if number < 0:
    print("Please enter a positive integer.")
else:
    original = number
    reversed_number = 0

    while number > 0:
        digit = number % 10
        reversed_number = reversed_number * 10 + digit
        number //= 10

    print(f"Original Number : {original}")
    print(f"Reversed Number : {reversed_number}")
````

# 🚀 Project 10 — Sum of Digits

## Problem

Calculate the sum of all digits in an integer.

Example:

```text
12345
↓
1 + 2 + 3 + 4 + 5
↓
15
```

````python
# Project 10 — Sum of Digits

number = abs(int(input("Enter an integer: ")))

digit_sum = 0

while number > 0:
    digit_sum += number % 10
    number //= 10

print(f"Sum of digits: {digit_sum}")
````

# 🚀 Project 11 — Multiplication Tables

## Problem

Print tables from 1 through a user-selected limit.

This is a practical nested-loop exercise.

````python
# Project 11 — Multiple Tables

limit = int(input("Enter table limit: "))

if limit < 1:
    print("Enter a positive limit.")
else:
    for number in range(1, limit + 1):
        print(f"\nTable of {number}")

        for multiplier in range(1, 11):
            print(
                f"{number} × {multiplier} = "
                f"{number * multiplier}"
            )
````
