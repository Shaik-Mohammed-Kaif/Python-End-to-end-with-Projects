# 1. What Makes a Function "Advanced"?

Advanced functions are not a completely different type of function.

They are techniques for treating functions as reusable values and for
building flexible, composable program logic.

The key idea is:

```text
Function
   ↓
Can be stored
   ↓
Can be passed
   ↓
Can be returned
   ↓
Can be composed
   ↓
Can build reusable behavior
```

These concepts appear in callbacks, validation pipelines, data processing,
decorators, event handling, APIs, and functional-style programming.

# 2. Lambda Functions

A lambda is a small anonymous function.

### Syntax

```python
lambda parameters: expression
```

Example:

```python
square = lambda x: x * x
```

A lambda contains a single expression. For larger logic, use `def` because
it is easier to read, document, test, and maintain.

````python
square = lambda x: x * x

print(square(5))
print(square(10))
````

# 5. Functions as Objects

Python functions are first-class objects.

That means a function can be:

- Assigned to a variable
- Stored in a list or dictionary
- Passed as an argument
- Returned from another function

This is the foundation of higher-order functions.

````python
def greet(name):
    return f"Hello, {name}!"

say_hello = greet

print(say_hello("Kaif"))
print(greet is say_hello)
````

Because functions are objects, they can be stored in dictionaries.

````python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

operations = {
    "add": add,
    "subtract": subtract,
    "multiply": multiply
}

print(operations["add"](10, 5))
print(operations["multiply"](10, 5))
````

# 7. Higher-Order Functions

A higher-order function is a function that:

1. Accepts another function as an argument, or
2. Returns a function.

Example:

```python
def apply_operation(value, operation):
    return operation(value)
```

This allows behavior to be supplied separately from the main function.

````python
def apply_operation(value, operation):
    return operation(value)

def square(x):
    return x * x

def cube(x):
    return x ** 3

print(apply_operation(5, square))
print(apply_operation(5, cube))
````

# 8. Higher-Order Function Returning a Function

A function can create and return another function.

This pattern is important for closures and configurable behavior.

````python
def create_multiplier(factor):
    def multiply(number):
        return number * factor

    return multiply

double = create_multiplier(2)
triple = create_multiplier(3)

print(double(10))
print(triple(10))
````

# 9. Nested Functions

A nested function is defined inside another function.

```python
def outer():
    def inner():
        ...
```

The inner function can access variables from its enclosing function.

````python
def outer():
    def inner():
        return "Inside the nested function"

    return inner()

print(outer())
````

# 10. Closures

A closure is a function that remembers values from its enclosing scope even
after the enclosing function has finished executing.

Typical structure:

```text
outer()
  ↓
creates local value
  ↓
defines inner function
  ↓
inner remembers value
  ↓
returns inner
```

Closures are useful for configurable behavior and maintaining small pieces
of state without exposing that state directly.

````python
def make_greeting(prefix):
    def greet(name):
        return f"{prefix}, {name}!"

    return greet

morning = make_greeting("Good morning")
evening = make_greeting("Good evening")

print(morning("Kaif"))
print(evening("Kaif"))
````

# 11. Inspecting a Closure

A closure may retain values through its function closure cells.

For learning/debugging, `__closure__` can show captured cells.

Do not rely on internal implementation details for ordinary application
logic; the important concept is that the returned function retains access
to the enclosing value.

````python
def make_power(exponent):
    def power(number):
        return number ** exponent

    return power

square = make_power(2)

print(square(5))
print(square.__closure__)
````

# 12. `nonlocal` in Closures

`nonlocal` allows a nested function to rebind a variable from its nearest
enclosing function scope.

This is useful for stateful closures.

````python
def make_counter():
    count = 0

    def counter():
        nonlocal count
        count += 1
        return count

    return counter

counter = make_counter()

print(counter())
print(counter())
print(counter())
````

# 13. Recursion

Recursion occurs when a function calls itself.

Every recursive algorithm needs a **base case** that stops recursion.

### Mental model

```text
recursive_call()
      ↓
smaller_problem
      ↓
recursive_call()
      ↓
...
      ↓
base_case
      ↓
return
```

Recursion can be elegant for naturally recursive structures such as trees,
but iterative solutions are often simpler for straightforward loops.

````python
def countdown(n):
    if n == 0:
        print("Done")
        return

    print(n)
    countdown(n - 1)

countdown(5)
````

# 14. Recursive Factorial

Mathematically:

```text
n! = n × (n-1)!
0! = 1
```

The base case is `0! = 1`.

````python
def factorial(n):
    if n < 0:
        raise ValueError("n must be non-negative")

    if n == 0:
        return 1

    return n * factorial(n - 1)

print(factorial(5))
````

# 15. Recursive Fibonacci

A simple recursive definition:

```text
F(0) = 0
F(1) = 1
F(n) = F(n-1) + F(n-2)
```

The naive recursive implementation is useful for learning but becomes
inefficient for larger `n` because it repeats work.

````python
def fibonacci(n):
    if n < 0:
        raise ValueError("n must be non-negative")

    if n < 2:
        return n

    return fibonacci(n - 1) + fibonacci(n - 2)

print([fibonacci(i) for i in range(10)])
````

# 21. `reduce()`

`reduce()` repeatedly combines items into one result.

It lives in the `functools` module.

### Syntax

```python
from functools import reduce

reduce(function, iterable)
```

Conceptually:

```text
a, b, c, d
 ↓
combine(a, b)
 ↓
combine(result, c)
 ↓
combine(result, d)
 ↓
final result
```

````python
from functools import reduce

numbers = [1, 2, 3, 4, 5]

total = reduce(
    lambda a, b: a + b,
    numbers
)

print(total)
````

# 22. `reduce()` With an Initial Value

An optional initializer can be supplied.

```python
reduce(function, iterable, initializer)
```

The initializer becomes the starting accumulator value.

````python
from functools import reduce

numbers = [1, 2, 3, 4]

total = reduce(
    lambda total, value: total + value,
    numbers,
    100
)

print(total)
````

# 24. `zip()`

`zip()` combines corresponding elements from multiple iterables.

```python
zip(iterable1, iterable2)
```

It produces an iterator of tuples.

By default, iteration stops when the shortest input iterable is exhausted.

````python
names = ["Kaif", "Sara", "Ali"]
scores = [88, 94, 76]

for name, score in zip(names, scores):
    print(name, score)
````

# 29. `map()` vs List Comprehension

Both can transform data.

### `map()`

```python
list(map(str.upper, names))
```

### Comprehension

```python
[name.upper() for name in names]
```

Choose based on readability and the problem. A comprehension is often very
clear for simple transformations, while `map()` can be elegant when a
named function already exists.

````python
names = ["kaif", "sara", "ali"]

with_map = list(map(str.upper, names))
with_comprehension = [name.upper() for name in names]

print(with_map)
print(with_comprehension)
````

# 30. `filter()` vs Comprehension

These are both valid:

```python
list(filter(predicate, values))
```

and:

```python
[value for value in values if predicate(value)]
```

Use whichever communicates the intent most clearly.

````python
numbers = range(1, 11)

with_filter = list(
    filter(lambda x: x % 2 == 0, numbers)
)

with_comprehension = [
    x for x in numbers
    if x % 2 == 0
]

print(with_filter)
print(with_comprehension)
````

# 31. Function Composition

Function composition means using the output of one function as the input
of another.

```text

function A
 ↓
function B
 ↓
function C

```

This is a powerful foundation for pipelines.

````python
def clean_text(text):
    return text.strip().lower()

def count_words(text):
    return len(text.split())

text = "  Python Makes Analytics Powerful  "

cleaned = clean_text(text)
result = count_words(cleaned)

print("Cleaned:", cleaned)
print("Word Count:", result)
````

# 32. Building a Reusable Pipeline

A higher-order function can apply multiple operations sequentially.

````python
def apply_pipeline(value, *functions):
    for function in functions:
        value = function(value)
    return value

def strip_text(text):
    return text.strip()

def lowercase(text):
    return text.lower()

def add_label(text):
    return f"TEXT: {text}"

result = apply_pipeline(
    "  Python  ",
    strip_text,
    lowercase,
    add_label
)

print(result)
````

# 35. Advanced Recursion — Directory-Like Tree

Recursive functions are naturally suited to tree structures.

Conceptually:

```text
Root
├── Child
│   ├── Child
│   └── Child
└── Child
```

A recursive function processes a node and then recursively processes its
children.

````python
def count_nodes(tree):
    total = 1

    for child in tree["children"]:
        total += count_nodes(child)

    return total

tree = {
    "name": "root",
    "children": [
        {
            "name": "A",
            "children": [
                {"name": "A1", "children": []},
                {"name": "A2", "children": []}
            ]
        },
        {
            "name": "B",
            "children": []
        }
    ]
}

print("Node count:", count_nodes(tree))
````

# 36. Advanced Function Pitfalls

### Pitfall 1 — Overusing lambda

Bad:

```python
complex_lambda = lambda x: ...
```

Prefer a named function when logic becomes difficult to understand.

### Pitfall 2 — Forgetting that `map()` and `filter()` are iterators

They are not lists.

```python
result = map(...)
```

Materialize them with `list()` only when needed.

### Pitfall 3 — Naive recursion can be expensive

Recursive algorithms can repeat work or hit Python's recursion depth.

### Pitfall 4 — Closures capture variables, not snapshots

Be careful with late binding when creating closures in loops.

### Pitfall 5 — `reduce()` can reduce readability

Use built-ins such as `sum()`, `min()`, and `max()` when they express the
intent directly.

# 37. Closure Late-Binding Example

Closures created in a loop can capture the same final loop variable.

This is an important advanced Python behavior.

The following intentionally demonstrates the issue:

````python
functions = []

for i in range(3):
    functions.append(lambda: i)

print([function() for function in functions])
````

# 38. Fixing Late Binding With a Default Argument

A default argument can capture the current value at function creation.

```python
lambda i=i: i
```

This is a common technique when creating callbacks in loops.

````python
functions = []

for i in range(3):
    functions.append(lambda i=i: i)

print([function() for function in functions])
````

# 🚀 Project 01 — Sales Transformation Pipeline

## Business Problem

Create a reusable functional pipeline that cleans product names, converts prices, applies discounts, and calculates final values.

## Workflow

```text
INPUT
 ↓
FUNCTIONS
 ↓
TRANSFORM / FILTER / VALIDATE
 ↓
COMPOSE
 ↓
RESULT
```

## Solution

````python
def clean_product(name):
    return name.strip().title()

def apply_discount(price):
    return price * 0.90

def format_price(price):
    return round(price, 2)

def pipeline(value, *functions):
    for function in functions:
        value = function(value)
    return value

products = [
    {"name": " laptop ", "price": 65000},
    {"name": " mouse ", "price": 1200},
    {"name": " monitor ", "price": 18000}
]

for product in products:
    product["name"] = clean_product(product["name"])
    product["final_price"] = pipeline(
        product["price"],
        apply_discount,
        format_price
    )

for product in products:
    print(product)
````

# 🚀 Project 02 — Employee Salary Analyzer

## Business Problem

Use `map()`, `filter()`, `max()`, and higher-order functions to analyze employee salaries.

## Workflow

```text
INPUT
 ↓
FUNCTIONS
 ↓
TRANSFORM / FILTER / VALIDATE
 ↓
COMPOSE
 ↓
RESULT
```

## Solution

````python
employees = [
    {"name": "Kaif", "salary": 65000},
    {"name": "Sara", "salary": 85000},
    {"name": "Ali", "salary": 72000},
    {"name": "Omar", "salary": 55000}
]

salaries = list(map(
    lambda employee: employee["salary"],
    employees
))

high_earners = list(filter(
    lambda employee: employee["salary"] >= 70000,
    employees
))

highest = max(
    employees,
    key=lambda employee: employee["salary"]
)

print("Salaries:", salaries)
print("High Earners:", high_earners)
print("Highest Paid:", highest)
````

# 🚀 Project 03 — Customer Risk Rules

## Business Problem

Build configurable customer-risk rules using closures and higher-order functions.

## Workflow

```text
INPUT
 ↓
FUNCTIONS
 ↓
TRANSFORM / FILTER / VALIDATE
 ↓
COMPOSE
 ↓
```

## Solution

````python
def make_threshold_rule(field, threshold):
    def rule(customer):
        return customer.get(field, 0) >= threshold

    return rule

high_spend_rule = make_threshold_rule("monthly_spend", 2000)
long_tenure_rule = make_threshold_rule("tenure_months", 24)

customers = [
    {
        "name": "Kaif",
        "monthly_spend": 2499,
        "tenure_months": 30
    },
    {
        "name": "Ali",
        "monthly_spend": 799,
        "tenure_months": 12
    },
    {
        "name": "Sara",
        "monthly_spend": 2999,
        "tenure_months": 36
    }
]

for customer in customers:
    print(
        customer["name"],
        "High Spend:", high_spend_rule(customer),
        "Long Tenure:", long_tenure_rule(customer)
    )
````

# 🚀 Project 04 — Transaction Validation Pipeline

## Business Problem

Create a validation pipeline where each function checks a different business rule.

## Workflow

```text
INPUT
 ↓
FUNCTIONS
 ↓
TRANSFORM / FILTER / VALIDATE
 ↓
COMPOSE
 ↓
RESULT
```

## Solution

````python
def has_customer_id(transaction):
    return bool(transaction.get("customer_id"))

def has_positive_amount(transaction):
    return transaction.get("amount", 0) > 0

def has_valid_category(transaction):
    return transaction.get("category") in {
        "Food",
        "Transport",
        "Shopping"
    }

def validate(transaction, *rules):
    return all(rule(transaction) for rule in rules)

rules = (
    has_customer_id,
    has_positive_amount,
    has_valid_category
)

transactions = [
    {"customer_id": 101, "amount": 500, "category": "Food"},
    {"customer_id": 102, "amount": -20, "category": "Food"},
    {"customer_id": 103, "amount": 800, "category": "Travel"}
]

for transaction in transactions:
    print(validate(transaction, *rules))
````

# 🚀 Project 05 — Grade Processing Engine

## Business Problem

Use `map()` and functions to transform marks, calculate averages, and classify grades.

## Workflow

```text
INPUT
 ↓
FUNCTIONS
 ↓
TRANSFORM / FILTER / VALIDATE
 ↓
COMPOSE
 ↓
RESULT
```

## Solution

````python
def average(marks):
    return sum(marks) / len(marks)

def grade(score):
    if score >= 90:
        return "A+"
    if score >= 80:
        return "A"
    if score >= 70:
        return "B"
    if score >= 60:
        return "C"
    if score >= 40:
        return "D"
    return "F"

students = [
    {"name": "Kaif", "marks": [88, 92, 85]},
    {"name": "Sara", "marks": [95, 91, 94]},
    {"name": "Ali", "marks": [72, 68, 76]}
]

averages = list(map(
    lambda student: average(student["marks"]),
    students
))

for student, score in zip(students, averages):
    print(
        student["name"],
        f"Average={score:.2f}",
        "Grade=" + grade(score)
    )
````

# 🚀 Project 06 — Inventory Filter & Ranking

## Business Problem

Filter inventory by stock and rank products by value using higher-order functions.

## Workflow

```text
INPUT
 ↓
FUNCTIONS
 ↓
TRANSFORM / FILTER / VALIDATE
 ↓
COMPOSE
 ↓
RESULT
```

## Solution

````python
inventory = [
    {"name": "Laptop", "price": 65000, "stock": 8},
    {"name": "Mouse", "price": 1200, "stock": 40},
    {"name": "Monitor", "price": 18000, "stock": 6},
    {"name": "Keyboard", "price": 2500, "stock": 25}
]

low_stock = list(filter(
    lambda item: item["stock"] <= 10,
    inventory
))

ranked = sorted(
    inventory,
    key=lambda item: item["price"] * item["stock"],
    reverse=True
)

for item in ranked:
    item["inventory_value"] = item["price"] * item["stock"]

print("Low Stock:", low_stock)

print("\nRanked Inventory:")

for item in ranked:
    print(
        item["name"],
        "₹",
        item["inventory_value"]
    )
````

# 🚀 Project 07 — Text Processing Toolkit

## Business Problem

Build reusable text functions and compose them into a configurable text-processing pipeline.

## Workflow

```text
INPUT
 ↓
FUNCTIONS
 ↓
TRANSFORM / FILTER / VALIDATE
 ↓
COMPOSE
 ↓
RESULT
```

## Solution

````python
def strip_text(text):
    return text.strip()

def lowercase(text):
    return text.lower()

def normalize_spaces(text):
    return " ".join(text.split())

def word_count(text):
    return len(text.split())

def process_text(text, *steps):
    for step in steps:
        text = step(text)
    return text

text = "   Python   makes   Data   Analysis   powerful   "

cleaned = process_text(
    text,
    strip_text,
    lowercase,
    normalize_spaces
)

print("Cleaned:", cleaned)
print("Word Count:", word_count(cleaned))
````

# 🚀 Project 08 — Financial Metrics Engine

## Business Problem

Use `reduce()` where aggregation is appropriate, and compare it with clearer built-ins.

## Workflow

```text
INPUT
 ↓
FUNCTIONS
 ↓
TRANSFORM / FILTER / VALIDATE
 ↓
COMPOSE
 ↓
RESULT
```

## Solution

````python
from functools import reduce

revenues = [12000, 18000, 9500, 22000, 14500]

total_with_reduce = reduce(
    lambda total, value: total + value,
    revenues,
    0
)

total_with_sum = sum(revenues)

average = total_with_sum / len(revenues)

print("Reduce Total:", total_with_reduce)
print("Built-in Total:", total_with_sum)
print(f"Average: {average:.2f}")
````

# 🚀 Project 09 — Multi-Rule Customer Segmentation

## Business Problem

Create reusable rule functions and classify customers into business segments.

## Workflow

```text
INPUT
 ↓
FUNCTIONS
 ↓
TRANSFORM / FILTER / VALIDATE
 ↓
COMPOSE
 ↓
RESULT
```

## Solution

````python
def is_high_value(customer):
    return customer["monthly_spend"] >= 2500

def is_long_term(customer):
    return customer["tenure"] >= 24

def segment_customer(customer):
    if is_high_value(customer) and is_long_term(customer):
        return "VIP"

    if is_high_value(customer):
        return "High Value"

    if is_long_term(customer):
        return "Loyal"

    return "Standard"

customers = [
    {"name": "Kaif", "monthly_spend": 3000, "tenure": 36},
    {"name": "Ali", "monthly_spend": 1000, "tenure": 30},
    {"name": "Sara", "monthly_spend": 2800, "tenure": 10},
    {"name": "Omar", "monthly_spend": 900, "tenure": 8}
]

for customer in customers:
    print(
        customer["name"],
        "→",
        segment_customer(customer)
    )
````

```text
TRANSFORM / FILTER / VALIDATE
 ↓
COMPOSE
 ↓
RESULT
```

## Solution

````python
def make_range_validator(field, minimum, maximum):
    def validate(record):
        value = record.get(field)
        return (
            isinstance(value, (int, float))
            and minimum <= value <= maximum
        )

    return validate

age_rule = make_range_validator("age", 18, 100)
score_rule = make_range_validator("score", 0, 100)

records = [
    {"name": "Kaif", "age": 22, "score": 88},
    {"name": "Sara", "age": 27, "score": 95},
    {"name": "Invalid", "age": 12, "score": 120}
]

rules = (age_rule, score_rule)

def is_valid(record):
    return all(rule(record) for rule in rules)

valid_records = list(filter(is_valid, records))
valid_names = list(map(
    lambda record: record["name"],
    valid_records
))

print("Valid Records:", valid_records)
print("Valid Names:", valid_names)
````

```text
        ↓
```
