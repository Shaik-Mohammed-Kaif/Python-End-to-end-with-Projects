# 1. What Is a Module?

A **module** is a Python file containing code that can be imported and reused.

For example:

```text
math_tools.py
```

can contain functions such as:

```python
def add(a, b):
    return a + b
```

Another Python file can import and use that functionality.

### Mental Model

```text
math_tools.py
     ↓
   module
     ↓
 import
     ↓
main_program.py
     ↓
reuse functions/classes
```

# 2. `import`

The basic import syntax is:

```python
import module_name
```

After importing, access members with dot notation:

```python
module_name.member
```

Example using the Python standard library:

````python
import math

print(math.sqrt(81))
print(math.pi)
````

# 3. `from`

`from` imports a specific member from a module.

```python
from module_name import member
```

This allows direct access to the imported member.

````python
from math import sqrt, pi

print(sqrt(64))
print(pi)
````

# 4. `as`

`as` creates an alias.

```python
import module_name as alias
```

or:

```python
from module_name import member as alias
```

Aliases can make long names shorter and can avoid naming conflicts.

````python
import datetime as dt

today = dt.date.today()

print(today)
````

# 5. Import Multiple Members

You can import multiple known members in one statement:

```python
from math import sqrt, floor, ceil
```

Prefer explicit imports when they improve readability.

````python
from math import sqrt, floor, ceil

number = 8.7

print(sqrt(81))
print(floor(number))
print(ceil(number))
````

# 11. Module Namespace

```python

```

Python binds the module name `math` in the current namespace.

Then:

```python

```

accesses `sqrt` through that module namespace.

This is often clearer than importing many names directly.

````python
import math

square_root = math.sqrt(144)

print(square_root)
````

# 14. What Is a Package?

A **package** is a way to organize related Python modules into a directory
structure.

Example:

```text
analytics/
├── __init__.py
├── cleaning.py
├── metrics.py
└── reports.py
```

Modern Python also supports namespace packages that do not require
`__init__.py` in every package directory, but a traditional package commonly
contains it and it remains useful for package initialization and compatibility.

# 15. Package Import

With a package such as:

```text
analytics/
├── __init__.py
└── metrics.py
```

you can import:

```python
from analytics.metrics import calculate_total
```

This gives your project a clean namespace.

````python
# Create and import a small traditional package.

from pathlib import Path
import sys

package_root = Path("package_demo")
analytics_dir = package_root / "analytics"
analytics_dir.mkdir(parents=True, exist_ok=True)

(analytics_dir / "__init__.py").write_text(
    "",
    encoding="utf-8"
)

(analytics_dir / "metrics.py").write_text(
    "def total(values):\n"
    "    return sum(values)\n",
    encoding="utf-8"
)

sys.path.insert(0, str(package_root.resolve()))

from analytics.metrics import total

print(total([100, 200, 300]))
````

# 16. `__init__.py`

In a traditional package, `__init__.py` is the package initialization module.

It can be empty or contain package-level setup/export logic.

Do not put unrelated application code there.

### Structure

```text
analytics/
└── __init__.py
```

The file itself is a module named `analytics.__init__` conceptually, while
the directory provides the package namespace.

# 17. `__name__`

Every Python module has a special `__name__` attribute.

When a file is run directly, Python normally sets:

```python
__name__ == "__main__"
```

When the same file is imported, its `__name__` is normally its module name.

This distinction is fundamental to reusable Python modules.

````python
print("Module name:", __name__)
````

# 18. `if __name__ == "__main__":`

This common pattern separates reusable definitions from code intended to run
when the file is executed directly.

```python
def main():
    print("Application started")

if __name__ == "__main__":
    main()
```

When imported, the `main()` call does not automatically execute because the
condition is false.

````python
def main():
    print("Main function executed.")

if __name__ == "__main__":
    main()
````

# 19. Why Use the Main Guard?

The main guard helps:

- prevent execution of script-only code during import
- make modules reusable
- provide a clear application entry point
- support testing and composition

### Architecture

```text
module definitions
      ↓
functions / classes
      ↓
main()
      ↓
if __name__ == "__main__":
      ↓
run application
```

# 20. Module Import Execution

When a module is imported for the first time in a Python process, its
top-level code is executed.

Therefore, keep import-time code lightweight and avoid unexpected side
effects.

Put executable application logic behind functions and, when appropriate,
the main guard.

````python
from pathlib import Path

Path("import_demo.py").write_text(
    "print('Module imported')\n"
    "def hello():\n"
    "    return 'Hello'\n"
    "\n"
    "if __name__ == '__main__':\n"
    "    print('Executed directly')\n",
    encoding="utf-8"
)

import sys
sys.path.insert(0, str(Path(".").resolve()))

import import_demo

print(import_demo.hello())
````

# 21. Importing a Module Multiple Times

Python caches imported modules in `sys.modules`.

In normal use, importing the same module again does not simply execute the
module's top-level code from scratch each time.

This contributes to predictable module behavior and avoids repeated loading.

````python
import sys
import math

print("math loaded:", "math" in sys.modules)
print("Module object:", sys.modules["math"])
````

# 29. Common Import Errors

### `ModuleNotFoundError`

Usually means Python could not locate the requested module/package.

### `ImportError`

The module may exist, but the requested import may not be available.

### `AttributeError`

The module loaded, but the requested attribute does not exist.

### Common causes

- wrong package/module name
- dependency not installed
- wrong virtual environment
- circular imports
- name conflicts
- incompatible versions

````python
try:
    import module_that_does_not_exist
except ModuleNotFoundError as error:
    print("Import failed:", error)
````

# 30. Circular Imports

A circular import occurs when modules depend on each other in a cycle.

Example:

```text
module_a
   ↓
module_b
   ↓
module_a
```

Symptoms can include partially initialized modules and confusing import
errors.

### Better design

Move shared functionality into a third module:

```text
module_a ──→ common
module_b ──→ common
```

Avoid unnecessary import cycles by designing clear dependency direction.

# 🚀 Project 01 — Calculator Module

## Business Problem

Create a reusable calculator module and import its functions from a main program.

## Architecture

```text
INPUT
 ↓
MODULE / PACKAGE
 ↓
REUSABLE LOGIC
 ↓
MAIN PROGRAM
 ↓
OUTPUT
```

## Solution

````python
from pathlib import Path
import sys
import importlib

root = Path("calculator_project")
root.mkdir(parents=True, exist_ok=True)

calculator_file = root / "calculator.py"

calculator_file.write_text(
    """
def add(a, b):
    return a + b


def multiply(a, b):
    return a * b
""".strip() + "\n",
    encoding="utf-8"
)

root_path = str(root.resolve())

if root_path not in sys.path:
    sys.path.insert(0, root_path)

importlib.invalidate_caches()

if "calculator" in sys.modules:
    del sys.modules["calculator"]

from calculator import add, multiply

print("Add:", add(10, 20))
print("Multiply:", multiply(10, 20))
````

# 🚀 Project 02 — Customer Analytics Package

## Business Problem

Build a package containing customer metrics and import those metrics into an application.

## Architecture

```text
INPUT
 ↓
MODULE / PACKAGE
 ↓
REUSABLE LOGIC
 ↓
MAIN PROGRAM
 ↓
OUTPUT
```

## Solution

````python
from pathlib import Path
import sys
import importlib

root = Path("customer_analytics_project")
package = root / "analytics"

package.mkdir(parents=True, exist_ok=True)

init_file = package / "__init__.py"
metrics_file = package / "metrics.py"

init_file.write_text(
    "",
    encoding="utf-8"
)

metrics_file.write_text(
    """
def average_revenue(values):
    if not values:
        return 0

    return sum(values) / len(values)
""".strip() + "\n",
    encoding="utf-8"
)

root_path = str(root.resolve())

if root_path not in sys.path:
    sys.path.insert(0, root_path)

importlib.invalidate_caches()

for module_name in list(sys.modules):
    if module_name == "analytics" or module_name.startswith("analytics."):
        del sys.modules[module_name]

from analytics.metrics import average_revenue

revenue = [1200, 1800, 2400]

average = average_revenue(revenue)

print("Revenue:", revenue)
print("Average Revenue:", average)
````

# 🚀 Project 07 — Validation Package

## Business Problem

Build reusable validation functions inside a package.

## Architecture

```text
INPUT
 ↓
MODULE / PACKAGE
 ↓
REUSABLE LOGIC
 ↓
MAIN PROGRAM
 ↓
OUTPUT
```

## Solution

````python
from pathlib import Path
import sys

root = Path("validation_project")
package = root / "validation"
package.mkdir(parents=True, exist_ok=True)

(package / "__init__.py").write_text("", encoding="utf-8")

(package / "rules.py").write_text(
    "def validate_age(age):\n"
    "    if not isinstance(age, int):\n"
    "        raise TypeError('age must be an integer')\n"
    "    if not 0 <= age <= 120:\n"
    "        raise ValueError('age must be between 0 and 120')\n"
    "    return True\n",
    encoding="utf-8"
)

sys.path.insert(0, str(root.resolve()))

from validation.rules import validate_age

for age in [22, 130]:
    try:
        print(age, validate_age(age))
    except (TypeError, ValueError) as error:
        print(age, "→", error)
````

# 🚀 Project 08 — `__name__` / `__main__` Demo

## Business Problem

Build a module that can be imported safely while also supporting direct execution.

## Architecture

```text
INPUT
 ↓
MODULE / PACKAGE
 ↓
REUSABLE LOGIC
 ↓
MAIN PROGRAM
 ↓
OUTPUT
```

## Solution

````python
from pathlib import Path
import subprocess
import sys

script = Path("main_guard_demo.py")

script.write_text(
    "def main():\n"
    "    print('Application entry point')\n"
    "\n"
    "if __name__ == '__main__':\n"
    "    main()\n",
    encoding="utf-8"
)

result = subprocess.run(
    [sys.executable, str(script)],
    capture_output=True,
    text=True,
    check=True
)

print(result.stdout.strip())
````

# 🚀 Project 09 — Multi-Module Analytics Pipeline

## Business Problem

Separate loading, transformation, and metrics into modules and orchestrate them from a main program.

## Architecture

```text
INPUT
 ↓
MODULE / PACKAGE
 ↓
REUSABLE LOGIC
 ↓
MAIN PROGRAM
 ↓
OUTPUT
```

## Solution

````python
from pathlib import Path
import sys

root = Path("analytics_pipeline")
root.mkdir(exist_ok=True)

(root / "loader.py").write_text(
    "def load_sales():\n"
    "    return [1000, 2500, 1800, 3200]\n",
    encoding="utf-8"
)

(root / "transform.py").write_text(
    "def positive_sales(values):\n"
    "    return [value for value in values if value > 0]\n",
    encoding="utf-8"
)

(root / "metrics.py").write_text(
    "def total(values):\n"
    "    return sum(values)\n"
    "\n"
    "def average(values):\n"
    "    return sum(values) / len(values) if values else 0\n",
    encoding="utf-8"
)

sys.path.insert(0, str(root.resolve()))

from loader import load_sales
from transform import positive_sales
from metrics import total, average

sales = positive_sales(load_sales())

print("Sales:", sales)
print("Total:", total(sales))
print("Average:", average(sales))
````


## Solution

````python
from pathlib import Path
import sys

root = Path("production_package")
package = root / "app"
package.mkdir(parents=True, exist_ok=True)

(package / "__init__.py").write_text("", encoding="utf-8")

(package / "config.py").write_text(
    "APP_NAME = 'Customer Analytics'\n"
    "VERSION = '1.0'\n",
    encoding="utf-8"
)

(package / "service.py").write_text(
    "def active_customers(customers):\n"
    "    return [c for c in customers if c.get('active')]\n",
    encoding="utf-8"
)

(package / "report.py").write_text(
    "def summary(customers):\n"
    "    return {'count': len(customers)}\n",
    encoding="utf-8"
)

sys.path.insert(0, str(root.resolve()))

from app.config import APP_NAME, VERSION
from app.service import active_customers
from app.report import summary

customers = [
    {"name": "Kaif", "active": True},
    {"name": "Sara", "active": False},
    {"name": "Ali", "active": True},
]

active = active_customers(customers)

print(APP_NAME, VERSION)
print(summary(active))
````

# 🧠 Import Decision Guide

```text
Need a complete module namespace?
        ↓
import module

Need only specific members?
        ↓
from module import member

Name is long/conflicting?
        ↓
use as alias

Codebase is growing?
        ↓
organize modules

Related modules belong together?
        ↓
create a package

File should run directly but stay reusable?
        ↓
main() + __name__ guard

Import cycle?
        ↓
redesign dependency direction
```
