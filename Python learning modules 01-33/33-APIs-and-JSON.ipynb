# 1. What Is an API?

**API** means Application Programming Interface.

A web API allows one application to communicate with another application over
a network.

Example:

```text
Python Application
       ↓
     API Request
       ↓
    Web Server
       ↓
   API Response
       ↓
     JSON Data
       ↓
Python Application
```

Real-world APIs can provide information such as products, weather, payments,
customers, orders, maps, or application data.

# 2. What Is HTTP?

**HTTP** (Hypertext Transfer Protocol) is a protocol commonly used for
communication between clients and web servers.

A request contains information such as:

```text
Method
URL
Headers
Optional Body
```

A response contains:

```text
Status Code
Headers
Optional Body
```

# 3. Request and Response

The basic API interaction is:

```text
CLIENT
  │
  │ HTTP REQUEST
  ▼
SERVER
  │
  │ HTTP RESPONSE
  ▼
CLIENT
```

Python commonly acts as the client when consuming an API.

# 4. HTTP Methods

# 9. Checking HTTP Success

`raise_for_status()` raises an HTTP error for unsuccessful HTTP status codes.

This is often cleaner than manually checking every possible 4xx/5xx code.

````python
import requests

response = requests.get(
    "https://jsonplaceholder.typicode.com/todos/1",
    timeout=10,
)

response.raise_for_status()

print("Request succeeded:", response.status_code)
print(response.json())
````

# 10. JSON

**JSON** (JavaScript Object Notation) is a common data-interchange format.

Example:

```json
{
  "id": 101,
  "name": "Kaif",
  "active": true
}
```

Python representation:

```python
{
    "id": 101,
    "name": "Kaif",
    "active": True,
}
```

Common JSON values include:

```text
object → Python dict
array → Python list
string → Python str
number → int / float
true / false → True / False
null → None
```

# 11. API JSON → Python

`response.json()` parses a JSON response into Python objects when the response
contains valid JSON.

````python
import requests

response = requests.get(
    "https://jsonplaceholder.typicode.com/users/1",
    timeout=10,
)

response.raise_for_status()

user = response.json()

print(type(user))
print("Name:", user["name"])
print("Email:", user["email"])
````

# 12. JSON Module

Python also provides the built-in `json` module.

Important functions:

```text
json.loads()  → JSON string to Python object
json.dumps()  → Python object to JSON string
json.load()   → JSON file to Python object
json.dump()   → Python object to JSON file
```

````python
import json

data = {
    "name": "Kaif",
    "scores": [90, 95, 88],
    "active": True,
}

json_text = json.dumps(data, indent=2)

print(json_text)

parsed = json.loads(json_text)

print(parsed["name"])
print(parsed["scores"])
````

# 20. Timeouts

Network requests should normally have an explicit timeout.

Prefer:

```python
requests.get(url, timeout=10)
```

rather than relying on an indefinite wait.

Timeouts help applications fail predictably when a network service is slow or
unavailable.

````python
import requests

response = requests.get(
    "https://jsonplaceholder.typicode.com/todos/1",
    timeout=10,
)

print(response.status_code)
````

# 21. API Error Handling

A network request can fail because of:

- Connection problems
- Timeout
- Invalid URL
- HTTP error
- Invalid JSON
- API rate limiting
- Authentication failure

Handle the type of failure that your application can meaningfully recover
from.

````python
import requests

url = "https://jsonplaceholder.typicode.com/todos/1"

try:
    response = requests.get(url, timeout=10)
    response.raise_for_status()
    data = response.json()
    print("Success:", data)

except requests.exceptions.Timeout:
    print("Request timed out")

except requests.exceptions.RequestException as error:
    print("Request failed:", error)

except ValueError:
    print("Response was not valid JSON")
````

# 24. Pagination

APIs often return data in pages instead of returning every record at once.

Common pagination styles include:

```text
page / limit
offset / limit
cursor
next URL
```

The API documentation defines which method is supported.

````python
import requests

all_records = []

for page in range(1, 4):
    response = requests.get(
        "https://jsonplaceholder.typicode.com/todos",
        params={"_page": page, "_limit": 5},
        timeout=10,
    )
    response.raise_for_status()
    all_records.extend(response.json())

print("Collected:", len(all_records))
print("First:", all_records[0])
````

# 27. Flattening Nested JSON

# 28. Sessions

`requests.Session()` can persist settings such as headers and can reuse
connections across multiple requests.

This is useful when an application makes many requests to the same service.

````python
import requests

with requests.Session() as session:
    session.headers.update({
        "Accept": "application/json"
    })

    response = session.get(
        "https://jsonplaceholder.typicode.com/todos/1",
        timeout=10,
    )

    response.raise_for_status()

    print(response.json())
````

# 29. API Client Function

Instead of writing request code everywhere, wrap API access in functions.

Benefits:

- Reusability
- Centralized error handling
- Easier testing
- Cleaner application logic

````python
import requests

def get_todo(todo_id: int) -> dict:
    url = f"https://jsonplaceholder.typicode.com/todos/{todo_id}"

    response = requests.get(url, timeout=10)
    response.raise_for_status()

    return response.json()

todo = get_todo(1)

print(todo["title"])
````

# 30. Common Mistakes

### Mistake 1 — No timeout

Network requests can wait unexpectedly.

### Mistake 2 — Hard-coded secrets

Never publish real credentials.

### Mistake 3 — Ignoring status codes

A response should not automatically be treated as successful.

### Mistake 4 — Assuming every response is JSON

Check the API contract/content before parsing.

### Mistake 5 — Ignoring rate limits

Respect the API's documented limits.

### Mistake 6 — Downloading unnecessary data

Use filtering and pagination when available.

### Mistake 7 — Hard-coding URLs everywhere

Centralize API configuration and client logic.

# 31. Best Practices

- Read the API documentation first.
- Use explicit timeouts.
- Check status codes.
- Use `raise_for_status()` where appropriate.
- Use parameterized query parameters.
- Use `json=` for JSON request bodies.
- Keep credentials outside source code.
- Handle network errors.
- Respect rate limits.
- Use pagination for large datasets.
- Validate important response data.
- Separate API clients from business logic.
- Log useful request failures without leaking secrets.
- Convert API data into structured Python objects before analytics.

# 🚀 Project 4 — API Pagination Pipeline

## Business Problem

Collect multiple API pages into one Python dataset.

## Architecture

```text
API
 ↓
HTTP REQUEST
 ↓
JSON RESPONSE
 ↓
VALIDATION
 ↓
PYTHON
 ↓
DATAFRAME / REPORT
```

## Implementation

````python
import requests

records = []

for page in range(1, 6):
    response = requests.get(
        "https://jsonplaceholder.typicode.com/todos",
        params={"_page": page, "_limit": 10},
        timeout=10,
    )
    response.raise_for_status()
    records.extend(response.json())

print("Records collected:", len(records))

completed = sum(
    1 for record in records
    if record["completed"]
)

print("Completed:", completed)
print("Pending:", len(records) - completed)
````

# 🚀 Project 5 — API Error Handling Service

## Business Problem

Build a reusable function that handles common request failures.

## Architecture

```text
API
 ↓
HTTP REQUEST
 ↓
JSON RESPONSE
 ↓
VALIDATION
 ↓
PYTHON
 ↓
DATAFRAME / REPORT
```

## Implementation

````python
import requests

def fetch_json(url: str) -> dict | list:
    try:
        response = requests.get(url, timeout=10)
        response.raise_for_status()
        return response.json()

    except requests.exceptions.Timeout:
        raise RuntimeError("API request timed out")

    except requests.exceptions.RequestException as error:
        raise RuntimeError(f"API request failed: {error}") from error

    except ValueError as error:
        raise RuntimeError("API returned invalid JSON") from error


data = fetch_json(
    "https://jsonplaceholder.typicode.com/todos/1"
)

print(data)
````

# 🚀 Project 7 — API Client Class

## Business Problem

Create a reusable object-oriented API client.

## Architecture

```text
API
 ↓
HTTP REQUEST
 ↓
JSON RESPONSE
 ↓
VALIDATION
 ↓
PYTHON
 ↓
DATAFRAME / REPORT
```

## Implementation

````python
import requests

class TodoAPI:
    def __init__(self, base_url: str):
        self.base_url = base_url.rstrip("/")
        self.session = requests.Session()
        self.session.headers.update({
            "Accept": "application/json"
        })

    def get_todo(self, todo_id: int) -> dict:
        response = self.session.get(
            f"{self.base_url}/todos/{todo_id}",
            timeout=10,
        )
        response.raise_for_status()
        return response.json()

    def close(self) -> None:
        self.session.close()


api = TodoAPI("https://jsonplaceholder.typicode.com")

try:
    todo = api.get_todo(1)
    print(todo)
finally:
    api.close()
````

# 🚀 Project 8 — API Monitoring Report

## Business Problem

Collect API records and generate simple operational metrics.

## Architecture

```text
API
 ↓
HTTP REQUEST
 ↓
JSON RESPONSE
 ↓
VALIDATION
 ↓
PYTHON
 ↓
DATAFRAME / REPORT
```

## Implementation

````python
import requests
from collections import Counter

response = requests.get(
    "https://jsonplaceholder.typicode.com/todos",
    timeout=10,
)
response.raise_for_status()

todos = response.json()

status_counts = Counter(
    "completed" if todo["completed"] else "pending"
    for todo in todos
)

user_counts = Counter(todo["userId"] for todo in todos)

print("Total records:", len(todos))
print("Status:", dict(status_counts))
print("Top users:", user_counts.most_common(5))
````

# 🚀 Project 10 — Production-Style API Analytics Pipeline

## Business Problem

Combine API client logic, validation, error handling, JSON normalization, and analytics.

## Architecture

```text
API
 ↓
HTTP REQUEST
 ↓
JSON RESPONSE
 ↓
VALIDATION
 ↓
PYTHON
 ↓
DATAFRAME / REPORT
```

## Implementation

````python
import requests
import pandas as pd


class APIClient:
    def __init__(self, base_url: str):
        self.base_url = base_url.rstrip("/")
        self.session = requests.Session()
        self.session.headers.update({
            "Accept": "application/json"
        })

    def get_users(self) -> list[dict]:
        response = self.session.get(
            f"{self.base_url}/users",
            timeout=10,
        )
        response.raise_for_status()

        data = response.json()

        if not isinstance(data, list):
            raise ValueError("Expected a JSON list")

        return data

    def close(self) -> None:
        self.session.close()


client = APIClient("https://jsonplaceholder.typicode.com")

try:
    users = client.get_users()

    df = pd.json_normalize(users)

    required_columns = ["id", "name", "email", "address.city"]

    missing = [
        column for column in required_columns
        if column not in df.columns
    ]

    if missing:
        raise ValueError(f"Missing columns: {missing}")

    report = df[required_columns].copy()

    print("API ANALYTICS REPORT")
    print("Records:", len(report))
    print("\nUsers by city:")
    print(report["address.city"].value_counts())

    print("\nSample:")
    print(report.head())

finally:
    client.close()
````
