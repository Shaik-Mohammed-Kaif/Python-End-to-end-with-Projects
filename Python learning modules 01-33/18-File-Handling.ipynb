# 1. What Is File Handling?

File handling means using Python to create, open, read, write, update, and manage files.

Common files:
- `.txt` — plain text
- `.csv` — tabular data
- `.json` — structured data
- `.log` — application logs
- Binary files such as images and other non-text data

### Basic workflow

```text
PATH → OPEN → READ/WRITE → PROCESS → CLOSE
```

Using `with open(...)` is normally preferred because Python manages the file resource safely.

````python
from pathlib import Path

path = Path("example.txt")
path.write_text("Hello, Python!", encoding="utf-8")

with open(path, "r", encoding="utf-8") as file:
    data = file.read()

print(data)
````

# 2. `open()`

The built-in `open()` function creates a file object connected to a file.

```python
open(file, mode="r", encoding=None)
```

Important arguments include the path, mode, encoding, newline handling, and other stream options.

````python
file = open("example_open.txt", "w", encoding="utf-8")
file.write("Python File Handling")
file.close()

with open("example_open.txt", "r", encoding="utf-8") as file:
    print(file.read())
````

# 3. File Modes

| Mode | Meaning |
|---|---|
| `r` | Read; file must exist |
| `w` | Write; creates or truncates |
| `a` | Append; creates if needed |
| `x` | Exclusive creation |
| `rb` | Read binary |
| `wb` | Write binary |

**Important:** `w` can destroy existing content by truncating the file.

````python
from pathlib import Path

Path("modes.txt").write_text("Original", encoding="utf-8")

with open("modes.txt", "a", encoding="utf-8") as file:
    file.write(" + appended")

print(Path("modes.txt").read_text(encoding="utf-8"))
````

# 4. `read()`

`read()` reads text from the current file position. With no size, it reads the remaining content.

````python
from pathlib import Path

Path("data.txt").write_text("Python\nSQL\nExcel", encoding="utf-8")

with open("data.txt", "r", encoding="utf-8") as file:
    print(file.read())
````

# 5. `read(size)`

Pass a size to read only part of the stream.

````python
from pathlib import Path

Path("message.txt").write_text("Python File Handling", encoding="utf-8")

with open("message.txt", "r", encoding="utf-8") as file:
    print(file.read(6))
````

# 6. `readline()`

Reads one line at a time.

````python
from pathlib import Path

Path("skills.txt").write_text("Python\nSQL\nExcel\n", encoding="utf-8")

with open("skills.txt", "r", encoding="utf-8") as file:
    print(file.readline().strip())
    print(file.readline().strip())
````

# 7. `readlines()`

Returns the remaining lines as a list. For very large files, line-by-line iteration is often more memory-friendly.

````python
from pathlib import Path

Path("skills.txt").write_text("Python\nSQL\nExcel\n", encoding="utf-8")

with open("skills.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()

print(lines)
````

# 8. Iterating Over a File

A file object is iterable, allowing large text files to be processed one line at a time.

````python
from pathlib import Path

Path("sales.log").write_text(
    "sale,1000\nsale,2500\nsale,1800\n",
    encoding="utf-8"
)

with open("sales.log", "r", encoding="utf-8") as file:
    for line in file:
        print(line.strip())
````

# 9. Writing — `write()`

`write()` writes a string and returns the number of characters written.

````python
with open("report.txt", "w", encoding="utf-8") as file:
    count = file.write("Monthly Sales Report\n")
    file.write("Total Sales: 125000\n")

print("Characters written:", count)
````

# 10. `writelines()`

`writelines()` writes an iterable of strings. It does not add separators automatically.

````python
lines = ["Python\n", "SQL\n", "Excel\n"]

with open("skills.txt", "w", encoding="utf-8") as file:
    file.writelines(lines)

print("File written successfully.")
````

# 11. Append Mode — `a`

Append mode preserves existing content and writes new content at the end.

````python
with open("activity.log", "w", encoding="utf-8") as file:
    file.write("Application started\n")

with open("activity.log", "a", encoding="utf-8") as file:
    file.write("Report generated\n")

with open("activity.log", "r", encoding="utf-8") as file:
    print(file.read())
````

# 12. Exclusive Creation — `x`

`x` creates a new file and raises `FileExistsError` if it already exists.

````python
from pathlib import Path

path = Path("new_report.txt")

try:
    with open(path, "x", encoding="utf-8") as file:
        file.write("First version")
except FileExistsError:
    print("The file already exists.")
````

# 13. `with open()`

A context manager automatically handles resource cleanup.

```python
with open("data.txt", "r", encoding="utf-8") as file:
    data = file.read()
```

This is the preferred pattern for ordinary file access.

````python
from pathlib import Path

Path("context.txt").write_text(
    "Context managers make cleanup safer.",
    encoding="utf-8"
)

with open("context.txt", "r", encoding="utf-8") as file:
    print(file.read())

print("File block completed.")
````

# 14. File Object Properties

Useful properties/methods include `name`, `mode`, `closed`, `readable()`, and `writable()`.

````python
with open("state.txt", "w+", encoding="utf-8") as file:
    file.write("Python")
    print("Name:", file.name)
    print("Mode:", file.mode)
    print("Readable:", file.readable())
    print("Writable:", file.writable())
    print("Closed inside with:", file.closed)

print("Closed after with:", file.closed)
````

# 15. `tell()` and `seek()`

`tell()` reports the current stream position. `seek()` changes the stream position.

````python
from pathlib import Path

Path("position.txt").write_text("Python", encoding="utf-8")

with open("position.txt", "r", encoding="utf-8") as file:
    print("Initial:", file.tell())
    print("Read:", file.read(2))
    print("After read:", file.tell())
    file.seek(0)
    print("After seek:", file.tell())
    print("Again:", file.read(2))
````

# 16. Text vs Binary

Text mode works with `str` and performs text encoding/decoding. Binary mode works with `bytes`.

````python
with open("binary_demo.bin", "wb") as file:
    file.write(b"Python")

with open("binary_demo.bin", "rb") as file:
    data = file.read()

print(data)
print(type(data).__name__)
````

# 17. Binary Files — `rb` / `wb`

Use binary modes for non-text content such as images and other binary data.

````python
source = b"Binary Python Data"

with open("source.bin", "wb") as file:
    file.write(source)

with open("source.bin", "rb") as file:
    copied = file.read()

print(copied)
print(copied == source)
````

# 19. Creating Directories

# 20. Listing Files

`Path.iterdir()` iterates over directory entries. `glob()` is useful for pattern-based discovery.

````python
from pathlib import Path

data_dir = Path("data_files")
data_dir.mkdir(exist_ok=True)

(data_dir / "sales.csv").write_text("id,amount\n1,500", encoding="utf-8")
(data_dir / "notes.txt").write_text("hello", encoding="utf-8")

csv_files = [
    path.name
    for path in data_dir.iterdir()
    if path.is_file() and path.suffix == ".csv"
]

print(csv_files)
````

# 21. CSV — `csv.reader` / `csv.writer`

Python's standard `csv` module provides `reader`, `writer`, `DictReader`, and `DictWriter` for tabular text files.

````python
import csv

rows = [
    ["id", "product", "amount"],
    [1, "Laptop", 65000],
    [2, "Mouse", 1200],
]

with open("sales.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerows(rows)

with open("sales.csv", "r", newline="", encoding="utf-8") as file:
    for row in csv.reader(file):
        print(row)
````

# 22. CSV — `DictReader`

`DictReader` represents each data row as a dictionary using the header names.

````python
import csv

with open("customers.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["id", "name", "city"])
    writer.writerow([101, "Kaif", "Mumbai"])
    writer.writerow([102, "Sara", "Delhi"])

with open("customers.csv", "r", newline="", encoding="utf-8") as file:
    for row in csv.DictReader(file):
        print(dict(row))
````

# 23. CSV — `DictWriter`

`DictWriter` writes dictionaries as CSV rows. Define `fieldnames` explicitly.

````python
import csv

records = [
    {"id": 101, "name": "Kaif", "score": 88},
    {"id": 102, "name": "Sara", "score": 95},
]

with open("scores.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(
        file,
        fieldnames=["id", "name", "score"]
    )
    writer.writeheader()
    writer.writerows(records)

print("CSV created.")
````

# 24. JSON — `dump()` / `load()`

JSON is commonly used for structured data exchange and configuration.

- `json.dump()` → write Python data to a file
- `json.load()` → read JSON from a file
- `json.dumps()` → create a JSON string
- `json.loads()` → parse a JSON string

````python
import json

customer = {
    "id": 101,
    "name": "Kaif",
    "skills": ["Python", "SQL", "Excel"]
}

with open("customer.json", "w", encoding="utf-8") as file:
    json.dump(customer, file, indent=4)

with open("customer.json", "r", encoding="utf-8") as file:
    loaded = json.load(file)

print(loaded)
````

# 25. JSON — `dumps()` / `loads()`

# 30. Large Files

For very large text files, avoid loading the entire file into memory with `read()`. Iterate line by line when practical.

````python
from pathlib import Path

Path("large_like.log").write_text(
    "INFO,100\nERROR,500\nINFO,250\n",
    encoding="utf-8"
)

error_count = 0

with open("large_like.log", "r", encoding="utf-8") as file:
    for line in file:
        if line.startswith("ERROR"):
            error_count += 1

print("Errors:", error_count)
````

# 33. Common Mistakes

### Mistake 1 — Forgetting `with`
Use `with open(...)` for managed file access.

### Mistake 2 — Accidentally using `w`
`w` truncates existing content.

### Mistake 3 — Reading huge files with `read()`
Stream line by line when appropriate.

### Mistake 4 — Missing encoding
Specify an appropriate text encoding.

### Mistake 5 — Wrong CSV newline handling
Use `newline=""` with the standard `csv` module.

### Mistake 6 — Confusing text and binary
Text streams use `str`; binary streams use `bytes`.

### Mistake 7 — Assuming a relative path is absolute
A relative path depends on the process's current working directory.

````python
from pathlib import Path

path = Path("safe_example.txt")

try:
    with path.open("r", encoding="utf-8") as file:
        print(file.read())
except FileNotFoundError:
    print("File not found:", path)
````

# 34. Best Practices

- Prefer `with open(...)` for managed file access.
- Specify text encoding explicitly.
- Use `newline=""` with Python's `csv` module.
- Use `pathlib` for filesystem paths.
- Choose the correct mode.
- Never use `w` when existing content must be preserved.
- Handle expected filesystem exceptions.
- Stream large files when appropriate.
- Validate imported data before processing it.
- Keep input, processing, and output stages separate.
- Avoid machine-specific hard-coded absolute paths.
- Never trust file contents blindly.
- Keep generated/temporary files separate from important source data.

# 🚀 Project 01 — Sales Text Report

## Business Problem

Read transaction text data, calculate total sales, and save a report.

## Workflow

```text
RAW FILE → OPEN → VALIDATE → PROCESS → ANALYZE → OUTPUT
```

## Solution

````python
from pathlib import Path

input_path = Path("sales.txt")
output_path = Path("sales_report.txt")

input_path.write_text(
    "Laptop,65000\nMouse,1200\nMonitor,18000\n",
    encoding="utf-8"
)

total = 0
count = 0

with input_path.open("r", encoding="utf-8") as file:
    for line in file:
        product, amount = line.strip().split(",")
        total += float(amount)
        count += 1

report = (
    f"Transactions: {count}\n"
    f"Total Sales: {total:.2f}\n"
)

output_path.write_text(report, encoding="utf-8")
print(report)
````

# 🚀 Project 02 — CSV Sales Analyzer

## Business Problem

Create a CSV, read it with `DictReader`, calculate KPIs, and save an analysis report.

## Workflow

```text
RAW FILE → OPEN → VALIDATE → PROCESS → ANALYZE → OUTPUT
```

## Solution

````python
import csv

records = [
    {"product": "Laptop", "amount": 65000},
    {"product": "Mouse", "amount": 1200},
    {"product": "Monitor", "amount": 18000},
]

with open("sales_data.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=["product", "amount"])
    writer.writeheader()
    writer.writerows(records)

total = 0
high_value = []

with open("sales_data.csv", "r", newline="", encoding="utf-8") as file:
    for row in csv.DictReader(file):
        amount = float(row["amount"])
        total += amount
        if amount >= 10000:
            high_value.append(row["product"])

report = (
    "Sales Analysis\n"
    f"Total Sales: {total:.2f}\n"
    f"High Value Products: {', '.join(high_value)}\n"
)

with open("sales_analysis.txt", "w", encoding="utf-8") as file:
    file.write(report)

print(report)
````

# 🚀 Project 03 — JSON Customer Manager

## Business Problem

Store customer records as JSON, load them, update a record, and save the revised data.

## Workflow

```text
RAW FILE → OPEN → VALIDATE → PROCESS → ANALYZE → OUTPUT
```

## Solution

````python
import json

customers = [
    {"id": 101, "name": "Kaif", "plan": "Premium", "active": True},
    {"id": 102, "name": "Sara", "plan": "Standard", "active": True},
]

with open("customers.json", "w", encoding="utf-8") as file:
    json.dump(customers, file, indent=4)

with open("customers.json", "r", encoding="utf-8") as file:
    loaded = json.load(file)

for customer in loaded:
    if customer["id"] == 102:
        customer["plan"] = "Premium"

with open("customers_updated.json", "w", encoding="utf-8") as file:
    json.dump(loaded, file, indent=4)

print(loaded)
````

# 🚀 Project 04 — Log File Analyzer

## Business Problem

Stream through a log file and count INFO, WARNING, and ERROR records.

## Workflow

```text
RAW FILE → OPEN → VALIDATE → PROCESS → ANALYZE → OUTPUT
```

## Solution

````python
from pathlib import Path

log_path = Path("application.log")

log_path.write_text(
    "INFO|Application started\n"
    "INFO|User logged in\n"
    "WARNING|Slow response\n"
    "ERROR|Database unavailable\n"
    "INFO|Application recovered\n",
    encoding="utf-8"
)

counts = {"INFO": 0, "WARNING": 0, "ERROR": 0}

with log_path.open("r", encoding="utf-8") as file:
    for line in file:
        level = line.split("|", 1)[0]
        if level in counts:
            counts[level] += 1

print(counts)
````

# 🚀 Project 06 — Backup File Utility

## Business Problem

Create a simple text-file backup using binary read/write.

## Workflow

```text
RAW FILE → OPEN → VALIDATE → PROCESS → ANALYZE → OUTPUT
```

## Solution

````python
from pathlib import Path

source = Path("important.txt")
backup = Path("important_backup.txt")

source.write_text(
    "Important project information.",
    encoding="utf-8"
)

with source.open("rb") as src:
    data = src.read()

with backup.open("wb") as dst:
    dst.write(data)

print("Backup created:", backup)
print("Same content:", source.read_bytes() == backup.read_bytes())
````

# 🚀 Project 07 — Customer Import Validator

## Business Problem

Read CSV customer records and separate valid records from rejected records.

## Workflow

```text
RAW FILE → OPEN → VALIDATE → PROCESS → ANALYZE → OUTPUT
```

## Solution

````python
import csv

rows = [
    ["id", "name", "age"],
    [101, "Kaif", 22],
    [102, "Sara", 27],
    [103, "Ali", 15],
    [104, "Omar", "unknown"],
]

with open("customer_import.csv", "w", newline="", encoding="utf-8") as file:
    csv.writer(file).writerows(rows)

valid = []
rejected = []

with open("customer_import.csv", "r", newline="", encoding="utf-8") as file:
    for row in csv.DictReader(file):
        try:
            age = int(row["age"])
            if not 18 <= age <= 100:
                raise ValueError("age outside valid range")
            valid.append(row)
        except (ValueError, TypeError) as error:
            rejected.append({"row": row, "error": str(error)})

print("Valid:", valid)
print("Rejected:", rejected)
````

# 🚀 Project 08 — Multi-File Report Merger

## Business Problem

Read several reports from a directory and combine them into a master report.

## Workflow

```text
RAW FILE → OPEN → VALIDATE → PROCESS → ANALYZE → OUTPUT
```

## Solution

````python
from pathlib import Path

report_dir = Path("daily_reports")
report_dir.mkdir(exist_ok=True)

(report_dir / "day1.txt").write_text("Day 1 Sales: 10000", encoding="utf-8")
(report_dir / "day2.txt").write_text("Day 2 Sales: 15000", encoding="utf-8")
(report_dir / "day3.txt").write_text("Day 3 Sales: 12000", encoding="utf-8")

master = []

for path in sorted(report_dir.glob("*.txt")):
    master.append(path.read_text(encoding="utf-8").strip())

master_report = "\n".join(master)
Path("master_report.txt").write_text(master_report, encoding="utf-8")

print(master_report)
````

# 🚀 Project 09 — JSON KPI Report

## Business Problem

Load sales data from JSON and generate a KPI summary JSON file.

## Workflow

```text
RAW FILE → OPEN → VALIDATE → PROCESS → ANALYZE → OUTPUT
```

## Solution

````python
import json

sales = [
    {"id": 1, "amount": 12000},
    {"id": 2, "amount": 18500},
    {"id": 3, "amount": 9200},
    {"id": 4, "amount": 24000},
]

with open("sales.json", "w", encoding="utf-8") as file:
    json.dump(sales, file, indent=4)

with open("sales.json", "r", encoding="utf-8") as file:
    data = json.load(file)

total = sum(item["amount"] for item in data)
average = total / len(data)
high_value = sum(
    1 for item in data
    if item["amount"] >= 15000
)

kpis = {
    "transaction_count": len(data),
    "total_sales": total,
    "average_sales": average,
    "high_value_transactions": high_value,
}

with open("sales_kpis.json", "w", encoding="utf-8") as file:
    json.dump(kpis, file, indent=4)

print(kpis)
````

# 🚀 Project 10 — End-to-End Data File Pipeline

## Business Problem

Build a complete mini pipeline: raw CSV → validation → clean JSON → KPI report.

## Workflow

```text
RAW FILE → OPEN → VALIDATE → PROCESS → ANALYZE → OUTPUT
```

## Solution

````python
import csv
import json
from pathlib import Path

raw_path = Path("raw_sales.csv")
clean_path = Path("clean_sales.json")
report_path = Path("final_report.txt")

raw_rows = [
    ["product", "amount"],
    ["Laptop", "65000"],
    ["Mouse", "1200"],
    ["Monitor", "18000"],
    ["Invalid", "not-a-number"],
]

with raw_path.open("w", newline="", encoding="utf-8") as file:
    csv.writer(file).writerows(raw_rows)

clean_records = []
rejected = []

with raw_path.open("r", newline="", encoding="utf-8") as file:
    for row in csv.DictReader(file):
        try:
            amount = float(row["amount"])
            if amount < 0:
                raise ValueError("negative amount")
            clean_records.append({
                "product": row["product"].strip(),
                "amount": amount,
            })
        except (ValueError, TypeError) as error:
            rejected.append({"row": row, "error": str(error)})

total = sum(item["amount"] for item in clean_records)
average = total / len(clean_records) if clean_records else 0

with clean_path.open("w", encoding="utf-8") as file:
    json.dump(clean_records, file, indent=4)

report = (
    "FINAL SALES REPORT\n"
    "==================\n"
    f"Valid Records: {len(clean_records)}\n"
    f"Rejected Records: {len(rejected)}\n"
    f"Total Sales: {total:.2f}\n"
    f"Average Sale: {average:.2f}\n"
)

report_path.write_text(report, encoding="utf-8")

print(report)
````

# 📚 File Handling Quick Reference

### Open

```python
with open("data.txt", "r", encoding="utf-8") as file:
    data = file.read()
```

### Modes

```text
r   → read
w   → write / truncate
a   → append
x   → exclusive create
rb  → binary read
wb  → binary write
```

### Read

```python
file.read()
file.readline()
file.readlines()
```

### Write

```python
file.write("text")
file.writelines(["a\n", "b\n"])
```

### Position

```python
file.tell()
file.seek(0)
```

### CSV

```python
import csv

csv.reader(...)
csv.writer(...)
csv.DictReader(...)
csv.DictWriter(...)
```

### JSON

```python
import json

json.load(file)
json.loads(text)
json.dump(data, file)
json.dumps(data)
```

### Paths

```python
from pathlib import Path

path = Path("data.csv")
path.exists()
path.is_file()
path.is_dir()
path.name
path.stem
path.suffix
path.parent
```

### Golden Rule

> **Use `with open(...)`, choose the correct mode, specify text encoding, and handle expected filesystem errors explicitly.**

# 🧠 File Handling Decision Guide

```text
READ TEXT?
  ↓ r

REPLACE / CREATE TEXT?
  ↓ w

ADD TO EXISTING CONTENT?
  ↓ a

CREATE ONLY IF ABSENT?
  ↓ x

BINARY DATA?
  ↓ rb / wb

TABULAR CSV?
  ↓ csv module

STRUCTURED JSON?
  ↓ json module

FILESYSTEM PATHS?
  ↓ pathlib.Path

SAFE RESOURCE CLEANUP?
  ↓ with open()

VERY LARGE TEXT?
  ↓ stream line by line
```

### Data Science Principle

```text
LOAD → VALIDATE → CLEAN → TRANSFORM → ANALYZE → SAVE
```
