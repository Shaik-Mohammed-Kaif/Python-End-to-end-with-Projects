# 1. `math`

`math` provides mathematical functions and constants for real-number calculations.

Important members: `sqrt()`, `ceil()`, `floor()`, `factorial()`, `gcd()`, `log()`, `log10()`, `exp()`, trigonometric functions, `pi`, and `e`.

## Example

````python
import math
print(math.sqrt(81))
print(math.ceil(4.2))
print(math.floor(4.8))
print(math.factorial(5))
print(math.gcd(24, 18))
print(math.pi)
````

# 2. `statistics`

Provides common statistical calculations such as `mean()`, `median()`, `mode()`, `fmean()`, `stdev()`, `pstdev()`, `variance()`, and `pvariance()`.

## Example

````python
import statistics
sales = [100, 150, 150, 200, 300]
print(statistics.mean(sales))
print(statistics.median(sales))
print(statistics.mode(sales))
print(statistics.pstdev(sales))
````

# 3. `decimal`

Provides decimal arithmetic with configurable precision. It is appropriate when exact decimal representation matters, such as many financial calculations.

Prefer constructing `Decimal` from strings when the exact decimal value matters.

## Example

````python
from decimal import Decimal
price = Decimal("19.99")
tax = Decimal("0.18")
total = price + price * tax
print(total)
````

# 4. `fractions`

`Fraction` represents rational numbers exactly.

## Example

````python
from fractions import Fraction
result = Fraction(1, 3) + Fraction(1, 6)
print(result)
print(float(result))
````

# 5. `datetime`

Provides `date`, `time`, `datetime`, `timedelta`, and timezone-related functionality for dates and times.

## Example

````python
from datetime import date, datetime, timedelta
today = date.today()
now = datetime.now()
print(today)
print(now)
print(today + timedelta(days=30))
````

# 6. `strftime()` and `strptime()`

`strftime()` formats date/time objects into strings. `strptime()` parses a string according to a format.

## Example

````python
from datetime import datetime
value = datetime(2026, 8, 20, 14, 30)
text = value.strftime("%Y-%m-%d %H:%M:%S")
parsed = datetime.strptime("2026-08-20", "%Y-%m-%d")
print(text)
print(parsed)
````

# 7. `calendar`

The `calendar` module supports calendar calculations such as leap-year checks and month information.

## Example

````python
import calendar
print(calendar.isleap(2028))
print(calendar.monthrange(2026, 8))
print(calendar.month(2026, 8))
````

# 8. `os`

`os` provides operating-system interfaces, including environment variables, the current working directory, and filesystem-related operations.

## Example

````python
import os
print(os.getcwd())
print(os.name)
os.environ["APP_MODE"] = "learning"
print(os.getenv("APP_MODE"))
````

# 9. `sys`

`sys` provides Python interpreter and runtime information such as `version`, `platform`, `argv`, `path`, and `modules`.

## Example

````python
import sys
print(sys.version.split()[0])
print(sys.platform)
print(len(sys.path))
````

# 10. `pathlib`

`pathlib.Path` provides an object-oriented interface for filesystem paths. Important members include `exists()`, `is_file()`, `is_dir()`, `mkdir()`, `glob()`, `iterdir()`, `read_text()`, `write_text()`, `name`, `stem`, `suffix`, and `parent`.

## Example

````python
from pathlib import Path
path = Path("standard_library_demo.txt")
path.write_text("Python Standard Library", encoding="utf-8")
print(path.exists())
print(path.name)
print(path.suffix)
print(path.read_text(encoding="utf-8"))
````

# 11. `json`

`json` converts JSON data to and from Python objects.

`dump()` / `load()` work with files. `dumps()` / `loads()` work with strings.

## Example

````python
import json
customer = {"id": 101, "name": "Kaif", "skills": ["Python", "SQL"], "active": True}
text = json.dumps(customer, indent=2)
restored = json.loads(text)
print(text)
print(restored["name"])
````

# 12. `csv`

The `csv` module supports `reader`, `writer`, `DictReader`, and `DictWriter` for tabular text files. Use `newline=""` when opening CSV files with this module.

## Example

````python
import csv
rows = [["name", "score"], ["Kaif", 90], ["Sara", 95]]
with open("scores.csv", "w", newline="", encoding="utf-8") as file:
    csv.writer(file).writerows(rows)
with open("scores.csv", "r", newline="", encoding="utf-8") as file:
    for row in csv.reader(file):
        print(row)
````

# 13. `sqlite3`

`sqlite3` provides access to SQLite databases. Use parameterized SQL rather than string concatenation.

## Example

````python
import sqlite3
with sqlite3.connect(":memory:") as connection:
    connection.execute("CREATE TABLE sales (product TEXT, amount REAL)")
    connection.execute("INSERT INTO sales VALUES (?, ?)", ("Laptop", 65000))
    rows = connection.execute("SELECT product, amount FROM sales").fetchall()
print(rows)
````

# 14. `random`

Provides pseudo-random generation. Common tools include `random()`, `randint()`, `randrange()`, `choice()`, `choices()`, `sample()`, and `shuffle()`.

**Security:** use `secrets` rather than `random` for security-sensitive values.

## Example

````python
import random
items = ["Python", "SQL", "Excel", "Power BI"]
print(random.random())
print(random.randint(1, 10))
print(random.choice(items))
print(random.sample(items, 2))
````

# 15. `collections`

Provides specialized containers including `Counter`, `defaultdict`, and `deque`.

## Example

````python
from collections import Counter, defaultdict, deque
words = ["python", "sql", "python", "excel", "python"]
print(Counter(words).most_common())
groups = defaultdict(list)
groups["data"].append("Python")
groups["data"].append("SQL")
print(dict(groups))
queue = deque(["A", "B"])
queue.append("C")
print(queue.popleft(), list(queue))
````

# 16. `itertools`

Provides efficient iterator-building tools such as `chain()`, `islice()`, `product()`, `permutations()`, and `combinations()`.

## Example

````python
from itertools import chain, islice, combinations
print(list(chain([1, 2], [3, 4])))
print(list(islice(range(100), 5)))
print(list(combinations(["Laptop", "Mouse", "Monitor"], 2)))
````

# 17. `functools`

Provides higher-order function utilities including `reduce()`, `partial()`, `lru_cache()`, `cache()`, and `wraps()`.

## Example

````python
from functools import reduce, partial, lru_cache
print(reduce(lambda a, b: a + b, [1, 2, 3, 4]))
double = partial(lambda x, factor: x * factor, factor=2)
print(double(10))
@lru_cache(maxsize=None)
def fib(n):
    if n < 2:
        return n
    return fib(n - 1) + fib(n - 2)
print(fib(20))
````

# 18. `re`

The `re` module provides regular-expression matching. Important operations include `search()`, `match()`, `fullmatch()`, `findall()`, `finditer()`, `sub()`, and `split()`.

## Example

````python
import re
text = "Order 101, Order 202, Order 303"
print(re.findall(r"\d+", text))
print(bool(re.fullmatch(r"Order-\d+", "Order-101")))
print(re.sub(r"\s+", " ", "Python   is   powerful").strip())
````

# 20. Common Mistakes

| Area | Common mistake | Better approach |
|---|---|---|
| `math` | expecting exact decimal money arithmetic | consider `Decimal` |
| `statistics` | treating it as a full dataframe library | use it for lightweight statistics |
| `decimal` | constructing from an imprecise float unnecessarily | prefer strings for exact decimal input |
| `datetime` | mixing incompatible time representations | keep date/time semantics explicit |
| `os` | hardcoding machine-specific paths | use `pathlib` |
| `json` | confusing file and string APIs | `load/dump` vs `loads/dumps` |
| `csv` | incorrect newline handling | use `newline=""` |
| `sqlite3` | concatenating user input into SQL | parameterized queries |
| `random` | using it for secrets | use `secrets` |
| `itertools` | materializing huge iterators unnecessarily | keep processing lazy |
| `re` | unreadable patterns | keep regexes simple and test them |

# 21. Best Practices

- Prefer the standard library when it genuinely satisfies the requirement.
- Use `pathlib` for modern path handling.
- Use context managers for files and database connections.
- Use parameterized SQLite queries.
- Use `Decimal` for appropriate exact-decimal financial calculations.
- Stream large files instead of loading everything into memory.
- Use iterator tools for memory-efficient workflows.
- Use `secrets` for security-sensitive randomness.
- Keep regular expressions readable.
- Validate external data.
- Separate loading, processing, analysis, and reporting.
- Avoid unnecessary third-party dependencies.

# 🚀 Project 01 — Sales File Analyzer

## Business Problem
Read CSV sales, calculate total/mean/median, and write a JSON report.

## Workflow

```text
INPUT → STANDARD LIBRARY → VALIDATE → PROCESS → ANALYZE → OUTPUT
```

## Solution

````python
import csv, json, statistics
from pathlib import Path

path = Path("sales_project.csv")
rows = [["product", "amount"], ["Laptop", "65000"], ["Mouse", "1200"], ["Monitor", "18000"]]
with path.open("w", newline="", encoding="utf-8") as f:
    csv.writer(f).writerows(rows)

amounts = []
with path.open("r", newline="", encoding="utf-8") as f:
    for row in csv.DictReader(f):
        amounts.append(float(row["amount"]))

report = {
    "count": len(amounts),
    "total": sum(amounts),
    "mean": statistics.mean(amounts),
    "median": statistics.median(amounts)
}
Path("sales_report.json").write_text(json.dumps(report, indent=2), encoding="utf-8")
print(report)
````

# 🚀 Project 07 — Cached Analytics Calculator

## Business Problem
Cache a deterministic recursive calculation.

## Workflow

```text
INPUT → STANDARD LIBRARY → VALIDATE → PROCESS → ANALYZE → OUTPUT
```

## Solution

````python
from functools import lru_cache
@lru_cache(maxsize=128)
def fibonacci(n):
    if n < 2:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)
print(fibonacci(25))
print(fibonacci.cache_info())
````

# 🚀 Project 09 — Regional Data Merger

## Business Problem
Combine CSV data and JSON targets, then calculate whether regions met target.

## Workflow

```text
INPUT → STANDARD LIBRARY → VALIDATE → PROCESS → ANALYZE → OUTPUT
```

## Solution

````python
import csv, json
from collections import defaultdict
from pathlib import Path

csv_path = Path("regions.csv")
csv_path.write_text(
    "region,amount\nWest,10000\nWest,15000\nEast,9000\n",
    encoding="utf-8"
)
targets = {"West": 20000, "East": 10000}
sales = defaultdict(float)
with csv_path.open("r", newline="", encoding="utf-8") as f:
    for row in csv.DictReader(f):
        sales[row["region"]] += float(row["amount"])

report = {
    region: {
        "sales": amount,
        "target": targets.get(region, 0),
        "target_met": amount >= targets.get(region, float("inf"))
    }
    for region, amount in sales.items()
}
print(json.dumps(report, indent=2))
````

# 🚀 Project 10 — Standard Library Analytics Pipeline

## Business Problem
Build an end-to-end CSV → validation → aggregation → JSON workflow using only standard-library modules.

## Workflow

```text
INPUT → STANDARD LIBRARY → VALIDATE → PROCESS → ANALYZE → OUTPUT
```

## Solution

````python
import csv, json, re, statistics
from collections import Counter
from datetime import date
from decimal import Decimal
from pathlib import Path

input_path = Path("pipeline.csv")
input_path.write_text(
    "customer,region,amount\nKaif,West,65000.00\nSara,East,18000.50\nAli,West,1200.00\n",
    encoding="utf-8"
)

records = []
with input_path.open("r", newline="", encoding="utf-8") as f:
    for row in csv.DictReader(f):
        if not re.fullmatch(r"[A-Za-z ]+", row["customer"]):
            continue
        records.append({
            "customer": row["customer"].strip(),
            "region": row["region"],
            "amount": Decimal(row["amount"])
        })

amounts = [float(r["amount"]) for r in records]
regions = Counter(r["region"] for r in records)

report = {
    "report_date": str(date.today()),
    "records": len(records),
    "total_sales": str(sum((r["amount"] for r in records), Decimal("0"))),
    "average_sales": statistics.mean(amounts),
    "region_counts": dict(regions)
}
Path("pipeline_report.json").write_text(json.dumps(report, indent=2), encoding="utf-8")
print(json.dumps(report, indent=2))
````
