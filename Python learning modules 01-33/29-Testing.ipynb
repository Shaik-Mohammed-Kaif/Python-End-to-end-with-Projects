# 1. Why Testing?

Testing provides evidence that code produces expected behavior.

Without tests:

```text
CODE CHANGE
    ↓
UNKNOWN IMPACT
    ↓
BUG RISK
```

With tests:

```text
CODE CHANGE
    ↓
AUTOMATED TESTS
    ↓
PASS / FAIL
    ↓
QUICK FEEDBACK
```

### Benefits

- catches regressions
- documents expected behavior
- improves refactoring confidence
- finds edge cases
- supports maintainability
- makes debugging easier

# 2. What Is a Test Case?

A test case defines:

```text
INPUT
 ↓
ACTION
 ↓
EXPECTED RESULT
 ↓
ACTUAL RESULT
 ↓
PASS / FAIL
```

Example:

```python
add(2, 3)
```

Expected result:

```text
5
```

# 3. Assertions

An assertion checks that an expected condition is true.

```python
assert condition
```

Example:

````python
def add(a: int, b: int) -> int:
    return a + b

assert add(2, 3) == 5

print("Test passed")
````

# 4. Assertions With Messages

A message helps explain why an assertion failed.

````python
def multiply(a: int, b: int) -> int:
    return a * b

assert multiply(4, 5) == 20, "Multiplication result is incorrect"

print("Test passed")
````

# 5. What Is Unit Testing?

A unit test checks a small, isolated unit of behavior, usually a function or
method.

Example:

```text
add()
 ↓
input
 ↓
expected output
```

A good unit test should be:

- focused
- repeatable
- understandable
- independent where practical
- fast

# 6. First `unittest` Example

Python includes `unittest` in the standard library.

```python
import unittest
```

A test method normally begins with `test_`.

````python
import unittest

def add(a: int, b: int) -> int:
    return a + b

class TestMath(unittest.TestCase):
    def test_add(self):
        self.assertEqual(add(2, 3), 5)

suite = unittest.defaultTestLoader.loadTestsFromTestCase(TestMath)
result = unittest.TextTestRunner(verbosity=2).run(suite)

print("Successful:", result.wasSuccessful())
````

# 7. Important `unittest` Assertions

Common assertions include:

```text
assertEqual()
assertNotEqual()
assertTrue()
assertFalse()
assertIsNone()
assertIsNotNone()
assertIn()
assertNotIn()
assertRaises()
```

These make test intent clearer than using raw `assert` in every situation.

````python
import unittest

class TestAssertions(unittest.TestCase):
    def test_common_assertions(self):
        self.assertEqual(10 + 5, 15)
        self.assertNotEqual(10, 20)
        self.assertTrue(5 > 2)
        self.assertFalse(5 < 2)
        self.assertIsNone(None)
        self.assertIsNotNone("Python")
        self.assertIn("Py", "Python")
        self.assertNotIn("Java", "Python")

suite = unittest.defaultTestLoader.loadTestsFromTestCase(TestAssertions)
result = unittest.TextTestRunner(verbosity=1).run(suite)

print("Successful:", result.wasSuccessful())
````

# 8. Testing Exceptions With `unittest`

Use `assertRaises()` when an exception is expected.

````python
import unittest

def divide(a: float, b: float) -> float:
    if b == 0:
        raise ValueError("Denominator cannot be zero")
    return a / b

class TestDivide(unittest.TestCase):
    def test_zero_denominator(self):
        with self.assertRaises(ValueError):
            divide(10, 0)

suite = unittest.defaultTestLoader.loadTestsFromTestCase(TestDivide)
result = unittest.TextTestRunner(verbosity=1).run(suite)

print("Successful:", result.wasSuccessful())
````

# 9. `setUp` and `tearDown`

`unittest` supports lifecycle methods.

```text
setUp()
   ↓
test
   ↓
tearDown()
```

Use them for setup/cleanup required by tests.

````python
import unittest

class TestCounter(unittest.TestCase):
    def setUp(self):
        self.values = [10, 20, 30]

    def test_sum(self):
        self.assertEqual(sum(self.values), 60)

    def tearDown(self):
        self.values = []

suite = unittest.defaultTestLoader.loadTestsFromTestCase(TestCounter)
result = unittest.TextTestRunner(verbosity=1).run(suite)

print("Successful:", result.wasSuccessful())
````

# 16. Arrange → Act → Assert

A clean test often follows:

```text
ARRANGE
  ↓
ACT
  ↓
ASSERT
```

Example:

````python
def calculate_total(price: float, quantity: int) -> float:
    return price * quantity

# Arrange
price = 250.0
quantity = 4

# Act
result = calculate_total(price, quantity)

# Assert
assert result == 1000.0

print("AAA test passed")
````

# 17. Test Edge Cases

Do not test only the normal path.

Consider:

- zero
- empty input
- minimum values
- maximum values
- negative values
- missing values
- duplicate values
- invalid input
- boundary conditions

````python
def average(values: list[float]) -> float:
    if not values:
        return 0.0
    return sum(values) / len(values)

assert average([10.0, 20.0]) == 15.0
assert average([]) == 0.0
assert average([5.0]) == 5.0

print("Edge-case tests passed")
````

# 18. Testing Boundaries

If a rule says:

```text
age >= 18
```

test:

```text
17
18
19
```

Boundary tests often reveal off-by-one errors.

````python
def is_adult(age: int) -> bool:
    return age >= 18

assert is_adult(17) is False
assert is_adult(18) is True
assert is_adult(19) is True

print("Boundary tests passed")
````

# 19. Negative Tests

A negative test verifies that invalid input is handled correctly.

````python
def withdraw(balance: float, amount: float) -> float:
    if amount <= 0:
        raise ValueError("Amount must be positive")
    if amount > balance:
        raise ValueError("Insufficient balance")
    return balance - amount

try:
    withdraw(1000.0, 1500.0)
except ValueError as error:
    print("Expected error:", error)
````

# 23. Parameterized Testing Concept

When several inputs should produce predictable outputs, parameterization avoids
duplicating the test structure.

In pytest, `@pytest.mark.parametrize` is commonly used.

Example structure:

```python
@pytest.mark.parametrize(
    "input_value, expected",
    [(1, 2), (2, 4), (3, 6)],
)
def test_double(input_value, expected):
    assert double(input_value) == expected
```

This feature requires pytest.

# 25. Mocking Concept

Sometimes a function depends on an external service:

```text
APPLICATION
   ↓
API / DATABASE
```

A unit test may replace the external dependency with a controlled mock:

```text
APPLICATION
   ↓
MOCK
```

Python's standard library provides `unittest.mock`.

Use mocks when they improve isolation and test speed; do not mock everything.

````python
from unittest.mock import Mock

payment_service = Mock()
payment_service.charge.return_value = True

result = payment_service.charge(1000)

assert result is True
payment_service.charge.assert_called_once_with(1000)

print("Mock test passed")
````

# 26. Testing With Dependencies

Dependency injection makes code easier to test.

Instead of hard-coding a service inside a function, pass it in.

````python
class TaxService:
    def rate(self) -> float:
        return 0.18

def final_price(
    amount: float,
    tax_service: TaxService,
) -> float:
    return amount * (1 + tax_service.rate())

service = TaxService()

assert final_price(1000.0, service) == 1180.0

print("Dependency test passed")
````

# 28. Regression Testing

A regression occurs when a previously working feature stops working after a
change.

```text
OLD CODE
 ↓
WORKING FEATURE
 ↓
CODE CHANGE
 ↓
BUG
```

A regression test captures the expected behavior so the same bug is less
likely to return unnoticed.

# 29. Test Coverage

Coverage measures which parts of the code are exercised by tests.

High coverage does **not** automatically mean high-quality tests.

Good testing asks:

```text
WHAT BEHAVIOR IS IMPORTANT?
        ↓
WHAT CAN FAIL?
        ↓
WHAT SHOULD BE VERIFIED?
```

Coverage is a useful signal, not the sole quality metric.

# 32. Common Testing Mistakes

### 1. Testing only the happy path

Include invalid and boundary cases.

### 2. Tests that depend on each other

A unit test should generally be independently runnable.

### 3. Testing implementation instead of behavior

Prefer stable observable behavior.

### 4. Huge test functions

Keep tests focused.

### 5. Random, uncontrolled data

Use deterministic inputs when possible.

### 6. Ignoring failed tests

A failing test needs investigation.

### 7. Overusing mocks

Mocks should isolate meaningful external dependencies, not hide the system
under test.

# 33. Best Practices

- Write tests for important business behavior.
- Keep unit tests small and fast.
- Use descriptive test names.
- Follow Arrange → Act → Assert.
- Test normal and edge cases.
- Test failure behavior.
- Keep tests deterministic.
- Avoid unnecessary shared state.
- Separate unit and integration tests when useful.
- Use `unittest` when standard-library testing is appropriate.
- Use pytest when its ecosystem/features benefit the project.
- Run tests before committing important changes.
- Automate tests in CI.
- Treat test failures as useful engineering feedback.

# 🚀 Project 1 — 01 — Calculator Unit Tests

## Business Problem

Create a small calculator and test its core operations.

## Testing Architecture

```text
BUSINESS LOGIC
 ↓
TEST INPUT
 ↓
EXECUTE
 ↓
EXPECTED vs ACTUAL
 ↓
ASSERT
 ↓
PASS / FAIL
```

## Implementation

````python
import unittest

def add(a: float, b: float) -> float:
    return a + b

def multiply(a: float, b: float) -> float:
    return a * b

def divide(a: float, b: float) -> float:
    if b == 0:
        raise ValueError("Cannot divide by zero")
    return a / b

class TestCalculator(unittest.TestCase):
    def test_add(self):
        self.assertEqual(add(2, 3), 5)

    def test_multiply(self):
        self.assertEqual(multiply(4, 5), 20)

    def test_divide(self):
        self.assertEqual(divide(10, 2), 5)

    def test_divide_by_zero(self):
        with self.assertRaises(ValueError):
            divide(10, 0)

suite = unittest.defaultTestLoader.loadTestsFromTestCase(TestCalculator)
result = unittest.TextTestRunner(verbosity=1).run(suite)

print("Successful:", result.wasSuccessful())
````

# 🚀 Project 4 — 04 — Password Validation Tests

## Business Problem

Test a password validation function against valid and invalid cases.

## Testing Architecture

```text
BUSINESS LOGIC
 ↓
TEST INPUT
 ↓
EXECUTE
 ↓
EXPECTED vs ACTUAL
 ↓
ASSERT
 ↓
PASS / FAIL
```

## Implementation

````python
def valid_password(password: str) -> bool:
    return (
        len(password) >= 8
        and any(character.isupper() for character in password)
        and any(character.islower() for character in password)
        and any(character.isdigit() for character in password)
    )

assert valid_password("Python123") is True
assert valid_password("python123") is False
assert valid_password("PYTHON123") is False
assert valid_password("Python") is False

print("Password validation tests passed")
````

# 🚀 Project 7 — 07 — Inventory Tests

## Business Problem

Test stock availability and inventory updates.

## Testing Architecture

```text
BUSINESS LOGIC
 ↓
TEST INPUT
 ↓
EXECUTE
 ↓
EXPECTED vs ACTUAL
 ↓
ASSERT
 ↓
PASS / FAIL
```

## Implementation

````python
class Inventory:
    def __init__(self, stock: dict[str, int]):
        self.stock = stock

    def available(self, product: str) -> bool:
        return self.stock.get(product, 0) > 0

    def remove(self, product: str, quantity: int) -> None:
        if quantity <= 0:
            raise ValueError("Quantity must be positive")
        if self.stock.get(product, 0) < quantity:
            raise ValueError("Insufficient stock")
        self.stock[product] -= quantity

inventory = Inventory({"Laptop": 5, "Mouse": 10})

assert inventory.available("Laptop") is True
inventory.remove("Laptop", 2)
assert inventory.stock["Laptop"] == 3

try:
    inventory.remove("Laptop", 10)
except ValueError:
    pass
else:
    raise AssertionError("Expected insufficient-stock error")

print("Inventory tests passed")
````

# 🚀 Project 9 — 09 — Analytics Pipeline Tests

## Business Problem

Test a multi-stage analytics pipeline using deterministic business data.

## Testing Architecture

```text
BUSINESS LOGIC
 ↓
TEST INPUT
 ↓
EXECUTE
 ↓
EXPECTED vs ACTUAL
 ↓
ASSERT
 ↓
PASS / FAIL
```

## Implementation

````python
def clean_sales(values: list[float]) -> list[float]:
    return [value for value in values if value >= 0]

def apply_tax(values: list[float], rate: float) -> list[float]:
    return [value * (1 + rate) for value in values]

def build_report(values: list[float]) -> dict[str, float | int]:
    if not values:
        return {"count": 0, "total": 0.0, "average": 0.0}

    total = sum(values)
    return {
        "count": len(values),
        "total": total,
        "average": total / len(values),
    }

raw = [1000.0, 2000.0, -500.0]

clean = clean_sales(raw)
assert clean == [1000.0, 2000.0]

taxed = apply_tax(clean, 0.18)
assert taxed == [1180.0, 2360.0]

report = build_report(taxed)
assert report["count"] == 2
assert report["total"] == 3540.0
assert report["average"] == 1770.0

print("Analytics pipeline tests passed")
````

# 🚀 Project 10 — 10 — Production-Style Test Suite

## Business Problem

Build a compact test suite for a sales service using validation, unit tests, edge cases, and mocked notification behavior.

## Testing Architecture

```text
BUSINESS LOGIC
 ↓
TEST INPUT
 ↓
EXECUTE
 ↓
EXPECTED vs ACTUAL
 ↓
ASSERT
 ↓
PASS / FAIL
```

## Implementation

````python
import unittest
from unittest.mock import Mock

class SalesService:
    def __init__(self, notifier):
        self.notifier = notifier

    def create_sale(
        self,
        customer: str,
        amount: float,
    ) -> dict[str, str | float]:
        if not customer.strip():
            raise ValueError("Customer is required")
        if amount <= 0:
            raise ValueError("Amount must be positive")

        sale = {
            "customer": customer.strip(),
            "amount": amount,
        }

        self.notifier.send(sale)
        return sale

class TestSalesService(unittest.TestCase):
    def setUp(self):
        self.notifier = Mock()
        self.service = SalesService(self.notifier)

    def test_create_sale(self):
        sale = self.service.create_sale(" Kaif ", 2500.0)

        self.assertEqual(sale["customer"], "Kaif")
        self.assertEqual(sale["amount"], 2500.0)
        self.notifier.send.assert_called_once_with(sale)

    def test_missing_customer(self):
        with self.assertRaises(ValueError):
            self.service.create_sale("   ", 2500.0)

    def test_invalid_amount(self):
        with self.assertRaises(ValueError):
            self.service.create_sale("Kaif", 0.0)

suite = unittest.defaultTestLoader.loadTestsFromTestCase(TestSalesService)
result = unittest.TextTestRunner(verbosity=1).run(suite)

print("Successful:", result.wasSuccessful())
````

| `unittest` | Python standard-library testing framework |

## Most Used Patterns

### Plain assertion

```python
assert add(2, 3) == 5
```

### unittest

```python
import unittest

class TestMath(unittest.TestCase):
    def test_add(self):
        self.assertEqual(add(2, 3), 5)
```

### Expected exception

```python
with self.assertRaises(ValueError):
    divide(10, 0)
```

### pytest

```python
def test_add():
    assert add(2, 3) == 5
```

### Mock

```python
from unittest.mock import Mock

service = Mock()
service.send("data")
service.send.assert_called_once_with("data")
```
