# 1. Why Virtual Environments?

A virtual environment creates an isolated Python environment for a project.

Without isolation, different projects may require different versions of the
same package.

```text
Project A
 ├── Python dependencies
 └── isolated environment

Project B
 ├── different dependencies
 └── separate environment
```

### Main benefits

- dependency isolation
- reproducibility
- cleaner projects
- fewer version conflicts
- safer experimentation

# 4. Create `.venv` from Python

You can create an environment programmatically for learning or automation.

````python
import subprocess
import sys
from pathlib import Path

environment = Path(".venv-demo")

subprocess.run(
    [sys.executable, "-m", "venv", str(environment)],
    check=True,
)

print("Created:", environment.resolve())
````

# 9. What Is `pip`?

`pip` is the standard Python package installer.

Common operations:

```text
pip install package
pip uninstall package
pip list
pip show package
pip freeze
```

A reliable practice inside a project environment is to invoke pip through the
active Python interpreter:

```text
python -m pip ...
```

This reduces confusion when multiple Python installations exist.

# 10. Upgrade pip

Inside the intended environment:

```text
python -m pip install --upgrade pip
```

Check the installed version:

```text
python -m pip --version
```

# 11. Install a Package

`pip freeze` outputs installed packages in requirements-style format.

Example:


# 16. `requirements.txt`

A `requirements.txt` file records project dependencies.

Example:

```text

pandas==2.2.3
```

The versions shown above are examples only. Use versions appropriate to the
project and verify compatibility before pinning them.

For larger projects, consider whether the complete environment or only direct

management strategy.

# 19. Package Versions

Check one package:


Check all installed packages:

Exact version constraints matter because different releases can change APIs,

# 20. Version Specifiers

Dependency requirements can express version constraints.

Examples:

```text
package==1.2.3
package>=1.2
package<2
package>=1.2,<2
```

Use constraints deliberately. Exact pins can improve reproducibility, while
ranges can allow compatible updates.

# 21. Dependency Management

A dependency is external software your project relies on.

```text
YOUR APPLICATION
      ↓
DIRECT DEPENDENCIES
      ↓
TRANSITIVE DEPENDENCIES
      ↓
PYTHON ENVIRONMENT
```

A production project should know:

- what it depends on
- why it depends on it
- which versions are supported
- how dependencies are installed
- how environments are reproduced

# 22. Direct vs Transitive Dependencies

### Direct dependency

Your project explicitly uses it.

Example:

```text
pandas
```

### Transitive dependency

A package you install may depend on another package.

```text
Your project
   ↓
Package A
   ↓
Package B
```

Package B is then a transitive dependency of your project.

# 30. `pyproject.toml`

`pyproject.toml` is the modern standardized configuration file for Python
projects.

It can hold project/build configuration and tool configuration.

A minimal packaging-oriented example:

```toml
[build-system]
requires = ["setuptools>=61"]
build-backend = "setuptools.build_meta"

[project]
name = "sales-analytics"
version = "0.1.0"
description = "Sales analytics project"
requires-python = ">=3.11"
dependencies = [
    "pandas>=2.0",
]
```

Exact build configuration depends on the packaging backend you choose.

# 31. `pyproject.toml` vs `requirements.txt`

They can serve different purposes.

| File | Typical purpose |
|---|---|
| `pyproject.toml` | Project metadata, build configuration, and declared project dependencies/tool configuration |
| `requirements.txt` | Installable dependency set, commonly used for environment reproduction/deployment workflows |

Do not assume that one universally replaces the other; the appropriate setup
depends on the project's packaging and deployment workflow.

# 35. Troubleshooting — Wrong Python

### Problem

A package is installed, but Python says:

```text
ModuleNotFoundError
```

### Common reason

`pip` installed the package into a different Python environment.

### Check

```text
python -c "import sys; print(sys.executable)"
python -m pip --version
```

The paths should correspond to the intended environment.

# 36. Troubleshooting — Package Version Conflict

### Problem

Two packages require incompatible versions of another dependency.

### Why it happens

Dependency constraints cannot all be satisfied by the selected versions.

### Fix

- inspect dependency requirements
- use a fresh virtual environment
- choose compatible package versions
- test the project
- document the working dependency set

# 37. Troubleshooting — Activation Problems

### Windows

Verify the command matches the shell:

```text
.venv\Scripts\activate
```

or:

```text
.venv\Scripts\Activate.ps1
```

### macOS / Linux

```text
source .venv/bin/activate
```

If activation is inconvenient, you can also invoke the environment's Python
executable directly.

# 38. Troubleshooting — Broken Environment

When an environment becomes confusing or corrupted, recreating it is often
simpler than manually repairing many packages.

```text
delete .venv
create a new .venv
install dependencies
run tests
```

Do not delete a production environment blindly; follow the deployment system's
process there.

# 39. Security Basics

Package management is part of software supply-chain security.

Best practices:

- install from trusted indexes
- review dependencies before adding them
- keep dependencies maintained
- avoid unnecessary packages
- pin or constrain versions appropriately
- review package ownership and source
- test dependency upgrades
- do not run untrusted installation scripts blindly
- protect private package credentials
- keep secrets out of `requirements.txt` and source control

# 40. Best Practices

### Virtual environments

- create one environment per project
- use a predictable name such as `.venv`
- do not commit `.venv`
- verify the active interpreter

### pip

- prefer `python -m pip`
- inspect versions
- use dependency files/configuration
- test upgrades

### Dependencies

- keep dependencies intentional
- avoid unnecessary packages
- document supported Python versions
- use reproducible installation workflows
- review dependency changes before production deployment

````python
import subprocess
import sys

result = subprocess.run(
    [sys.executable, "-m", "pip", "list", "--format=freeze"],
    check=True,
    capture_output=True,
    text=True,
)

lines = result.stdout.splitlines()

print("Installed dependency count:", len(lines))
print("\nFirst dependencies:")

for line in lines[:10]:
    print(line)
````

# 🚀 Project 6 — 06 — Reproducible Environment Builder

## Business Problem

Build a fresh environment and install dependencies from a requirements file.

```text
 ↓
```

# 🚀 Project 9 — 09 — Dependency Audit Workflow

## Business Problem

Create a repeatable workflow for checking dependency state before release.

## Architecture

```text
PROJECT
 ↓
.venv
 ↓
PYTHON + PIP
 ↓
DEPENDENCIES
 ↓
CONFIGURATION
 ↓
VERIFICATION
 ↓
REPRODUCIBLE WORKFLOW
```

## Implementation

````python
import subprocess
import sys

commands = [
    ["--version"],
    ["list"],
    ["check"],
]

for command in commands:
    print("\n>>> python -m pip", *command)
    result = subprocess.run(
        [sys.executable, "-m", "pip", *command],
        capture_output=True,
        text=True,
    )

    print(result.stdout)

    if result.stderr:
        print(result.stderr)

    print("Return code:", result.returncode)
````

# 🚀 Project 10 — 10 — Production-Style Python Environment Workflow

## Business Problem

Create a complete project environment, dependency specification, source layout, verification step, and reproducibility workflow.

## Architecture

```text
PROJECT
 ↓
.venv
 ↓
PYTHON + PIP
 ↓
DEPENDENCIES
 ↓
CONFIGURATION
 ↓
VERIFICATION
 ↓
REPRODUCIBLE WORKFLOW
```

## Implementation

````python
from pathlib import Path
import subprocess
import sys

root = Path("project_10")
package = root / "src" / "sales_platform"

package.mkdir(parents=True, exist_ok=True)

(package / "__init__.py").write_text(
    '__version__ = "1.0.0"\n',
    encoding="utf-8",
)

(package / "analytics.py").write_text(
    "def total_sales(values: list[float]) -> float:\n"
    "    return sum(values)\n",
    encoding="utf-8",
)

(root / "pyproject.toml").write_text(
    '[build-system]\n'
    'requires = ["setuptools>=61"]\n'
    'build-backend = "setuptools.build_meta"\n\n'
    '[project]\n'
    'name = "sales-platform"\n'
    'version = "1.0.0"\n'
    'description = "Production-style analytics environment demo"\n'
    'requires-python = ">=3.11"\n',
    encoding="utf-8",
)

(root / "requirements.txt").write_text(
    "requests>=2.0,<3\n",
    encoding="utf-8",
)

(root / ".gitignore").write_text(
    ".venv/\n__pycache__/\n*.pyc\n",
    encoding="utf-8",
)

venv = root / ".venv"

subprocess.run(
    [sys.executable, "-m", "venv", str(venv)],
    check=True,
)

if sys.platform.startswith("win"):
    python_executable = venv / "Scripts" / "python.exe"
else:
    python_executable = venv / "bin" / "python"

subprocess.run(
    [str(python_executable), "-m", "pip", "install", "-r", str(root / "requirements.txt")],
    check=True,
)

subprocess.run(
    [str(python_executable), "-m", "pip", "install", "-e", str(root)],
    check=True,
)

subprocess.run(
    [str(python_executable), "-m", "pip", "check"],
    check=True,
)

print("Project environment is ready.")
print("Python:", python_executable.resolve())
print("Project:", root.resolve())
````
