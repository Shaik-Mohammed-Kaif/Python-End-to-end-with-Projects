# 1. What Is a List?

A Python `list` is an ordered, mutable collection.

### Example

```python
sales = [100, 200, 150]
```

A list can contain multiple values:

```python
numbers = [10, 20, 30]
names = ["Kaif", "Ali", "Sara"]
mixed = [10, "Python", 3.14, True]
```

### Important Properties

- Ordered
- Mutable
- Allows duplicate values
- Supports indexing
- Supports slicing
- Can contain different Python object types

````python
sales = [100, 200, 150]

print(sales)
print(type(sales))
````

# 2. Creating Lists

The most common way is with square brackets:

```python
numbers = [10, 20, 30]
```

### Empty List

```python
items = []
```

### From an Iterable

```python
letters = list("Python")
```

Result:

```text
['P', 'y', 't', 'h', 'o', 'n']
```

````python
numbers = [10, 20, 30]
empty_list = []
letters = list("Python")

print(numbers)
print(empty_list)
print(letters)
````

# 3. List Indexing

List indexing starts at `0`.

```text
Index:   0       1       2       3
Value:  "A"     "B"     "C"     "D"
```

Example:

```python
names = ["Kaif", "Ali", "Sara", "Omar"]

names[0]   # Kaif
names[2]   # Sara
```

If an index is outside the valid range, Python raises `IndexError`.

````python
names = ["Kaif", "Ali", "Sara", "Omar"]

print(names[0])
print(names[2])
print(names[3])
````

# 4. Negative Indexing

Negative indexes count from the end.

```text
Value:  "A"    "B"    "C"    "D"
Index:   0      1      2      3
        -4     -3     -2     -1
```

Therefore:

```python
names[-1]
```

returns the last item.

````python
names = ["Kaif", "Ali", "Sara", "Omar"]

print(names[-1])
print(names[-2])
print(names[-4])
````

# 5. List Slicing

Slicing extracts a portion of a list.

### Syntax

```python
list[start:stop:step]
```

The `stop` index is excluded.

Example:

```python
numbers = [10, 20, 30, 40, 50]

numbers[1:4]
```

Result:

```text
[20, 30, 40]
```

### Useful Slices

```python
numbers[:3]     # first three
numbers[2:]     # from index 2
numbers[::2]    # every second item
numbers[::-1]   # reversed copy
```

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[1:4])
print(numbers[:3])
print(numbers[2:])
print(numbers[::2])
print(numbers[::-1])
````

# 6. Updating List Elements

Lists are mutable, so an existing element can be changed.

```python
prices = [100, 200, 300]

prices[1] = 250
```

Now:

```text
[100, 250, 300]
```

You can also update a slice.

````python
prices = [100, 200, 300]

prices[1] = 250

print(prices)

prices[0:2] = [110, 260]

print(prices)
````

# 7. Adding Items — `append()`

`append()` adds one item to the end of the list.

### Syntax

```python
list.append(item)
```

Example:

````python
shopping = ["Milk", "Bread"]

shopping.append("Eggs")

print(shopping)
````

# 8. Adding Items — `extend()`

`extend()` adds all items from an iterable.

```python
numbers = [1, 2]

numbers.extend([3, 4, 5])
```

Result:

```text
[1, 2, 3, 4, 5]
```

### Difference

```python
append([3, 4])
```

adds the list as one item.

```python
extend([3, 4])
```

adds `3` and `4` separately.

````python
items = [1, 2]

items.append([3, 4])

print("After append:", items)

items = [1, 2]

items.extend([3, 4])

print("After extend:", items)
````

# 9. Adding Items — `insert()`

`insert()` adds an item at a specific position.

### Syntax

```python
list.insert(index, item)
```

Example:

````python
names = ["Ali", "Sara"]

names.insert(1, "Kaif")

print(names)
````

# 10. Removing Items — `remove()`

`remove()` removes the first matching value.

```python
items = ["A", "B", "C", "B"]

items.remove("B")
```

Result:

```text
["A", "C", "B"]
```

If the value does not exist, `remove()` raises `ValueError`.

````python
items = ["A", "B", "C", "B"]

items.remove("B")

print(items)
````

# 11. Removing Items — `pop()`

`pop()` removes and returns an item.

### Last item

```python
items.pop()
```

### Specific index

```python
items.pop(1)
```

This is useful when you need both removal and the removed value.

````python
items = ["A", "B", "C"]

removed_item = items.pop()

print("Removed:", removed_item)
print("Remaining:", items)

removed_item = items.pop(0)

print("Removed:", removed_item)
print("Remaining:", items)
````

# 12. Removing All Items — `clear()`

`clear()` removes all items from the list.

```python
items = [1, 2, 3]

items.clear()
```

Result:

```text
[]
```

````python
items = [1, 2, 3, 4]

items.clear()

print(items)
````

# 13. Searching — `in`

The `in` operator checks membership.

```python
products = ["Laptop", "Mouse", "Keyboard"]

print("Mouse" in products)
```

Output:

```text
True
```

Use `not in` to check absence.

````python
products = ["Laptop", "Mouse", "Keyboard"]

print("Mouse" in products)
print("Phone" in products)
print("Phone" not in products)
````

# 14. Searching — `index()`

`index()` returns the position of the first matching value.

```python
names = ["Ali", "Kaif", "Sara"]

names.index("Kaif")
```

Result:

```text
1
```

If the value does not exist, Python raises `ValueError`.

Use membership checking first when absence is possible.

````python
names = ["Ali", "Kaif", "Sara"]

if "Kaif" in names:
    print("Index:", names.index("Kaif"))
else:
    print("Name not found")
````

# 15. Counting — `count()`

`count()` returns the number of occurrences of a value.

```python
numbers = [10, 20, 10, 30, 10]

print(numbers.count(10))
```

Output:

```text
3
```

````python
numbers = [10, 20, 10, 30, 10]

print("10 appears:", numbers.count(10), "times")
````

# 16. Sorting — `sort()`

`sort()` sorts the existing list in place.

### Ascending

```python
numbers.sort()
```

### Descending

```python
numbers.sort(reverse=True)
```

The method changes the original list.

````python
numbers = [50, 10, 40, 20, 30]

numbers.sort()

print("Ascending:", numbers)

numbers.sort(reverse=True)

print("Descending:", numbers)
````

# 17. `sorted()`

`sorted()` returns a new sorted list and leaves the original unchanged.

```python
numbers = [30, 10, 20]

new_numbers = sorted(numbers)
```

### Difference

```text
sort()
  → modifies the original list

sorted()
  → returns a new sorted list
```

````python
numbers = [30, 10, 20]

sorted_numbers = sorted(numbers)

print("Original:", numbers)
print("Sorted:", sorted_numbers)
````

# 18. Reversing — `reverse()`

`reverse()` reverses the list in place.

```python

numbers.reverse()
```

Result:

```text
[4, 3, 2, 1]
```

````python
numbers = [1, 2, 3, 4]

numbers.reverse()

print(numbers)
````

# 19. Copying — `copy()`

`copy()` creates a shallow copy of a list.

```python
original = [10, 20, 30]

backup = original.copy()
```

Now the two list objects are separate at the outer level.

### Important

Avoid:

```python
backup = original
```

when you intend to create a separate list, because both variables then
refer to the same list object.

````python
original = [10, 20, 30]

backup = original.copy()

backup.append(40)

print("Original:", original)
print("Backup:", backup)
````

# 26. Common List Mistakes

### Mistake 1 — Index out of range

```python
items = [10, 20]

items[5]
```

There is no index `5`.

### Mistake 2 — `append()` vs `extend()`

```python
items.append([3, 4])
```

adds one nested list.

```python
items.extend([3, 4])
```

adds two elements.

### Mistake 3 — Assuming `sort()` returns the sorted list

`sort()` modifies the list in place and returns `None`.

### Mistake 4 — Removing a missing value

`remove()` raises `ValueError` when the value is absent.

### Mistake 5 — Aliasing

```python
backup = original
```

does not create an independent list.

# 🚀 Project 01 — Student Marks Analyzer

## Problem

Create a program that stores student marks in a list and calculates:

- Number of subjects
- Total marks
- Average
- Highest mark
- Lowest mark
- Sorted marks
- Pass/fail status

### Rule

A student passes when every subject mark is at least `40`.

### Skills

- `list`
- `len()`
- `sum()`
- `min()`
- `max()`
- `sorted()`
- `for`
- `all()`

````python
# Project 01 — Student Marks Analyzer

marks = [78, 85, 67, 92, 74]

total = sum(marks)
average = total / len(marks)
highest = max(marks)
lowest = min(marks)
sorted_marks = sorted(marks)

passed = all(mark >= 40 for mark in marks)

print("Marks:", marks)
print("Number of Subjects:", len(marks))
print("Total:", total)
print(f"Average: {average:.2f}")
print("Highest:", highest)
print("Lowest:", lowest)
print("Sorted:", sorted_marks)
print("Result:", "Pass" if passed else "Fail")
````

# 🚀 Project 02 — Shopping Cart

## Problem

Build a simple shopping cart using lists.

The program should:

- Add products
- Remove a product
- Display the cart
- Count items
- Check whether an item exists
- Sort the cart
- Clear the cart

This demonstrates practical list mutation.

````python
# Project 02 — Shopping Cart

cart = []

cart.append("Laptop")
cart.append("Mouse")
cart.append("Keyboard")

print("Initial Cart:", cart)

if "Mouse" in cart:
    cart.remove("Mouse")

cart.insert(1, "Monitor")

print("Updated Cart:", cart)
print("Item Count:", len(cart))

cart.sort()

print("Sorted Cart:", cart)

print("Keyboard available:", "Keyboard" in cart)

cart.clear()

print("After Clear:", cart)
````

# 🚀 Project 03 — To-Do List

## Problem

Create a basic to-do list.

Operations:

- Add tasks
- Complete/remove a task
- Display tasks
- Search for a task
- Sort tasks

Use list methods instead of manually rebuilding the list.

````python
# Project 03 — To-Do List

tasks = [
    "Study Python",
    "Practice SQL",
    "Build Project"
]

print("Current Tasks:")

for task in tasks:
    print("-", task)

new_task = input("\nEnter a new task: ").strip()

if new_task:
    tasks.append(new_task)

remove_task = input("Enter a task to remove (or press Enter): ").strip()

if remove_task:
    if remove_task in tasks:
        tasks.remove(remove_task)
        print("Task removed.")
    else:
        print("Task not found.")

tasks.sort()

print("\nFinal To-Do List:")

for index, task in enumerate(tasks, start=1):
    print(f"{index}. {task}")
````

# 🚀 Project 04 — Expense Tracker

## Problem

Store expenses in a list and calculate:

- Total expense
- Average expense
- Highest expense
- Lowest expense
- Number of transactions
- Sorted expenses

This simulates a basic financial-analysis workflow.

````python
# Project 04 — Expense Tracker

expenses = [250, 120, 800, 450, 300]

total = sum(expenses)
average = total / len(expenses)
highest = max(expenses)
lowest = min(expenses)

print("Expenses:", expenses)
print(f"Transactions: {len(expenses)}")
print(f"Total Expense: ₹{total:.2f}")
print(f"Average Expense: ₹{average:.2f}")
print(f"Highest Expense: ₹{highest:.2f}")
print(f"Lowest Expense: ₹{lowest:.2f}")

expenses.sort(reverse=True)

print("Expenses High → Low:", expenses)
````

# 🚀 Project 05 — Inventory System

## Problem

Create a basic inventory system using lists.

Requirements:

- Store products
- Add a product
- Search for a product
- Remove a product
- Display inventory
- Sort inventory
- Count inventory items

This project demonstrates how list operations can model a simple
business workflow.

````python
# Project 05 — Inventory System

inventory = [
    "Laptop",
    "Mouse",
    "Keyboard",
    "Monitor"
]

print("=" * 50)
print("             INVENTORY SYSTEM")
print("=" * 50)

print("\nCurrent Inventory:")

for product in inventory:
    print("-", product)

new_product = input("\nEnter product to add: ").strip()

if new_product:
    if new_product not in inventory:
        inventory.append(new_product)
        print("Product added.")
    else:
        print("Product already exists.")

search_product = input("Enter product to search: ").strip()

if search_product in inventory:
    print(f"{search_product} is available.")
    print(f"Position: {inventory.index(search_product)}")
else:
    print(f"{search_product} is not available.")

remove_product = input("Enter product to remove (or press Enter): ").strip()

if remove_product:
    if remove_product in inventory:
        inventory.remove(remove_product)
        print("Product removed.")
    else:
        print("Product not found.")

inventory.sort()

print("\nFinal Inventory:")

for index, product in enumerate(inventory, start=1):
    print(f"{index}. {product}")

print(f"\nTotal Products: {len(inventory)}")
````
