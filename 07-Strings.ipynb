# 1. What Is a String?

A **string** is a sequence of characters enclosed inside quotes.

Characters can include:

- Letters
- Numbers
- Spaces
- Symbols
- Special characters

### Example

```python
name = "Mohammed Kaif"

print(name)
````

### Output

```text
Mohammed Kaif
```

The variable `name` stores a string value.

---

## Checking the Data Type

Python provides the `type()` function to check the type of a value.

```python
name = "Mohammed Kaif"

print(type(name))
```

### Output

```text
<class 'str'>
```

`str` means **string**.

# 6. String Indexing

A string is a sequence of characters.

Each character has a position called an **index**.

Python uses **zero-based indexing**.

Example:

```python
word = "Python"
```

The positions are:

```text
Character:  P   y   t   h   o   n
Index:      0   1   2   3   4   5
```

Therefore:

```python
word = "Python"

print(word[0])
print(word[1])
print(word[2])
```

### Output

```text
P
y
t
```

---

## Accessing Individual Characters

```python
language = "Python"

print(language[0])
print(language[3])
print(language[5])
```

### Output

```text
P
h
n
```

---

# 7. Negative Indexing

Python also supports negative indexing.

Negative indexes start from the end of the string.

```text
Character:       P    y    t    h    o    n
Positive Index:  0    1    2    3    4    5
Negative Index: -6   -5   -4   -3   -2   -1
```

### Example

```python
word = "Python"

print(word[-1])
print(word[-2])
print(word[-6])
```

### Output

```text
n
o
P
```

The most useful negative index is often:

```python
text[-1]
```

which returns the **last character**.

# 8. String Slicing

Slicing allows you to extract part of a string.

### Syntax

```python
string[start:stop]
```

Important:

The `stop` index is **excluded**.

### Example

```python
word = "Python"

print(word[0:3])
```

### Output

```text
Pyt
```

Indexes `0`, `1`, and `2` are included.

Index `3` is excluded.

---

## Slicing From the Beginning

```python
word = "Python"

print(word[:3])
```

### Output

```text
Pyt
```

---

## Slicing Until the End

```python
word = "Python"

print(word[2:])
```

### Output

```text
thon
```

---

## Slicing With a Step

### Syntax

```python
string[start:stop:step]
```

Example:

```python
word = "Python"

print(word[0:6:2])
```

### Output

```text
Pto
```

---

## Reverse a String

A common Python technique:

```python
word = "Python"

print(word[::-1])
```

### Output

```text
nohtyP
```

# 9. Strings Are Immutable

Strings in Python are **immutable**.

That means you cannot directly change an individual character.

This will cause an error:

```python
word = "Python"

word[0] = "J"
```

Strings cannot be modified character-by-character.

Instead, create a new string.

```python
word = "Python"

word = "J" + word[1:]

print(word)
```

### Output

```text
Jython
```

Many string methods also return a **new string** rather than modifying the original.

# 12. Case Conversion Methods

## `lower()`

Converts characters to lowercase.

```python
text = "PYTHON PROGRAMMING"

result = text.lower()

print(result)
```

### Output

```text
python programming
```

### Real-World Use

Useful when comparing text without worrying about capitalization.

```python
email = "USER@GMAIL.COM"

if email.lower() == "user@gmail.com":
    print("Email matched")
```

### Output

```text
Email matched
```

---

## `upper()`

Converts characters to uppercase.

```python
text = "python"

print(text.upper())
```

### Output

```text
PYTHON
```

---

## `title()`

Converts the first character of each word to uppercase.

```python
name = "mohammed kaif"

print(name.title())
```

### Output

```text
Mohammed Kaif
```

---

## `capitalize()`

Capitalizes the first character of the string.

```python
text = "python programming"

print(text.capitalize())
```

### Output

```text
Python programming
```

### Difference

```text
title()       → Capitalizes each word
capitalize()  → Capitalizes the first character of the string
```

# 17. Search Methods

## `find()`

Returns the index of the first occurrence of a substring.

```python
text = "Python Programming"

position = text.find("Programming")

print(position)
```

### Output

```text
7
```

If the substring is not found:

```python
text = "Python"

print(text.find("Java"))
```

### Output

```text
-1
```

This makes `find()` useful when you want a non-error result for a missing substring.

---

## `index()`

`index()` also searches for a substring.

```python
text = "Python Programming"

print(text.index("Programming"))
```

### Output

```text
7
```

### Important Difference

If the substring does not exist, `index()` raises a `ValueError`.

```python
text = "Python"

print(text.index("Java"))
```

This produces an error.

### Comparison

| Method    | Found | Not Found    |
| --------- | ----- | ------------ |
| `find()`  | Index | `-1`         |
| `index()` | Index | `ValueError` |

# 23. String Formatting

String formatting allows you to insert values into text.

Python provides multiple approaches.

Modern Python code generally favors **f-strings** for straightforward formatting.

---

## Basic Concatenation

```python
name = "Kaif"

message = "Hello " + name

print(message)
```

### Output

```text
Hello Kaif
```

This works, but f-strings are usually cleaner when many values are involved.

---

# 24. f-Strings

An f-string is created by placing `f` before the string.

### Example

```python
name = "Mohammed"
age = 22

message = f"My name is {name} and I am {age} years old."

print(message)
```

### Output

```text
My name is Mohammed and I am 22 years old.
```

---

## Expressions Inside f-Strings

Python expressions can be placed inside `{}`.

```python
price = 100
quantity = 3

print(f"Total = {price * quantity}")
```

### Output

```text
Total = 300
```

---

## Formatting Numbers

```python
price = 1250.5678

print(f"Price: ₹{price:.2f}")
```

### Output

```text
Price: ₹1250.57
```

`.2f` means display two digits after the decimal point.

---

## Real-World Example — Customer Information

```python
name = "Mohammed Kaif"
plan = "Premium"
monthly_charge = 999.50

message = (
    f"Customer: {name}\n"
    f"Plan: {plan}\n"
    f"Monthly Charge: ₹{monthly_charge:.2f}"
)

print(message)
```

### Output

```text
Customer: Mohammed Kaif
Plan: Premium
Monthly Charge: ₹999.50
```

# 29. Important String Rules

### Rule 1 — Strings Use Quotes

```python
name = "Kaif"
```

### Rule 2 — Indexing Starts at Zero

```text
Python
012345
```

### Rule 3 — Negative Indexing Starts From `-1`

```text
Python
-6 -5 -4 -3 -2 -1
```

### Rule 4 — Slice Stop Is Excluded

```python
text[0:3]
```

returns indexes:

```text
0, 1, 2
```

not index `3`.

### Rule 5 — Strings Are Immutable

You cannot directly change:

```python
text[0]
```

### Rule 6 — String Methods Usually Return a New String

```python
text = "python"

new_text = text.upper()

print(text)
print(new_text)
```

### Output

```text
python
PYTHON
```

print("=" * 65)

# ----------------------------------------------------------

username = input("Enter your username: ")

# ----------------------------------------------------------
# 2. Clean Input
# ----------------------------------------------------------

email = email.strip().lower()
username = username.strip().lower()

# ----------------------------------------------------------
# 3. Basic Email Analysis
# ----------------------------------------------------------

at_count = email.count("@")
dot_count = email.count(".")

has_at = "@" in email
has_dot = "." in email

# ----------------------------------------------------------
# 4. Extract Email Username and Domain
# ----------------------------------------------------------

if has_at:
    email_parts = email.split("@")

    email_username = email_parts[0]
    domain = "@".join(email_parts[1:])
else:
    email_username = ""
    domain = ""

# ----------------------------------------------------------
# 5. Domain Validation
# ----------------------------------------------------------

if has_dot:
    domain_parts = domain.split(".")

    domain_name = domain_parts[0]
    extension = domain_parts[-1]
else:
    domain_name = ""
    extension = ""

# ----------------------------------------------------------
# 6. Email Validation
# ----------------------------------------------------------

if (
    email
    and at_count == 1
    and email_username
    and domain
    and "." in domain
    and not email.startswith("@")
    and not email.endswith("@")
    and not domain.startswith(".")
    and not domain.endswith(".")
):
    email_status = "Valid Basic Format"
else:
    email_status = "Invalid Basic Format"

# ----------------------------------------------------------
# 7. Username Analysis
# ----------------------------------------------------------

username_length = len(username)

has_digit = any(character.isdigit() for character in username)

has_letter = any(character.isalpha() for character in username)

username_is_alphanumeric = username.replace("_", "").isalnum()

starts_with_letter = bool(username) and username[0].isalpha()

# ----------------------------------------------------------
# 8. Username Validation
# ----------------------------------------------------------

if (
    username
    and username_length >= 4
    and username_is_alphanumeric
    and starts_with_letter
):
    username_status = "Valid"
else:
    username_status = "Invalid"

# ----------------------------------------------------------
# 9. Generate Clean Username
# ----------------------------------------------------------

# 🔍 Project Explanation

## 1. `strip()`

Removes unnecessary whitespace from the beginning and end.

```python
text = text.strip()
````

---

## 2. `lower()`

Converts the complete text to lowercase.

```python
lower_text = text.lower()
```

This makes searching case-insensitive.

For example:

```text
Python
PYTHON
python
```

can all be normalized to:

```text
python
```

---

## 3. `replace()`

We remove selected punctuation characters.

```python
clean_text = (
    lower_text
    .replace(",", "")
    .replace(".", "")
    .replace("!", "")
    .replace("?", "")
)
```

---

## 4. `split()`

Converts the cleaned text into separate words.

```python
words = clean_text.split()
```

Example:

```text
"python is easy"

        ↓

["python", "is", "easy"]
```

---

## 5. `len()`

Used to calculate:

* Total characters
* Number of words
* Length of individual words

Example:

```python
len(text)
```

---

## 6. `count()`

Counts how many times text appears.

```python
clean_text.count(search_word)
```

Example:

```text
python python sql
```

Searching for:

```text
python
```

returns:

```text
2
```

---

## 7. `find()`

Returns the position of the first occurrence.

```python
clean_text.find(search_word)
```

If the text is not found, it returns:

```text
-1
```

---

## 8. `isalpha()`

Checks whether a character is alphabetic.

```python
character.isalpha()
```

---

## 9. `isdigit()`

Checks whether a character is a digit.

```python
character.isdigit()
```

---

## 10. `isspace()`

Checks whether a character is whitespace.

```python
character.isspace()
```

---

## 11. `startswith()`

Checks how the text begins.

```python
lower_text.startswith("python")
```

---

## 12. `endswith()`

Checks how the text ends.

```python
text.endswith(".")
```

---

## 13. Finding the Longest Word

```python
longest_word = max(words, key=len)
```

Python compares the lengths of the words and returns the longest one.

---

## 14. Finding the Shortest Word

```python
shortest_word = min(words, key=len)
```

Python compares word lengths and returns the shortest one.

---

## 15. Counting Characters With a Loop

This pattern:

```python
digit_count = sum(
    character.isdigit()
    for character in text
)
```

checks every character.

Conceptually:

```text
Character
    ↓
isdigit()
    ↓
True / False
    ↓
Count True values
```
