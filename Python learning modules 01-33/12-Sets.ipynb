# 1. What Is a Set?

A Python `set` is a mutable collection of **distinct hashable objects**.

### Main characteristics
- Duplicate values are automatically removed.
- Sets do not support sequence-style indexing or slicing.
- A set itself can be modified.
- Set elements must be hashable.
- Sets are especially useful for uniqueness, membership testing, and set relationships.

```python
skills = {"Python", "SQL", "Python", "Excel"}
```

Only one `"Python"` remains.

````python
skills = {"Python", "SQL", "Python", "Excel"}

print(skills)
print(type(skills))
print(len(skills))
````

# 2. Creating Sets

### Literal syntax

```python
numbers = {10, 20, 30}
```

### Empty set

```python
empty_set = set()
```

Important:

```python
{}
```

creates an empty **dictionary**, not an empty set.

### From an iterable

```python
letters = set("Python")
```

The resulting set contains the distinct characters.

````python
numbers = {10, 20, 30}
empty_set = set()
letters = set("Python")

print(numbers)
print(empty_set)
print(letters)
print(type(empty_set))
print(type({}))
````

# 3. Unique Values

Converting an iterable to a set is a simple way to remove duplicates.

```python
values = [10, 20, 10, 30, 20, 40]
unique_values = set(values)
```

Result:

```text
{10, 20, 30, 40}
```

````python
values = [10, 20, 10, 30, 20, 40]
unique_values = set(values)

print("Original:", values)
print("Unique:", unique_values)
````

# 4. `add()`

`add()` inserts one element.

```python
skills.add("Excel")
```

If the element is already present, the set remains unchanged.

````python
skills = {"Python", "SQL"}

skills.add("Excel")
skills.add("Python")

print(skills)
````

# 5. `update()`

`update()` adds elements from one or more iterables.

```python
skills.update(["Excel", "Power BI"])
```

### Difference

```text
add(x)
→ adds one element

update(iterable)
→ adds elements from an iterable
```

````python
skills = {"Python"}

skills.update(["SQL", "Excel", "Power BI"])

print(skills)
````

# 6. `remove()`

`remove(x)` removes a specified element.

If the element does not exist, it raises `KeyError`.

Use it when the element is expected to exist.

````python
skills = {"Python", "SQL", "Excel"}

skills.remove("SQL")

print(skills)
````

# 7. `discard()`

`discard(x)` removes an element if present.

If it is absent, nothing happens.

```text
remove()  → missing item = KeyError
discard() → missing item = no error
```

````python
skills = {"Python", "SQL", "Excel"}

skills.discard("Java")
skills.discard("SQL")

print(skills)
````

# 8. `pop()`

`pop()` removes and returns an **arbitrary** element.

Because sets do not provide positional indexing, do not use `pop()` when a particular item must be removed.

````python
skills = {"Python", "SQL", "Excel"}

removed = skills.pop()

print("Removed:", removed)
print("Remaining:", skills)
````

# 9. `clear()`

`clear()` removes all elements and leaves an empty set.

````python
skills = {"Python", "SQL", "Excel"}

skills.clear()

print(skills)
````

# 10. Membership Testing

Use:

```python
value in my_set
value not in my_set
```

Sets are particularly useful when the main operation is membership testing.

````python
skills = {"Python", "SQL", "Excel"}

print("Python" in skills)
print("Java" in skills)
print("Java" not in skills)
````

# 11. Iterating Through a Set

Use a `for` loop.

```python
for skill in skills:
    print(skill)
```

Do not write program logic that depends on an ordinary set having a particular iteration/display order.

````python
skills = {"Python", "SQL", "Excel"}

for skill in skills:
    print(skill)
````

# 12. `union()`

Returns a new set containing elements from both sets.

```python
a.union(b)
```

Example:

```text
A = Python, SQL, Excel
B = SQL, Power BI

Union = Python, SQL, Excel, Power BI
```

````python
a = {"Python", "SQL", "Excel"}
b = {"SQL", "Power BI"}

print(a.union(b))
````

# 13. Union Operator `|`

```python
a | b
```

is the operator form of union.

````python
a = {"Python", "SQL"}
b = {"SQL", "Excel"}

print(a | b)
print(a.union(b))
````

# 14. `intersection()`

Returns elements common to both sets.

```python
a.intersection(b)
```

For example, if A contains Python, SQL, Excel and B contains SQL, Excel, Power BI, the intersection is SQL and Excel.

````python
a = {"Python", "SQL", "Excel"}
b = {"SQL", "Excel", "Power BI"}

print(a.intersection(b))
````

# 15. Intersection Operator `&`

```python
a & b
```

is the operator form of intersection.

````python
a = {"Python", "SQL", "Excel"}
b = {"SQL", "Excel", "Power BI"}

print(a & b)
print(a.intersection(b))
````

# 16. `difference()`

Returns elements that are in the first set but not the second.

```python
a.difference(b)
```

Difference is directional:

```text
A - B
```

is generally different from:

```text
B - A
```

````python
a = {"Python", "SQL", "Excel"}
b = {"SQL", "Excel", "Power BI"}

print("A - B:", a.difference(b))
print("B - A:", b.difference(a))
````

# 17. Difference Operator `-`

```python
a - b
```

is the operator form of difference.

````python
a = {"Python", "SQL", "Excel"}
b = {"SQL", "Excel", "Power BI"}

print(a - b)
````

# 18. `symmetric_difference()`

Returns elements that occur in exactly one of the two sets.

```text
A = Python, SQL, Excel
B = SQL, Excel, Power BI

Result = Python, Power BI
```

````python
a = {"Python", "SQL", "Excel"}
b = {"SQL", "Excel", "Power BI"}

print(a.symmetric_difference(b))
````

# 19. Symmetric Difference Operator `^`

```python
a ^ b
```

is the operator form of symmetric difference.

````python
a = {"Python", "SQL", "Excel"}
b = {"SQL", "Excel", "Power BI"}

print(a ^ b)
print(a.symmetric_difference(b))
````

# 21. `issubset()`

Checks whether every element of one set is contained in another.

```python
required.issubset(skills)
```

````python
required = {"Python", "SQL"}
skills = {"Python", "SQL", "Excel", "Power BI"}

print(required.issubset(skills))
print(skills.issubset(required))
````

# 22. Subset Operators

```python
A <= B
```

means A is a subset of B.

```python
A < B
```

means A is a proper subset of B: A is contained in B and the sets are not equal.

````python
required = {"Python", "SQL"}
skills = {"Python", "SQL", "Excel"}

print(required <= skills)
print(required < skills)
print(skills <= skills)
print(skills < skills)
````

# 23. `issuperset()`

Checks whether a set contains every element of another set.

```python
skills.issuperset(required)
```

````python
skills = {"Python", "SQL", "Excel"}
required = {"Python", "SQL"}

print(skills.issuperset(required))
print(required.issuperset(skills))
````

# 24. Superset Operators

```python
A >= B
```

means A is a superset of B.

```python
A > B
```

means A is a proper superset of B.

````python
skills = {"Python", "SQL", "Excel"}
required = {"Python", "SQL"}

print(skills >= required)
print(skills > required)
print(skills >= skills)
print(skills > skills)
````

# 25. `isdisjoint()`

Checks whether two sets have no elements in common.

```python
a.isdisjoint(b)
```

Returns `True` when the sets share no elements.

````python
frontend = {"HTML", "CSS", "JavaScript"}
backend = {"Python", "SQL", "Java"}

print(frontend.isdisjoint(backend))

data = {"Python", "SQL"}
analytics = {"SQL", "Excel"}

print(data.isdisjoint(analytics))
````

# 26. In-Place Set Operations

These methods modify the existing set:

- `update()`
- `intersection_update()`
- `difference_update()`
- `symmetric_difference_update()`

Use them when you intentionally want to change the original set.

````python
a = {"Python", "SQL", "Excel"}
b = {"SQL", "Power BI"}

a.update(b)
print("update:", a)

a = {"Python", "SQL", "Excel"}
a.intersection_update(b)
print("intersection_update:", a)

a = {"Python", "SQL", "Excel"}
a.difference_update(b)
print("difference_update:", a)

a = {"Python", "SQL", "Excel"}
a.symmetric_difference_update(b)
print("symmetric_difference_update:", a)
````

# 27. `copy()`

`copy()` creates a shallow copy of the set.

```python
backup = original.copy()
```

Adding/removing an element from the copied set does not change the original set.

````python
original = {"Python", "SQL"}
backup = original.copy()

backup.add("Excel")

print("Original:", original)
print("Backup:", backup)
````

# 29. Hashable Elements

Set elements must be hashable.

Common valid examples:

```python
{1, "Python", (10, 20)}
```

A list cannot be a set element because a list is mutable and unhashable.

This is why:

```python
{[1, 2, 3]}
```

raises `TypeError`.

````python
valid = {1, "Python", (10, 20)}
print(valid)

try:
    invalid = {[1, 2, 3]}
except TypeError as error:
    print("Error:", error)
````

# 30. Set vs List vs Tuple

| Feature | Set | List | Tuple |
|---|---|---|---|
| Mutable | Yes | Yes | No |
| Duplicate values | No | Yes | Yes |
| Indexing | No | Yes | Yes |
| Slicing | No | Yes | Yes |
| Main strength | Unique values / relationships | Ordered mutable data | Fixed ordered data |
| Add one | `add()` | `append()` | Not supported |
| Union | Yes | No | No |
| Intersection | Yes | No | No |

### Simple Rule

```text
Unique values / set relationships → SET
Ordered changeable collection → LIST
Fixed ordered collection → TUPLE
```

# 🚀 Project 01 — Student Course Analyzer

## Problem

Two groups of students take Python and SQL.

Find:
- All unique students
- Students taking both
- Python-only students
- SQL-only students
- Students taking exactly one course
- Whether the groups are disjoint

````python
python_students = {"Kaif", "Ali", "Sara", "Omar"}
sql_students = {"Ali", "Sara", "Zoya", "Ravi"}

all_students = python_students | sql_students
both_courses = python_students & sql_students
python_only = python_students - sql_students
sql_only = sql_students - python_students
exactly_one = python_students ^ sql_students

print("All Students:", all_students)
print("Both Courses:", both_courses)
print("Python Only:", python_only)
print("SQL Only:", sql_only)
print("Exactly One Course:", exactly_one)
print("Groups Disjoint:", python_students.isdisjoint(sql_students))
````

# 🚀 Project 02 — Unique Shopping Products

Use sets to analyze products from two shopping sessions.

Find:
- Unique products
- Number of unique products
- Common products
- Products only in session 1
- Products only in session 2

````python
session_1 = {
    "Laptop",
    "Mouse",
    "Keyboard",
    "Monitor",
    "Mouse"
}

session_2 = {
    "Mouse",
    "Monitor",
    "Webcam",
    "Headset"
}

unique_products = session_1 | session_2
common_products = session_1 & session_2
session_1_only = session_1 - session_2
session_2_only = session_2 - session_1

print("Unique Products:", unique_products)
print("Unique Product Count:", len(unique_products))
print("Common Products:", common_products)
print("Session 1 Only:", session_1_only)
print("Session 2 Only:", session_2_only)
````

# 🚀 Project 03 — Skill Matching System

A job requires:

```text
Python, SQL, Statistics, Excel
```

A candidate has:

```text
Python, SQL, Excel, Power BI
```

Determine:
- Matching skills
- Missing skills
- Extra skills
- Whether every required skill is present

````python
required_skills = {
    "Python",
    "SQL",
    "Statistics",
    "Excel"
}

candidate_skills = {
    "Python",
    "SQL",
    "Excel",
    "Power BI"
}

matching = required_skills & candidate_skills
missing = required_skills - candidate_skills
extra = candidate_skills - required_skills
eligible = required_skills.issubset(candidate_skills)

print("Matching Skills:", matching)
print("Missing Skills:", missing)
print("Extra Skills:", extra)
print("Meets All Requirements:", eligible)
````

# 🚀 Project 04 — Inventory Category Analyzer

Analyze products shared across three inventory categories.

Find:
- All products
- Products common to all categories
- Products unique to each category
- Products shared by exactly two categories

````python
electronics = {"Laptop", "Mouse", "Keyboard", "Monitor"}
office = {"Notebook", "Keyboard", "Pen", "Monitor"}
accessories = {"Mouse", "Monitor", "Webcam", "Headset"}

all_products = electronics | office | accessories
common_all = electronics & office & accessories

electronics_only = electronics - office - accessories
office_only = office - electronics - accessories
accessories_only = accessories - electronics - office

shared_two = (
    (electronics & office)
    ^ (electronics & accessories)
    ^ (office & accessories)
)

print("All Products:", all_products)
print("Common to All:", common_all)
print("Electronics Only:", electronics_only)
print("Office Only:", office_only)
print("Accessories Only:", accessories_only)
print("Shared by Exactly Two:", shared_two)
````

# 🚀 Project 05 — Duplicate Data Cleaner

A customer dataset contains repeated IDs.

Use sets to:
- Remove duplicates
- Count unique customers
- Find customers common to two sources
- Find customers only in source 1
- Find customers only in source 2
- Find the combined unique customer population

````python
source_1 = [101, 102, 103, 101, 104, 105, 103]
source_2 = [103, 104, 106, 107, 104, 108]

customers_1 = set(source_1)
customers_2 = set(source_2)

common = customers_1 & customers_2
only_source_1 = customers_1 - customers_2
only_source_2 = customers_2 - customers_1
all_customers = customers_1 | customers_2

print("Unique Source 1:", customers_1)
print("Unique Source 2:", customers_2)
print("Total Unique Customers:", len(all_customers))
print("Common Customers:", common)
print("Only Source 1:", only_source_1)
print("Only Source 2:", only_source_2)
````

# 📚 Set Methods — Complete Quick Reference

| Method | Purpose |
|---|---|
| `add(x)` | Add one element |
| `update(iterable)` | Add elements |
| `remove(x)` | Remove; raises `KeyError` if absent |
| `discard(x)` | Remove if present |
| `pop()` | Remove and return an arbitrary element |
| `clear()` | Remove everything |
| `copy()` | Shallow copy |
| `union()` | Elements from either set |
| `intersection()` | Common elements |
| `difference()` | Elements only in first set |
| `symmetric_difference()` | Elements in exactly one |
| `issubset()` | Subset test |
| `issuperset()` | Superset test |
| `isdisjoint()` | No-common-elements test |
| `intersection_update()` | Keep common elements in place |
| `difference_update()` | Remove common elements in place |
| `symmetric_difference_update()` | Replace with symmetric difference |

### Operators

```text
A | B → Union
A & B → Intersection
A - B → Difference
A ^ B → Symmetric Difference
A <= B → Subset
A < B → Proper Subset
A >= B → Superset
A > B → Proper Superset
```

# 🧠 Set Decision Guide

```text
Need UNIQUE values?
        ↓
       SET

Need values from EITHER set?
        ↓
     union() / |

Need values COMMON to both?
        ↓
 intersection() / &

Need values ONLY in A?
        ↓
 difference() / -

Need values in EXACTLY ONE?
        ↓
symmetric_difference() / ^

Does A fit inside B?
        ↓
     issubset()

Does A contain B?
        ↓
    issuperset()

Do they have NOTHING in common?
        ↓
    isdisjoint()

Need safe removal?
        ↓
     discard()
```
