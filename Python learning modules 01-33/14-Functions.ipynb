# 1. What Is a Function?

A function is a reusable block of code designed to perform a specific task.

### Why use functions?

Without functions:

```text
Repeat code
↓
Long program
↓
Hard to maintain
↓
More bugs
```

With functions:

```text
Write once
↓
Call many times
↓
Reusable
↓
Readable
↓
Maintainable
```

### Basic Syntax

```python
def function_name(parameters):
    # function body
    return result
```

````python
def add(a, b):
    return a + b

result = add(10, 20)

print(result)
````

# 2. `def`

`def` starts a function definition.

```python
def greet():
    print("Hello")
```

The function body must be indented.

Defining a function does not execute its body immediately. The body runs when the function is called.

````python
def greet():
    print("Hello, Python!")

print("Before function call")

greet()

print("After function call")
````

# 3. Calling a Function

A function is called using parentheses:

```python
greet()
```

For functions with parameters:

```python
add(10, 20)
```

The values supplied during the call are arguments.

````python
def square(number):
    return number ** 2

print(square(5))
print(square(10))
````

# 4. Parameters vs Arguments

### Parameter

A variable written in the function definition:

```python
def add(a, b):
```

`a` and `b` are parameters.

### Argument

A value supplied when calling the function:

```python
add(10, 20)
```

`10` and `20` are arguments.

### Remember

```text
Definition → Parameters
Call       → Arguments
```

# 5. Positional Arguments

Arguments can be passed according to their position.

```python
def introduce(name, age):
    ...

introduce("Kaif", 22)
```

Here `"Kaif"` goes to `name` and `22` goes to `age`.

````python
def introduce(name, age):
    return f"{name} is {age} years old."

print(introduce("Kaif", 22))
````

# 6. Keyword Arguments

Arguments can be supplied using parameter names.

```python
introduce(name="Kaif", age=22)
```

This makes calls more explicit and can improve readability.

````python
def introduce(name, age):
    return f"{name} is {age} years old."

print(introduce(age=22, name="Kaif"))
````

# 7. Positional + Keyword Arguments

Python allows positional arguments followed by keyword arguments.

```python
function(10, name="Kaif")
```

Important rule:

**Positional arguments must come before keyword arguments.**

Incorrect:

```python
function(name="Kaif", 22)
```

This produces a `SyntaxError`.

````python
def profile(name, age, city):
    return f"{name}, {age}, {city}"

print(profile("Kaif", age=22, city="Mumbai"))
````

# 8. Default Arguments

A parameter can have a default value.

```python
def greet(name, message="Welcome"):
    return f"{message}, {name}"
```

If `message` is omitted, its default is used.

A parameter with a default value must not be followed by a required positional parameter in the same parameter list.

````python
def greet(name, message="Welcome"):
    return f"{message}, {name}!"

print(greet("Kaif"))
print(greet("Kaif", "Good morning"))
````

# 9. Return Values

`return` sends a value back to the caller.

```python
def add(a, b):
    return a + b
```

The returned value can be stored, printed, compared, or used in another calculation.

If a function reaches the end without a `return` statement, it returns `None`.

````python
def add(a, b):
    return a + b

total = add(15, 25)

print("Total:", total)
print("Type:", type(total))
````

# 10. `print()` vs `return`

These are different.

### `print()`

Displays a value.

### `return`

Sends a value back to the caller.

Prefer `return` when a function's result needs to be reused by other code.

````python
def using_print(a, b):
    print(a + b)

def using_return(a, b):
    return a + b

result = using_print(10, 20)
print("using_print returned:", result)

result = using_return(10, 20)
print("using_return returned:", result)
print("Reusable result:", result * 2)
````

# 11. Multiple Return Values

A function can return multiple values separated by commas.

```python
def calculate(a, b):
    return a + b, a - b
```

Python returns them together as a tuple, which can be unpacked.

````python
def calculate(a, b):
    return a + b, a - b, a * b

addition, subtraction, multiplication = calculate(20, 5)

print("Addition:", addition)
print("Subtraction:", subtraction)
print("Multiplication:", multiplication)
````

# 12. Returning Early

A `return` immediately exits the current function.

This is useful for validation and guard clauses.

````python
def divide(a, b):
    if b == 0:
        return None
    return a / b

print(divide(10, 2))
print(divide(10, 0))
````

# 13. Variable-Length Arguments — `*args`

`*args` collects extra positional arguments into a tuple.

### Syntax

```python
def function(*args):
    ...
```

Inside the function, `args` is a tuple.

````python
def total(*numbers):
    return sum(numbers)

print(total(10, 20))
print(total(10, 20, 30, 40, 50))
````

# 14. `*args` With Required Parameters

A normal parameter can appear before `*args`.

```python
def report(name, *scores):
    ...
```

The first positional argument goes to `name`; remaining positional arguments are collected into `scores`.

````python
def score_report(name, *scores):
    total = sum(scores)
    average = total / len(scores) if scores else 0
    return name, total, average

print(score_report("Kaif", 80, 90, 85, 95))
````

# 15. Variable-Length Keyword Arguments — `**kwargs`

`**kwargs` collects extra keyword arguments into a dictionary.

```python
def profile(**kwargs):
    ...
```

Inside the function, `kwargs` is a dictionary.

````python
def show_profile(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

show_profile(
    name="Kaif",
    role="Data Scientist",
    experience=2
)
````

# 16. Combining `*args` and `**kwargs`

A flexible function can accept both.

```python
def function(*args, **kwargs):
    ...
```

Typical order:

```text
positional parameters
*
*args
keyword-only parameters
**
**kwargs
```

The exact signature depends on the design of the function.

````python
def inspect_data(*args, **kwargs):
    print("Positional:", args)
    print("Keyword:", kwargs)

inspect_data(
    10,
    20,
    name="Kaif",
    role="Analyst"
)
````

# 17. Unpacking Arguments With `*`

The `*` operator can unpack an iterable into positional arguments.

```python
numbers = [10, 20]
add(*numbers)
```

The list becomes two positional arguments.

````python
def add(a, b):
    return a + b

values = [10, 20]

print(add(*values))
````

# 18. Unpacking Keyword Arguments With `**`

The `**` operator can unpack a dictionary into keyword arguments.

```python
data = {"name": "Kaif", "age": 22}
function(**data)
```

````python
def introduce(name, age):
    return f"{name} is {age} years old."

data = {
    "name": "Kaif",
    "age": 22
}

print(introduce(**data))
````

# 19. Local Variables

A variable created inside a function normally belongs to that function's local scope.

```python
def calculate():
    total = 100
```

`total` is local to `calculate()`.

````python
def calculate():
    total = 100
    return total

print(calculate())

# `total` is not available here because it is local to the function.
````

# 20. Global Variables

A variable defined at module/global scope can be read inside a function.

```python
tax_rate = 0.18

def calculate_tax(amount):
    return amount * tax_rate
```

However, relying heavily on mutable global state can make programs harder to test and maintain.

````python
tax_rate = 0.18

def calculate_tax(amount):
    return amount * tax_rate

print(calculate_tax(1000))
````

# 21. `global`

The `global` statement tells Python that an assignment inside a function should target a global variable.

Use it sparingly. Prefer returning values and passing dependencies as parameters when possible.

````python
counter = 0

def increment():
    global counter
    counter += 1

increment()
increment()

print(counter)
````

# 22. Scope and LEGB

Python resolves names using the LEGB rule:

```text
L → Local
E → Enclosing
G → Global
B → Built-in
```

### Local
Inside the current function.

### Enclosing
Inside an enclosing function when nested functions are involved.

### Global
At module level.

### Built-in
Names supplied by Python, such as `len`, `sum`, and `print`.

Python searches these scopes in that order.

````python
name = "global"

def outer():
    name = "enclosing"

    def inner():
        name = "local"
        return name

    return inner()

print(outer())
print(len([1, 2, 3]))
````

# 23. `nonlocal`

Inside nested functions, `nonlocal` allows an inner function to rebind a variable from the nearest enclosing function scope.

It is different from `global`.

```text
nonlocal → enclosing function scope
global   → module/global scope
```

````python
def counter_factory():
    count = 0

    def increment():
        nonlocal count
        count += 1
        return count

    return increment

counter = counter_factory()

print(counter())
print(counter())
print(counter())
````

# 24. Positional-Only and Keyword-Only Parameters

Modern Python function signatures can control how arguments are supplied.

### Positional-only

Parameters before `/` are positional-only.

```python
def power(base, exponent, /):
    return base ** exponent
```

### Keyword-only

Parameters after `*` are keyword-only.

```python
def connect(host, *, timeout=30):
    ...
```

These features are useful for designing clear and stable APIs.

````python
def power(base, exponent, /):
    return base ** exponent

print(power(2, 5))

def connect(host, *, timeout=30):
    return f"Connecting to {host} with timeout={timeout}"

print(connect("server.example.com"))
print(connect("server.example.com", timeout=60))
````

# 25. Function Annotations / Type Hints

Type hints document expected types.

```python
def add(a: int, b: int) -> int:
    return a + b
```

They improve readability and tooling but do not by themselves enforce runtime types.

````python
def add(a: int, b: int) -> int:
    return a + b

print(add(10, 20))
````

# 26. Docstrings

A docstring documents what a function does.

```python
def calculate_total(values):
    """Return the total of numeric values."""
    return sum(values)
```

Use clear docstrings for reusable or public functions.

````python
def calculate_total(values):
    """Return the total of numeric values."""
    return sum(values)

print(calculate_total([100, 200, 300]))
print(calculate_total.__doc__)
````

# 27. Functions as Objects

Functions are first-class objects in Python.

You can:
- assign a function to a variable
- pass a function to another function
- return a function from a function

This foundation becomes important later for decorators, callbacks, and functional programming.

````python
def greet(name):
    return f"Hello, {name}"

say_hello = greet

print(say_hello("Kaif"))
````

# 28. Passing Functions to Functions

A function can receive another function as an argument.

````python
def square(number):
    return number ** 2

def apply_operation(value, operation):
    return operation(value)

print(apply_operation(5, square))
````

# 29. Lambda Functions

A lambda is a small anonymous function.

### Syntax

```python
lambda arguments: expression
```

Example:

```python
square = lambda x: x ** 2
```

Use lambdas for short expressions, especially as `key=` functions. For complex logic, prefer `def`.

# 30. Recursion — Function Foundation

A recursive function calls itself.

Every useful recursive function needs a base case.

```text
Problem
↓
Base case?
↓ yes → return
↓ no
Smaller problem
↓
Recursive call
```

Recursion is a separate advanced topic, but understanding functions makes it possible.

````python
def countdown(n):
    if n == 0:
        print("Done")
        return

    print(n)
    countdown(n - 1)

countdown(5)
````

# 31. Mutable Default Argument Warning

Avoid mutable objects such as lists or dictionaries as default values when you intend to create a fresh object for every call.

Prefer:

```python
def collect(value, items=None):
    if items is None:
        items = []
```

This avoids accidental state sharing between calls.

````python
def add_item(item, items=None):
    if items is None:
        items = []

    items.append(item)
    return items

print(add_item("Python"))
print(add_item("SQL"))
````

# 32. Function Design Best Practices

### Good function

```text
One clear responsibility
↓
Meaningful name
↓
Small input/output contract
↓
Returns useful result
↓
Easy to test
```

### Best practices

- Use descriptive names.
- Keep functions focused.
- Prefer explicit parameters.
- Return values instead of printing when results must be reused.
- Avoid unnecessary global state.
- Use type hints for important interfaces.
- Add docstrings to reusable functions.
- Avoid excessive `*args` / `**kwargs` when a clear signature is better.
- Keep side effects visible and intentional.

````python
def calculate_discount(price: float, rate: float = 0.10) -> float:
    """Return the price after applying a percentage discount."""
    return price * (1 - rate)

print(calculate_discount(1000))
print(calculate_discount(1000, 0.20))
````

# 🚀 Project 02 — Grade Calculator

## Business Problem

Create functions that calculate total, average, grade, and pass/fail status.

## Approach

```text
Input
↓
Validation
↓
Reusable Functions
↓
Business Logic
↓
Return Result
↓
Output
```

## Solution

````python
def calculate_total(marks):
    return sum(marks)

def calculate_average(marks):
    return calculate_total(marks) / len(marks)

def get_grade(average):
    if average >= 90:
        return "A+"
    if average >= 80:
        return "A"
    if average >= 70:
        return "B"
    if average >= 60:
        return "C"
    if average >= 40:
        return "D"
    return "F"

def get_status(average):
    return "Pass" if average >= 40 else "Fail"

marks = [88, 92, 76, 85, 90]
average = calculate_average(marks)

print("Total:", calculate_total(marks))
print(f"Average: {average:.2f}")
print("Grade:", get_grade(average))
print("Status:", get_status(average))
````

# 🚀 Project 03 — Sales KPI Analyzer

```text
Input
↓
Validation

Reusable Functions

Business Logic
```

````python
def total_sales(sales):
    return sum(sales)

def average_sales(sales):
    return total_sales(sales) / len(sales) if sales else 0

def highest_sale(sales):
    return max(sales) if sales else None

def lowest_sale(sales):
    return min(sales) if sales else None

sales = [12000, 18500, 9200, 24000, 15600]

print("Total:", total_sales(sales))
print(f"Average: {average_sales(sales):.2f}")
print("Highest:", highest_sale(sales))
print("Lowest:", lowest_sale(sales))
````

# 🚀 Project 04 — Expense Tracker

## Business Problem

Build functions that add expenses, calculate totals, calculate category totals, and identify the largest expense.

## Approach

```text
Input
↓
Validation
↓
Reusable Functions
↓
Business Logic
↓
Return Result
↓
Output
```

## Solution

````python
def add_expense(expenses, category, amount):
    expenses.append({
        "category": category,
        "amount": amount
    })

def total_expenses(expenses):
    return sum(item["amount"] for item in expenses)

def category_total(expenses, category):
    return sum(
        item["amount"]
        for item in expenses
        if item["category"] == category
    )

def largest_expense(expenses):
    return max(expenses, key=lambda item: item["amount"])

expenses = []

add_expense(expenses, "Food", 250)
add_expense(expenses, "Transport", 450)
add_expense(expenses, "Food", 500)
add_expense(expenses, "Shopping", 2500)

print("Total:", total_expenses(expenses))
print("Food:", category_total(expenses, "Food"))
print("Largest:", largest_expense(expenses))
````

# 🚀 Project 06 — Inventory Manager

## Business Problem

Create functions to add products, update stock, calculate inventory value, and identify low-stock items.

## Approach

```text
Input
↓
Validation
↓
Reusable Functions
↓
Business Logic
↓
Return Result
↓
Output
```

## Solution

````python
def add_product(inventory, product_id, name, price, stock):
    inventory[product_id] = {
        "name": name,
        "price": price,
        "stock": stock
    }

def update_stock(inventory, product_id, quantity):
    if product_id in inventory:
        inventory[product_id]["stock"] += quantity

def inventory_value(inventory):
    return sum(
        product["price"] * product["stock"]
        for product in inventory.values()
    )

def low_stock(inventory, threshold=10):
    return [
        product
        for product in inventory.values()
        if product["stock"] <= threshold
    ]

inventory = {}

add_product(inventory, 101, "Laptop", 65000, 8)
add_product(inventory, 102, "Mouse", 1200, 25)
add_product(inventory, 103, "Monitor", 18000, 6)

update_stock(inventory, 101, 5)

print("Inventory Value:", inventory_value(inventory))
print("Low Stock:", low_stock(inventory))
````

# 🚀 Project 08 — Text Analyzer

## Business Problem

Create functions for character count, word count, unique words, and most common words using core Python.

## Approach

```text
Input
↓
Validation
↓
Reusable Functions
↓
Business Logic
↓
Return Result
↓
Output
```

## Solution

````python
def word_count(text):
    return len(text.split())

def character_count(text):
    return len(text)

def unique_words(text):
    words = text.lower().split()
    return set(words)

def word_frequency(text):
    frequency = {}

    for word in text.lower().split():
        frequency[word] = frequency.get(word, 0) + 1

    return frequency

text = "Python makes data analysis powerful and Python makes automation powerful"

print("Characters:", character_count(text))
print("Words:", word_count(text))
print("Unique Words:", len(unique_words(text)))
print("Frequency:", word_frequency(text))
````

# 🚀 Project 09 — Banking Account System

## Business Problem

Build functions for deposit, withdrawal, balance checking, and transaction summaries.

## Approach

```text
Input
↓
Validation
↓
Reusable Functions
↓
Business Logic
↓
Return Result
↓
Output
```

## Solution

````python
def deposit(balance, amount):
    if amount <= 0:
        return balance
    return balance + amount

def withdraw(balance, amount):
    if amount <= 0 or amount > balance:
        return balance
    return balance - amount

def account_summary(balance, transactions):
    return {
        "balance": balance,
        "transaction_count": len(transactions),
        "total_deposits": sum(
            amount for kind, amount in transactions
            if kind == "deposit"
        ),
        "total_withdrawals": sum(
            amount for kind, amount in transactions
            if kind == "withdraw"
        )
    }

balance = 10000
transactions = []

balance = deposit(balance, 5000)
transactions.append(("deposit", 5000))

balance = withdraw(balance, 2500)
transactions.append(("withdraw", 2500))

print(account_summary(balance, transactions))
````

# 🚀 Project 10 — Data Quality / Validation Engine

## Business Problem

Create reusable validation functions for names, emails, ages, numeric ranges, and complete records.

## Approach

```text
Input
↓
Validation
↓
Reusable Functions
↓
Business Logic
↓
Return Result
↓
Output
```

## Solution

````python
def is_non_empty(value):
    return isinstance(value, str) and bool(value.strip())

def is_valid_email(email):
    return (
        isinstance(email, str)
        and "@" in email
        and "." in email.split("@")[-1]
    )

def is_valid_age(age, minimum=18, maximum=100):
    return isinstance(age, int) and minimum <= age <= maximum

def validate_customer(customer):
    errors = []

    if not is_non_empty(customer.get("name", "")):
        errors.append("Invalid name")

    if not is_valid_email(customer.get("email", "")):
        errors.append("Invalid email")

    if not is_valid_age(customer.get("age", 0)):
        errors.append("Invalid age")

    return errors

customer = {
    "name": "Kaif",
    "email": "kaif@example.com",
    "age": 22
}

errors = validate_customer(customer)

if errors:
    print("Validation failed:", errors)
else:
    print("Customer record is valid.")
````

| `return` | Sends result to caller |
| Multiple return | Multiple values, commonly packed into a tuple |
| `*args` | Extra positional arguments as tuple |
| `**kwargs` | Extra keyword arguments as dictionary |
| Local scope | Current function scope |
| Enclosing scope | Outer function scope |
| Global scope | Module-level scope |
| Built-in scope | Python's built-in names |
| `nonlocal` | Rebind enclosing function variable |
| `global` | Rebind global variable |
| Type hint | Documents expected types |
| Docstring | Documents a function |
| Lambda | Small anonymous function |

### Function Design Formula

```text
CLEAR INPUT
    ↓
VALIDATE
    ↓
PROCESS
    ↓
RETURN
```

### Most Important Rule

> **A good function should have a clear purpose, a clear input/output contract, and minimal unnecessary side effects.**
