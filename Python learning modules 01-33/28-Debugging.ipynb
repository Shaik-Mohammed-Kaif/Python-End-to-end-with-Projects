# 1. What Is Debugging?

Debugging is not simply "removing an error."

A proper debugging process asks:

1. What happened?
2. Where did it happen?
3. Why did it happen?
4. What should have happened?
5. What change fixes the root cause?
6. Does the fix break anything else?

```text
OBSERVE
  ↓
REPRODUCE
  ↓
ISOLATE
  ↓
UNDERSTAND
  ↓
FIX
  ↓
VERIFY
```

# 2. Syntax Errors

A syntax error means Python cannot parse the code.

Example:

```python
if True
    print("Hello")
```

The missing `:` prevents Python from parsing the statement.

````python
# Correct version
if True:
    print("Hello")
````

# 3. Runtime Errors

A runtime error occurs while the program is executing.

Example:

```python
10 / 0
```

Python raises `ZeroDivisionError`.

````python
try:
    result = 10 / 0
except ZeroDivisionError as error:
    print("Error:", error)
````

# 4. Logical Errors

Logical errors are often harder because the program runs without an exception,
but produces the wrong result.

Example:

```text
Expected total: 600
Actual total:   500
```

The debugging task is to discover where the calculation logic differs from the
intended business rule.

````python
prices = [100, 200, 300]

# Bug: accidentally skips the last value.
total = sum(prices[:-1])

print("Expected:", 600)
print("Actual:", total)
````

# 5. Reading a Traceback

A traceback usually gives:

- exception type
- error message
- file
- line number
- call stack

Example:

```text
Traceback (most recent call last):
    ...
ValueError: invalid literal for int()
```

### Read from the bottom first

The final line often tells you the immediate exception and message.

Then move upward to understand how execution reached that point.

````python
def convert_age(value: str) -> int:
    return int(value)

try:
    convert_age("twenty")
except ValueError as error:
    print(type(error).__name__ + ":", error)
````

# 6. Finding the Error Line

When a traceback points to a line, inspect:

1. the exact line
2. variables used there
3. the function containing it
4. the caller
5. assumptions made earlier

Do not automatically assume the highlighted line is the root cause; it can be
where an earlier bad value finally causes a failure.

````python
def calculate_average(values):
    total = sum(values)
    count = len(values)
    return total / count

try:
    print(calculate_average([]))
except ZeroDivisionError as error:
    print("Failure:", error)
````

# 7. Call Stack

A function can call another function, which calls another function.

```text
main()
  ↓
process_data()
  ↓
calculate_metric()
  ↓
error
```

The traceback helps you follow this chain.

Understanding the call stack is essential when debugging larger applications.

````python
def divide(a, b):
    return a / b

def calculate():
    return divide(100, 0)

def main():
    return calculate()

try:
    main()
except ZeroDivisionError as error:
    print("Caught:", error)
````

# 8. Reproduce the Bug

Before changing code, make the problem reproducible.

```text
INPUT
 ↓
PROGRAM
 ↓
WRONG RESULT / ERROR
```

A small reproducible example is much easier to debug than a large application.

````python
def discount(price, percentage):
    return price - percentage

price = 1000
percentage = 20

print("Expected:", 800)
print("Actual:", discount(price, percentage))
````

# 10. Print Debugging

Print debugging is simple and useful for small problems.

Good debug output should identify:

- variable
- value
- relevant state

Example:

```python

```

````python
sales = [1200, 2500, 1800]

for index, sale in enumerate(sales):
    print(f"DEBUG index={index}, sale={sale}")

print("Total:", sum(sales))
````

# 11. Debugging Logic

When output is wrong, break the calculation into meaningful steps.

Instead of:

```python
result = complex_expression
```

temporarily inspect the intermediate values.

````python
sales = [1200, 2500, 1800]

total = sum(sales)
count = len(sales)
average = total / count

print("DEBUG total:", total)
print("DEBUG count:", count)
print("DEBUG average:", average)
````

# 12. Debugging Loops

For loops, inspect each iteration when necessary.

Useful values include:

- index
- current item
- accumulator
- condition result

````python
numbers = [10, 20, 30, 40]
total = 0

for number in numbers:
    total += number
    print(f"DEBUG number={number}, total={total}")

print("Final:", total)
````

# 13. Debugging Conditions

A condition can be syntactically correct but logically wrong.

Inspect the individual parts of a complex condition.

````python
age = 22
has_id = True

age_allowed = age >= 18

print("DEBUG age_allowed:", age_allowed)
print("DEBUG has_id:", has_id)

if age_allowed and has_id:
    print("Eligible")
else:
    print("Not eligible")
````

# 14. Breakpoints

A breakpoint pauses execution at a selected line.

At a breakpoint you can inspect:

- variables
- call stack
- current line
- expressions
- program state

This is usually more powerful than adding many temporary `print()` statements.

````python
def calculate_total(price: float, quantity: int) -> float:
    total = price * quantity
    return total

print(calculate_total(250.0, 4))

# In a supported debugger, place a breakpoint on the `total` line.
````

# 15. Python's Built-in `breakpoint()`

Python provides:

```python
breakpoint()
```

It enters the configured debugger at that location.

Basic workflow:

```text
PROGRAM
  ↓
breakpoint()
  ↓
PAUSE
  ↓
INSPECT
  ↓
STEP
  ↓
CONTINUE
```

````python
def calculate_total(price: float, quantity: int) -> float:
    total = price * quantity
    # breakpoint()  # Uncomment when you want to start an interactive debug session.
    return total

print(calculate_total(250.0, 4))
````

# 16. Jupyter Debugging

Jupyter environments can provide interactive debugging depending on the
Jupyter interface and kernel configuration.

A common Python approach is:

```python

```

You can also use IDE/Jupyter debugger controls when your environment supports
them.

For notebook debugging:

```text

 ↓
INSPECT STATE
```

````python
def calculate_average(values: list[float]) -> float:
    if not values:
        return 0.0

    total = sum(values)
    count = len(values)

    # breakpoint()  # Uncomment to inspect total/count interactively.
    return total / count

print(calculate_average([100.0, 200.0, 300.0]))
````

# 17. VS Code Debugger

In VS Code, a common debugging workflow is:

1. Open the Python project.
2. Open the source file.
3. Set a breakpoint beside a line number.
4. Start the Python debugger.
5. Inspect variables.
6. Step over / into / out of functions.
7. Continue execution.
8. Fix the root cause.
9. Run tests again.

The exact UI can vary by VS Code and Python extension versions.

# 18. Step Over

**Step Over** executes the current line without entering the implementation of
a called function.

Useful when you trust the called function and want to inspect the next line.

# 19. Step Into

**Step Into** enters a function call so you can inspect its internal execution.

Useful when you suspect the problem is inside the called function.

# 20. Step Out

**Step Out** finishes the current function and returns to its caller.

Useful when you have inspected enough of the current function.

# 22. Assertions

Use:

```python
assert condition
```

An assertion checks that an expected condition is true.

If the condition is false, Python raises `AssertionError`.

Basic syntax:

```python
assert condition
assert condition, "message"
```

````python
age = 22

assert age >= 18

print("Age check passed")
````

# 23. Assertions with Messages

A message can explain what assumption failed.

````python
sales = [100, 200, 300]

assert len(sales) > 0, "Sales data must not be empty"

print("Sales records:", len(sales))
````

# 24. Assertions for Development Invariants

Assertions are useful for internal assumptions and invariants.

Example:

```text
PROCESS DATA
     ↓
EXPECTED INVARIANT
     ↓
assert
     ↓
CONTINUE
```

Use explicit validation for external/untrusted input rather than relying on
assertions as security checks.

````python
def calculate_average(values: list[float]) -> float:
    assert values, "Expected at least one value"
    return sum(values) / len(values)

print(calculate_average([100.0, 200.0, 300.0]))
````

# 25. Assertions Are Not Input Validation

Do not use:

```python
assert user_is_authenticated
```

as a security mechanism.

Assertions can be disabled with Python optimization settings.

For required runtime validation, use explicit checks and appropriate exception
handling.

````python
def validate_age(age: int) -> None:
    if age < 0:
        raise ValueError("Age cannot be negative")

validate_age(22)
print("Valid age")
````

# 26. Debugging Exceptions

Use exception information to understand failures.

```python
try:
    ...
except SomeError as error:
    print(type(error).__name__)
    print(error)
```

During development, avoid hiding unexpected errors with broad exception
handling.

````python
try:
    number = int("abc")
except ValueError as error:
    print("Exception:", type(error).__name__)
    print("Message:", error)
````

# 27. Avoid Bare `except`

This pattern is usually too broad:

```python
try:
    ...
except:
    ...
```

Prefer specific exceptions when possible:

```python
except ValueError:
    ...
```

If you intentionally catch a broad exception at an application boundary,
log useful diagnostic information and handle it deliberately.

````python
try:
    value = int("abc")
except ValueError as error:
    print("Could not convert value:", error)
````

# 29. Debugging a Business Rule

A business rule should be translated into explicit conditions.

Example:

```text
Customer eligible if:
age >= 18
AND
account is active
AND
balance >= required amount
```

Debug each component separately.

````python
age = 25
active = True
balance = 15000
required_balance = 10000

age_ok = age >= 18
active_ok = active
balance_ok = balance >= required_balance

print("age_ok:", age_ok)
print("active_ok:", active_ok)
print("balance_ok:", balance_ok)

eligible = age_ok and active_ok and balance_ok

print("Eligible:", eligible)
````

# 30. Debugging State Changes

Many bugs occur because a variable changes unexpectedly.

Track state transitions:

```text
BEFORE
 ↓
OPERATION
 ↓
AFTER
```

Only inspect the state you actually need; excessive logging can make debugging
harder.

````python
balance = 10000

print("Before:", balance)

withdrawal = 2500
balance -= withdrawal

print("After:", balance)

assert balance == 7500
````

Inspect:

````python
value = "100"

print("Value:", value)
print("Type:", type(value).__name__)

if isinstance(value, str):
    value = int(value)

print("Converted:", value)
print("New type:", type(value).__name__)
````

# 33. Debugging With Small Functions

Large functions are difficult to debug.

Break complicated logic into small, testable functions:

```text
INPUT
 ↓
VALIDATE
 ↓
TRANSFORM
 ↓
CALCULATE
 ↓
FORMAT
```

Each stage becomes easier to inspect and test.

````python
def validate_amount(amount: float) -> None:
    if amount < 0:
        raise ValueError("Amount cannot be negative")

def calculate_tax(amount: float, rate: float) -> float:
    return amount * rate

def final_amount(amount: float, rate: float) -> float:
    validate_amount(amount)
    tax = calculate_tax(amount, rate)
    return amount + tax

print(final_amount(1000.0, 0.18))
````

# 34. Debugging Workflow

Use this professional workflow:

```text
1. READ ERROR
      ↓
2. REPRODUCE
      ↓
3. LOCATE
      ↓
4. INSPECT
      ↓
5. FORM HYPOTHESIS
      ↓
6. TEST HYPOTHESIS
      ↓
7. FIX ROOT CAUSE
      ↓
8. RUN TESTS
      ↓
9. VERIFY EDGE CASES
      ↓
10. DOCUMENT IF NEEDED
```

Avoid random code changes. Debugging is a reasoning process.

# 35. Common Debugging Mistakes

### Mistake 1 — Ignoring the traceback

Read it carefully.

### Mistake 2 — Fixing only the symptom

Find the root cause.

### Mistake 3 — Changing many things at once

Make one controlled change at a time.

### Mistake 4 — Using only `print()`

Use breakpoints and a debugger for complex state.

### Mistake 5 — Catching every exception

Broad exception handling can hide the actual problem.

### Mistake 6 — Not reproducing the bug

Without a reproducible case, verification becomes difficult.

### Mistake 7 — Forgetting to retest

A fix is incomplete until the relevant behavior has been verified.

# 36. Best Practices

- Read the complete traceback.
- Start with the final exception line.
- Reproduce the problem.
- Reduce the problem when possible.
- Inspect values and types.
- Use breakpoints for complex flows.
- Use `breakpoint()` when interactive inspection is useful.
- Use assertions for development invariants.
- Use explicit validation for required runtime checks.
- Catch specific exceptions where possible.
- Fix root causes rather than symptoms.
- Run tests after the fix.
- Verify edge cases.
- Remove temporary debugging code before production when it is no longer needed.
- Keep useful production logging separate from temporary debugging.

# 🚀 Project 10 — 10 — Production-Style Debugging Workflow

## Business Problem

Build a small analytics workflow using validation, explicit diagnostics, assertions, exception handling, and final verification.

## Debugging Architecture

```text
INPUT
 ↓
REPRODUCE
 ↓
INSPECT
 ↓
VALIDATE
 ↓
TRACE LOGIC
 ↓
FIX
 ↓
ASSERT / TEST
 ↓
FINAL OUTPUT
```

## Implementation

````python
from typing import TypedDict

class Sale(TypedDict):
    sale_id: int
    customer: str
    amount: float

def validate_sales(sales: list[Sale]) -> list[Sale]:
    valid: list[Sale] = []

    for sale in sales:
        try:
            assert sale["sale_id"] > 0
            assert sale["customer"].strip()
            assert sale["amount"] >= 0
        except (AssertionError, KeyError) as error:
            print("DEBUG invalid sale:", sale, "|", type(error).__name__)
            continue

        valid.append(sale)

    return valid

def build_report(sales: list[Sale]) -> dict[str, float | int]:
    assert sales, "Expected at least one valid sale"

    total = sum(sale["amount"] for sale in sales)
    count = len(sales)
    average = total / count

    print("DEBUG count:", count)
    print("DEBUG total:", total)
    print("DEBUG average:", average)

    return {
        "count": count,
        "total": total,
        "average": average,
    }

sales: list[Sale] = [
    {"sale_id": 1, "customer": "Kaif", "amount": 65000.0},
    {"sale_id": 2, "customer": "Sara", "amount": 18000.0},
    {"sale_id": 3, "customer": "Ali", "amount": -500.0},
    {"sale_id": 4, "customer": "Maya", "amount": 42000.0},
]

clean_sales = validate_sales(sales)
report = build_report(clean_sales)

assert report["count"] == 3
assert report["total"] == 125000.0

print("FINAL REPORT:", report)
````
