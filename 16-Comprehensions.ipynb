# 1. What Is a Comprehension?

A comprehension combines iteration and collection creation into a compact
expression.

Instead of:

```python
squares = []

for x in range(10):
    squares.append(x ** 2)
```

we can write:

```python
squares = [x ** 2 for x in range(10)]
```

### Mental Model

```text
SOURCE ITERABLE
      ↓
    LOOP
      ↓
OPTIONAL CONDITION
      ↓
EXPRESSION
      ↓
NEW COLLECTION
```

A comprehension does not automatically make code better. Use it when the
shorter form remains clear.

````python
squares = [x ** 2 for x in range(10)]

print(squares)
````

# 2. List Comprehension — Basic Syntax

```python
[expression for item in iterable]
```

### Parts

- `expression` → value placed in the new list
- `item` → current element
- `iterable` → source collection

Example:

```python
[x * 2 for x in numbers]
```

````python
numbers = [1, 2, 3, 4, 5]

doubled = [x * 2 for x in numbers]

print(doubled)
````

# 3. List Comprehension From `range()`

`range()` is commonly used with comprehensions.

````python
numbers = [x for x in range(1, 11)]

print(numbers)
````

# 4. Transforming Strings

Comprehensions can transform each string in a collection.

````python
names = ["kaif", "sara", "ali"]

uppercase_names = [name.upper() for name in names]

print(uppercase_names)
````

# 5. Conditional List Comprehension

### Syntax

```python
[expression for item in iterable if condition]
```

Only elements satisfying the condition are included.

````python
numbers = range(1, 11)

even_numbers = [
    x for x in numbers
    if x % 2 == 0
]

print(even_numbers)
````

# 6. Conditional Transformation — `if / else`

There is another form when every item should produce a result but the
result depends on a condition.

### Syntax

```python
[expression_if_true if condition else expression_if_false
 for item in iterable]
```

Important: the conditional expression comes **before** the `for`.

````python
numbers = range(1, 8)

labels = [
    "Even" if x % 2 == 0 else "Odd"
    for x in numbers
]

print(labels)
````

# 7. Filtering vs Conditional Transformation

### Filtering

```python
[x for x in numbers if x > 5]
```

This changes how many elements are returned.

### Conditional transformation

```python
["High" if x > 5 else "Low" for x in numbers]
```

This normally keeps the same number of elements but changes their values.

````python
numbers = range(1, 8)

filtered = [x for x in numbers if x > 5]

classified = [
    "High" if x > 5 else "Low"
    for x in numbers
]

print("Filtered:", filtered)
print("Classified:", classified)
````

# 8. Multiple Conditions

Multiple conditions can be combined.

```python
[x for x in numbers if x > 0 and x % 2 == 0]
```

Use parentheses when they make a complex condition easier to read.

````python
numbers = range(1, 21)

result = [
    x for x in numbers
    if x > 5 and x % 2 == 0
]

print(result)
````

# 9. List Comprehension From a Dictionary

Iterating over a dictionary normally iterates over its keys.

````python
prices = {
    "Laptop": 65000,
    "Mouse": 1200,
    "Monitor": 18000
}

products = [name for name in prices]

print(products)
````

# 10. Dictionary `.items()` With List Comprehension

Use `.items()` when both keys and values are needed.

````python
prices = {
    "Laptop": 65000,
    "Mouse": 1200,
    "Monitor": 18000
}

expensive = [
    name
    for name, price in prices.items()
    if price >= 10000
]

print(expensive)
````

# 11. Nested List Comprehension

A nested comprehension can represent nested loops.

Normal:

```python
result = []

for row in matrix:
    for value in row:
        result.append(value)
```

Comprehension:

```python
[value for row in matrix for value in row]
```

Read the clauses from left to right as nested loops.

````python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

flattened = [
    value
    for row in matrix
    for value in row
]

print(flattened)
````

# 12. Nested Comprehension With a Condition

````python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

even_values = [
    value
    for row in matrix
    for value in row
    if value % 2 == 0
]

print(even_values)
````

# 13. Building a Matrix

A comprehension can generate a nested list.

````python
matrix = [
    [0 for column in range(4)]
    for row in range(3)
]

print(matrix)
````

# 14. Avoid the Shared-Row Trap

Do not create a matrix like this when independent rows are required:

```python
matrix = [[0] * 4] * 3
```

The rows reference the same inner list.

Prefer:

```python
matrix = [[0] * 4 for _ in range(3)]
```

````python
matrix = [[0] * 4 for _ in range(3)]

matrix[0][0] = 99

print(matrix)
````

# 15. Dictionary Comprehension

Dictionary comprehensions create dictionaries.

### Syntax

```python
{key_expression: value_expression
 for item in iterable}
```

````python
numbers = range(1, 6)

squares = {
    x: x ** 2
    for x in numbers
}

print(squares)
````

# 16. Dictionary Comprehension With Condition

````python
prices = {
    "Laptop": 65000,
    "Mouse": 1200,
    "Monitor": 18000,
    "Keyboard": 2500
}

expensive = {
    name: price
    for name, price in prices.items()
    if price >= 10000
}

print(expensive)
````

# 17. Dictionary Transformation

A dictionary comprehension can transform keys and values.

````python
sales = {
    "laptop": 65000,
    "mouse": 1200,
    "monitor": 18000
}

formatted = {
    product.title(): price
    for product, price in sales.items()
}

print(formatted)
````

# 18. Set Comprehension

Set comprehensions use `{}` with an expression and `for`.

### Syntax

```python
{expression for item in iterable}
```

The resulting set contains unique values.

````python
numbers = [1, 2, 2, 3, 3, 4, 4, 5]

squares = {
    x ** 2
    for x in numbers
}

print(squares)
````

# 19. Set Comprehension With Condition

````python
numbers = range(1, 21)

even_squares = {
    x ** 2
    for x in numbers
    if x % 2 == 0
}

print(even_squares)
````

# 20. Generator Expressions

A generator expression looks similar to a list comprehension but uses
parentheses.

```python
(x ** 2 for x in range(10))
```

It produces a generator that yields values lazily.

### Key Difference

```text
List comprehension
→ builds the list immediately

Generator expression
→ produces values on demand
```

````python
numbers = (x ** 2 for x in range(5))

print(numbers)

for value in numbers:
    print(value)
````

# 21. Generator Expression With `sum()`

Generator expressions are useful when you need to process values without
materializing a full list.

````python
numbers = range(1, 1_000_001)

total = sum(x * x for x in numbers)

print(total)
````

# 22. Generator Expression vs List Comprehension

### List

```python
sum([x * x for x in numbers])
```

### Generator

```python
sum(x * x for x in numbers)
```

For aggregations, the generator expression avoids creating a temporary list.

Use a list comprehension when you actually need a list and a generator
when lazy iteration is useful.

````python
numbers = range(1, 11)

list_result = [x * x for x in numbers]
generator_result = (x * x for x in numbers)

print("List:", list_result)
print("Generator:", generator_result)
print("Generator total:", sum(generator_result))
````

# 23. Comprehension With `enumerate()`

Comprehensions can combine with built-in iteration helpers.

````python
skills = ["Python", "SQL", "Excel"]

numbered = [
    f"{index}. {skill}"
    for index, skill in enumerate(skills, start=1)
]

print(numbered)
````

# 24. Comprehension With `zip()`

`zip()` is useful when corresponding values from multiple collections need
to be combined.

````python
names = ["Kaif", "Sara", "Ali"]
scores = [88, 94, 76]

results = [
    f"{name}: {score}"
    for name, score in zip(names, scores)
]

print(results)
````

# 25. Comprehension With Function Calls

The expression can call a function.

````python
def clean_name(name):
    return name.strip().title()

names = [" kaif ", " sara ", " ali "]

cleaned = [
    clean_name(name)
    for name in names
]

print(cleaned)
````

# 26. Comprehension With Nested Conditions

Complex conditions are possible, but readability matters.

If a comprehension becomes difficult to read, use a normal `for` loop or
move logic into a named function.

````python
numbers = range(1, 31)

result = [
    x
    for x in numbers
    if x % 2 == 0
    and x % 3 == 0
]

print(result)
````

# 27. Conditional Dictionary Mapping

A dictionary comprehension can classify values.

````python
scores = {
    "Kaif": 88,
    "Sara": 95,
    "Ali": 72
}

grades = {
    name: "Pass" if score >= 40 else "Fail"
    for name, score in scores.items()
}

print(grades)
````

# 28. Conditional Set Creation

````python
emails = [
    "kaif@example.com",
    "invalid",
    "sara@example.com",
    "admin@example.com"
]

valid_domains = {
    email.split("@")[1]
    for email in emails
    if "@" in email
}

print(valid_domains)
````

# 29. Comprehension Readability Rule

A short comprehension is excellent:

```python
squares = [x ** 2 for x in numbers]
```

A very complex comprehension can become difficult to maintain.

### Prefer a loop when:

- There are many nested levels.
- Conditions are complex.
- Multiple side effects are involved.
- Debugging the expression is difficult.
- A beginner cannot understand it quickly.

**Pythonic does not mean "shortest possible code."**

# 30. Common Mistakes

### Mistake 1 — Confusing filter syntax

Correct:

```python
[x for x in numbers if x > 10]
```

### Mistake 2 — Putting `if/else` in the wrong position

Correct:

```python
["Even" if x % 2 == 0 else "Odd" for x in numbers]
```

### Mistake 3 — Forgetting dictionary key/value syntax

Correct:

```python
{x: x ** 2 for x in numbers}
```

### Mistake 4 — Expecting a generator to be reusable

Generators are consumed as they are iterated.

### Mistake 5 — Overusing nested comprehensions

Readable loops are sometimes the better solution.

````python
numbers = [1, 2, 3, 4, 5]

# Correct filtering
result = [x for x in numbers if x > 2]

# Correct conditional expression
labels = [
    "Even" if x % 2 == 0 else "Odd"
    for x in numbers
]

print(result)
print(labels)
````

# 🚀 Project 01 — Sales Data Transformation

## Business Problem

Transform product sales into cleaned, formatted records using list and dictionary comprehensions.

## Workflow

```text
RAW INPUT
   ↓
ITERATE
   ↓
FILTER / TRANSFORM
   ↓
COMPREHENSION
   ↓
STRUCTURED OUTPUT
```

## Solution

````python
sales = [
    {"product": " laptop ", "amount": 65000},
    {"product": " mouse ", "amount": 1200},
    {"product": " monitor ", "amount": 18000}
]

clean_sales = [
    {
        "product": item["product"].strip().title(),
        "amount": item["amount"],
        "status": "High Value" if item["amount"] >= 10000 else "Standard"
    }
    for item in sales
]

print(clean_sales)
````

# 🚀 Project 02 — Student Performance Analyzer

## Business Problem

Create grade labels and extract high-performing students with comprehensions.

## Workflow

```text
RAW INPUT
   ↓
ITERATE
   ↓
FILTER / TRANSFORM
   ↓
COMPREHENSION
   ↓
STRUCTURED OUTPUT
```

````python
students = {
    "Kaif": 88,
    "Sara": 95,
    "Ali": 72,
    "Omar": 35
}

grades = {
    name: "Pass" if score >= 40 else "Fail"
    for name, score in students.items()
}

top_students = [
    name
    for name, score in students.items()
    if score >= 80
]

print("Grades:", grades)
print("Top Students:", top_students)
````

# 🚀 Project 03 — E-Commerce Inventory Filter

## Business Problem

Create inventory values and identify low-stock products using dictionary and list comprehensions.

## Workflow

```text
RAW INPUT
   ↓
ITERATE
   ↓
FILTER / TRANSFORM
   ↓
COMPREHENSION
   ↓
STRUCTURED OUTPUT
```

## Solution

````python
inventory = {
    "Laptop": {"price": 65000, "stock": 8},
    "Mouse": {"price": 1200, "stock": 40},
    "Monitor": {"price": 18000, "stock": 6},
    "Keyboard": {"price": 2500, "stock": 25}
}

inventory_value = {
    name: data["price"] * data["stock"]
    for name, data in inventory.items()
}

low_stock = [
    name
    for name, data in inventory.items()
    if data["stock"] <= 10
]

print("Inventory Value:", inventory_value)
print("Low Stock:", low_stock)
````

# 🚀 Project 04 — Text Cleaning Toolkit

## Business Problem

Clean and analyze text using string methods inside comprehensions.

## Workflow

```text
RAW INPUT
   ↓
ITERATE
   ↓
FILTER / TRANSFORM
   ↓
COMPREHENSION
   ↓
STRUCTURED OUTPUT
```

## Solution

````python
text = "Python makes data analysis powerful and Python is readable"

words = text.lower().split()

clean_words = [
    word.strip(".,!?")
    for word in words
]

long_words = [
    word
    for word in clean_words
    if len(word) >= 7
]

unique_words = {
    word
    for word in clean_words
}

print("Clean Words:", clean_words)
print("Long Words:", long_words)
print("Unique Words:", unique_words)
````

# 🚀 Project 05 — Employee Salary Bands

## Business Problem

Classify employees into salary bands using dictionary comprehensions.

## Workflow

```text
RAW INPUT
   ↓
ITERATE
   ↓
FILTER / TRANSFORM
   ↓
COMPREHENSION
   ↓
STRUCTURED OUTPUT
```

## Solution

````python
employees = {
    "Kaif": 65000,
    "Sara": 85000,
    "Ali": 52000,
    "Omar": 38000
}

bands = {
    name: (
        "Senior"
        if salary >= 80000
        else "Mid"
        if salary >= 50000
        else "Entry"
    )
    for name, salary in employees.items()
}

senior_employees = [
    name
    for name, salary in employees.items()
    if salary >= 80000
]

print("Bands:", bands)
print("Senior:", senior_employees)
````

# 🚀 Project 07 — Matrix / Grid Analyzer

## Business Problem

Flatten a matrix, extract values meeting conditions, and create a transformed matrix.

## Workflow

```text
RAW INPUT
   ↓
ITERATE
   ↓
FILTER / TRANSFORM
   ↓
COMPREHENSION
   ↓
STRUCTURED OUTPUT
```

## Solution

````python
matrix = [
    [10, 15, 20],
    [25, 30, 35],
    [40, 45, 50]
]

flattened = [
    value
    for row in matrix
    for value in row
]

multiples_of_10 = [
    value
    for value in flattened
    if value % 10 == 0
]

doubled_matrix = [
    [value * 2 for value in row]
    for row in matrix
]

print("Flattened:", flattened)
print("Multiples of 10:", multiples_of_10)
print("Doubled Matrix:", doubled_matrix)
````

# 🚀 Project 08 — Customer Data Normalization

## Business Problem

Normalize names, filter valid ages, and create a dictionary keyed by customer ID.

## Workflow

```text
RAW INPUT
   ↓
ITERATE
   ↓
FILTER / TRANSFORM
   ↓
COMPREHENSION
   ↓
STRUCTURED OUTPUT
```

## Solution

````python
customers = [
    {"id": 101, "name": " kaif ", "age": 22},
    {"id": 102, "name": " sara ", "age": 27},
    {"id": 103, "name": " ali ", "age": 15}
]

normalized = [
    {
        **customer,
        "name": customer["name"].strip().title()
    }
    for customer in customers
]

adult_customers = [
    customer
    for customer in normalized
    if customer["age"] >= 18
]

by_id = {
    customer["id"]: customer
    for customer in adult_customers
}

print("Normalized:", normalized)
print("Adults:", adult_customers)
print("By ID:", by_id)
````

# 🚀 Project 09 — KPI Generator

## Business Problem

Use a generator expression to calculate analytics without creating unnecessary intermediate lists.

## Workflow

```text
RAW INPUT
   ↓
ITERATE
   ↓
FILTER / TRANSFORM
   ↓
COMPREHENSION
   ↓
STRUCTURED OUTPUT
```

## Solution

````python
sales = [12000, 18500, 9200, 24000, 15600]

total_sales = sum(
    sale for sale in sales
)

high_value_count = sum(
    1 for sale in sales
    if sale >= 15000
)

average_sales = total_sales / len(sales)

print("Total Sales:", total_sales)
print("High Value Transactions:", high_value_count)
print(f"Average Sales: {average_sales:.2f}")
````

# 🚀 Project 10 — Multi-Collection Business Report

## Business Problem

Combine `zip()`, `enumerate()`, list comprehensions, dictionary comprehensions, and set comprehensions to build a compact report.

## Workflow

```text
RAW INPUT
   ↓
ITERATE
   ↓
FILTER / TRANSFORM
   ↓
COMPREHENSION
   ↓
STRUCTURED OUTPUT
```

## Solution

````python
products = ["Laptop", "Mouse", "Monitor", "Keyboard"]
prices = [65000, 1200, 18000, 2500]
quantities = [5, 30, 8, 20]

records = [
    {
        "rank": index,
        "product": product,
        "price": price,
        "quantity": quantity,
        "value": price * quantity,
        "status": "High"
        if price * quantity >= 50000
        else "Normal"
    }
    for index, (product, price, quantity)
    in enumerate(zip(products, prices, quantities), start=1)
]

high_value_products = {
    record["product"]
    for record in records
    if record["status"] == "High"
}

value_by_product = {
    record["product"]: record["value"]
    for record in records
}

print("Records:")
for record in records:
    print(record)

print("\nHigh Value Products:", high_value_products)
print("Value By Product:", value_by_product)
````

# 📚 Comprehensions Quick Reference

### List

```python
[x for x in values]
```

### List + Filter

```python
[x for x in values if condition]
```

### List + Conditional Expression

```python
["Yes" if condition else "No" for x in values]
```

### Nested

```python
[value for row in matrix for value in row]
```

### Dictionary

```python
{x: x ** 2 for x in numbers}
```

### Dictionary + Filter

```python
{k: v for k, v in data.items() if condition}
```

### Set

```python
{x for x in values}
```

### Generator Expression

```python
(x ** 2 for x in numbers)
```

### With `enumerate()`

```python
[
    (index, value)
    for index, value in enumerate(values)
]
```

### With `zip()`

```python
[
    (a, b)
    for a, b in zip(first, second)
]
```

### Golden Rule

> **Use comprehensions for clear transformations and filtering. Use a normal loop when the comprehension becomes difficult to read.**

# 🧠 Decision Guide

```text
Need a new list?
      ↓
List comprehension

Need a new dictionary?
      ↓
Dictionary comprehension

Need unique generated values?
      ↓
Set comprehension

Need lazy / on-demand values?
      ↓
Generator expression

Need simple filtering?
      ↓
Comprehension + if

Need if/else transformation?
      ↓
Conditional expression before for

Need nested data?
      ↓
Nested comprehension
      ↓
If it becomes hard to read → normal loop
```

### Core Principle

```text
Readable
   >
Short
```

The goal is clean Python, not the fewest possible characters.
