## Module 26 — Type Hints

Type hints describe the **expected types** of variables, function parameters,
return values, collections, and custom structures.

> **Important:** Type hints are mainly used for readability, documentation,
> IDE support, and static type checking. Python normally does not enforce
> annotations automatically at runtime.

# 1. What Are Type Hints?

Type hints are annotations that communicate what type of value is expected.

Basic syntax:

```python
name: str
age: int
salary: float
```

A variable annotation does not by itself perform runtime validation.

````python
name: str = "Kaif"
age: int = 22
salary: float = 55000.0

print(name)
print(age)
print(salary)
````

# 2. Type Hints Do Not Automatically Enforce Types

Python is dynamically typed.

A static type checker may report this as incorrect:

```python
age: int = "twenty"
```

But ordinary Python execution does not automatically reject the assignment.

````python
age: int = "twenty"

print(age)
print(type(age).__name__)
````

# 3. Function Parameters

Annotate function parameters to document expected input types.

````python
def greet(name: str) -> str:
    return f"Hello, {name}"

print(greet("Kaif"))
````

# 4. Return Types

Use `->` to describe the expected return type.

````python
def calculate_total(price: float, quantity: int) -> float:
    return price * quantity

print(calculate_total(250.0, 4))
````

# 5. `None` Return Type

Use `-> None` when a function does not return a meaningful value.

````python
def print_report(title: str) -> None:
    print(f"Report: {title}")

print_report("Monthly Sales")
````

# 6. `list[T]`

Modern Python supports built-in generic collection syntax.

Examples:

```python
list[int]
list[str]
list[float]
```

````python
def average(values: list[float]) -> float:
    if not values:
        return 0.0
    return sum(values) / len(values)

sales: list[float] = [1200.0, 2500.0, 1800.0]

print(average(sales))
````

# 7. `dict[K, V]`

Use:

```python
dict[str, int]
```

for a dictionary with string keys and integer values.

````python
scores: dict[str, int] = {
    "Kaif": 90,
    "Sara": 85,
    "Ali": 88,
}

print(scores)
````

# 8. `tuple[...]`

A fixed tuple can describe each position.

```python
tuple[str, int]
```

````python
def employee_record() -> tuple[str, int]:
    return ("Kaif", 22)

name, age = employee_record()

print(name)
print(age)
````

# 9. Variable-Length Tuples

Use:

```python
tuple[int, ...]
```

for zero or more integers.

````python
def total(values: tuple[int, ...]) -> int:
    return sum(values)

numbers: tuple[int, ...] = (10, 20, 30, 40)

print(total(numbers))
````

# 10. Sets

Modern syntax also supports parameterized sets:

```python
set[str]
set[int]
```

````python
regions: set[str] = {"West", "East", "North"}

print(regions)
````

# 11. `Optional`

`Optional[T]` means `T` or `None`.

Modern Python can write the same idea as:

```python
T | None
```

For modern Python, prefer `T | None` when the project's Python version
supports it.

````python
def find_customer(customer_id: int) -> str | None:
    customers = {
        101: "Kaif",
        102: "Sara",
    }
    return customers.get(customer_id)

print(find_customer(101))
print(find_customer(999))
````

# 12. `Union`

A union represents multiple possible types.

Modern syntax:

```python
int | str
```

Traditional typing syntax:

```python
Union[int, str]
```

````python
def format_id(value: int | str) -> str:
    return f"ID-{value}"

print(format_id(101))
print(format_id("A102"))
````

# 13. `Any`

`Any` tells static type checkers to accept the value without a specific type
constraint.

Use it carefully because excessive `Any` reduces the benefits of static
typing.

````python
from typing import Any

def display(value: Any) -> None:
    print(value)

display(100)
display("Python")
display([1, 2, 3])
````

# 14. `Callable`

`Callable` describes a function or another callable object.

Example:

```python
Callable[[int, int], int]
```

means a callable receiving two integers and returning an integer.

````python
from collections.abc import Callable

def apply_operation(
    operation: Callable[[int, int], int],
    a: int,
    b: int,
) -> int:
    return operation(a, b)

def add(a: int, b: int) -> int:
    return a + b

print(apply_operation(add, 10, 20))
````

# 16. Type Aliases — Compatible Syntax

A type alias gives a meaningful name to a type.

For broad compatibility with Python versions before 3.12, use
`TypeAlias`:

```python
CustomerID: TypeAlias = int
```

This notebook intentionally uses this form instead of:

```python
type CustomerID = int
```

because the latter requires Python 3.12+.

````python
from typing import TypeAlias

CustomerID: TypeAlias = int
SalesRecord: TypeAlias = dict[str, float]

customer_id: CustomerID = 101

record: SalesRecord = {
    "revenue": 125000.0,
}

print(customer_id)
print(record)
````

# 17. Recursive JSON-Like Type Alias

A JSON-like value can contain:

- strings
- integers
- floats
- booleans
- `None`
- lists of JSON values
- dictionaries containing JSON values

For Python versions where the `type` statement is unavailable, define the
alias with `TypeAlias`.

````python
from typing import TypeAlias

JSONValue: TypeAlias = (
    str
    | int
    | float
    | bool
    | None
    | list["JSONValue"]
    | dict[str, "JSONValue"]
)

data: JSONValue = {
    "name": "Kaif",
    "scores": [90, 95, 88],
    "active": True,
}

print(data)
````

# 18. `TypedDict`

`TypedDict` describes the expected keys and value types of a dictionary.

This is especially useful for structured business or JSON-like records.

````python
from typing import TypedDict

class Customer(TypedDict):
    name: str
    age: int
    active: bool

customer: Customer = {
    "name": "Kaif",
    "age": 22,
    "active": True,
}

print(customer)
````

# 19. `NotRequired`

`NotRequired` can mark a `TypedDict` key as optional while other keys remain
required.

````python
from typing import NotRequired, TypedDict

class CustomerProfile(TypedDict):
    name: str
    email: str
    phone: NotRequired[str]

profile: CustomerProfile = {
    "name": "Kaif",
    "email": "kaif@example.com",
}

print(profile)
````

# 20. `Literal`

`Literal` restricts a value to a defined set of literal values.

Useful for:

- statuses
- modes
- configuration options
- controlled categories

````python
from typing import Literal

Status = Literal["pending", "approved", "rejected"]

def update_status(status: Status) -> str:
    return f"Status: {status}"

print(update_status("approved"))
````

# 21. `Final`

`Final` communicates that a name is intended not to be reassigned.

Static type checkers can detect inappropriate reassignment.

````python
from typing import Final

MAX_RETRIES: Final[int] = 3

print(MAX_RETRIES)
````

# 22. `ClassVar`

`ClassVar` identifies a class-level variable rather than an instance field.

````python
from typing import ClassVar

class Employee:
    company: ClassVar[str] = "DataTech"

    def __init__(self, name: str):
        self.name: str = name

employee = Employee("Kaif")

print(employee.name)
print(Employee.company)
````

# 23. `TypeVar`

`TypeVar` creates a generic type variable.

It is useful when the relationship between input and output types should be
preserved.

````python
from typing import TypeVar

T = TypeVar("T")

def first(items: list[T]) -> T:
    return items[0]

print(first([10, 20, 30]))
print(first(["Python", "SQL"]))
````

# 24. Generic Function

Generic functions can work with many types while preserving type information.

````python
from typing import TypeVar

T = TypeVar("T")

def duplicate(value: T) -> tuple[T, T]:
    return value, value

print(duplicate(10))
print(duplicate("Python"))
````

# 25. Generic Class

`Generic[T]` can be used to create reusable type-parameterized classes.

````python
from typing import Generic, TypeVar

T = TypeVar("T")

class Box(Generic[T]):
    def __init__(self, value: T):
        self.value = value

    def get(self) -> T:
        return self.value

number_box = Box(100)
text_box = Box("Python")

print(number_box.get())
print(text_box.get())
````

# 26. `Protocol`

A `Protocol` describes expected behavior.

This supports structural typing: an object can satisfy the protocol based on
its structure rather than inheriting from the protocol class.

````python
from typing import Protocol

class HasName(Protocol):
    name: str

def display_name(item: HasName) -> str:
    return item.name

class Employee:
    def __init__(self, name: str):
        self.name = name

print(display_name(Employee("Kaif")))
````

# 27. `Annotated`

`Annotated` attaches metadata to a type annotation.

The metadata can be interpreted by external tools or frameworks.

````python
from typing import Annotated

UserID = Annotated[int, "positive customer identifier"]

customer_id: UserID = 101

print(customer_id)
````

# 28. `TypeGuard`

`TypeGuard` can communicate type narrowing to static type checkers after a
predicate returns `True`.

````python
from typing import TypeGuard

def is_string_list(value: list[object]) -> TypeGuard[list[str]]:
    return all(isinstance(item, str) for item in value)

items: list[object] = ["Python", "SQL"]

if is_string_list(items):
    print([item.upper() for item in items])
````

# 29. `Self`

`Self` represents the current class type and is useful for fluent APIs and
methods returning the current instance.

````python
from typing import Self

class Builder:
    def set_name(self, name: str) -> Self:
        self.name = name
        return self

builder = Builder().set_name("Kaif")

print(builder.name)
````

# 30. Typed Classes

Type hints work naturally with class attributes, constructors, and methods.

````python
class Product:
    name: str
    price: float

    def __init__(self, name: str, price: float):
        self.name = name
        self.price = price

    def final_price(self, tax_rate: float) -> float:
        return self.price * (1 + tax_rate)

product = Product("Laptop", 65000.0)

print(product.name)
print(product.final_price(0.18))
````

# 31. Default Arguments with Type Hints

Annotations can be combined with default values.

````python
def create_user(
    name: str,
    role: str = "Analyst",
    active: bool = True,
) -> dict[str, object]:
    return {
        "name": name,
        "role": role,
        "active": active,
    }

print(create_user("Kaif"))
````

# 32. `*args` and `**kwargs`

Variable-length arguments can also be typed.

````python
def total(*values: float) -> float:
    return sum(values)

def create_record(**fields: str) -> dict[str, str]:
    return fields

print(total(10.5, 20.0, 30.5))
print(create_record(name="Kaif", role="Analyst"))
````

# 33. Type Hints for Analytics Records

Type aliases can make business data easier to understand.

````python
from typing import TypeAlias

SalesRecord: TypeAlias = dict[str, str | float]

def revenue(record: SalesRecord) -> float:
    value = record["amount"]

    if not isinstance(value, float):
        raise TypeError("amount must be a float")

    return value

record: SalesRecord = {
    "customer": "Kaif",
    "amount": 2500.0,
}

print(revenue(record))
````

# 34. Optional Lookup Pattern

A lookup can return an object or `None`.

Always handle the `None` case before using the value as the expected type.

````python
def customer_name(customer_id: int) -> str | None:
    customers = {101: "Kaif"}
    return customers.get(customer_id)

name = customer_name(101)

if name is not None:
    print(name.upper())
````

# 35. Forward References

A quoted annotation can refer to a class name in contexts where the class
name is not yet available during annotation evaluation.

````python
class Employee:
    def __init__(self, manager: "Employee | None" = None):
        self.manager = manager

employee = Employee()

print(employee.manager)
````

# 36. Recursive Data Structures

Recursive type aliases are useful for nested structures such as JSON data.

This version deliberately uses `TypeAlias` so it does not require the Python
3.12 `type` statement.

````python
from typing import TypeAlias

JSONValue: TypeAlias = (
    str
    | int
    | float
    | bool
    | None
    | list["JSONValue"]
    | dict[str, "JSONValue"]
)

data: JSONValue = {
    "customer": {
        "name": "Kaif",
        "scores": [90, 95, 88],
    },
    "active": True,
}

print(data)
````

# 38. Runtime Annotations

Function annotations can be inspected at runtime through metadata.

This is different from automatic runtime type validation.

````python
def calculate(price: float, quantity: int) -> float:
    return price * quantity

print(calculate.__annotations__)
````

# 39. Type Hints vs Runtime Validation

```text
TYPE HINT
   ↓
DOCUMENTATION
   ↓
IDE SUPPORT
   ↓
STATIC ANALYSIS
```

Runtime validation is a separate concern.

If an application must reject invalid external input, validate the input
explicitly or use an appropriate runtime validation library.

# 40. Common Mistakes

### 1. Thinking annotations enforce types

They normally do not.

### 2. Using `Any` everywhere

This removes useful type information.

### 3. Confusing `Optional[T]` with a default

`Optional[T]` means `T | None`; it does not mean the argument has a default.

### 4. Overengineering annotations

Simple code should remain simple.

### 5. Using Python 3.12-only syntax on older Python

For example:

```python
type CustomerID = int
```

requires Python 3.12+.

For compatibility, use:

```python
from typing import TypeAlias

CustomerID: TypeAlias = int
```

### 6. Forgetting to handle `None`

If a function returns `T | None`, check for `None` before using it as `T`.

# 41. Best Practices

- Type important public functions.
- Use precise types where they improve clarity.
- Prefer modern built-in generics such as `list[int]`.
- Prefer `T | None` on supported Python versions.
- Use `TypedDict` for known dictionary structures.
- Use `Callable` for function arguments.
- Use `TypeVar` when relationships between types matter.
- Use `Protocol` when behavior matters more than inheritance.
- Avoid unnecessary `Any`.
- Keep annotations readable.
- Use static type checking on larger projects.
- Keep the project's supported Python version clear.

# 🚀 Project 1 — 01 — Typed Sales Analytics

## Business Problem

Calculate total and average sales with precise type annotations.

## Architecture

```text
INPUT
 ↓
TYPE HINTS
 ↓
STRUCTURE / VALIDATION
 ↓
BUSINESS LOGIC
 ↓
TYPED OUTPUT
 ↓
REPORT
```

## Implementation

````python
def sales_total(values: list[float]) -> float:
    return sum(values)

def sales_average(values: list[float]) -> float:
    if not values:
        return 0.0
    return sales_total(values) / len(values)

sales: list[float] = [1200.0, 2500.0, 1800.0, 3200.0]

print("Total:", sales_total(sales))
print("Average:", sales_average(sales))
````

```text
 ↓
TYPED OUTPUT
```

## Implementation

````python
from collections.abc import Callable

def apply_transform(
    values: list[float],
    transform: Callable[[float], float],
) -> list[float]:
    return [transform(value) for value in values]

def add_tax(value: float) -> float:
    return value * 1.18

sales = [1000.0, 2000.0, 3000.0]

print(apply_transform(sales, add_tax))
````

# 🚀 Project 9 — 09 — Typed Data Pipeline

## Business Problem

Build a typed filter → transform → aggregate workflow.

## Architecture

```text
INPUT
 ↓
TYPE HINTS
 ↓
STRUCTURE / VALIDATION
 ↓
BUSINESS LOGIC
 ↓
TYPED OUTPUT
 ↓
REPORT
```

## Implementation

````python
from collections.abc import Callable

def filter_values(
    values: list[float],
    condition: Callable[[float], bool],
) -> list[float]:
    return [value for value in values if condition(value)]

def transform_values(
    values: list[float],
    transform: Callable[[float], float],
) -> list[float]:
    return [transform(value) for value in values]

sales = [1000.0, -200.0, 2500.0, 3200.0, 800.0]

valid = filter_values(sales, lambda value: value > 0)
taxed = transform_values(valid, lambda value: value * 1.18)

print("Valid:", valid)
print("Taxed:", taxed)
print("Total:", sum(taxed))
````

# 🚀 Project 10 — 10 — Typed Analytics Reporting System

## Business Problem

Build an end-to-end typed reporting workflow with structured records, validation, transformation, and reporting.

## Architecture

```text
INPUT
 ↓
TYPE HINTS
 ↓
STRUCTURE / VALIDATION
 ↓
BUSINESS LOGIC
 ↓
TYPED OUTPUT
 ↓
REPORT
```

## Implementation

````python
from collections.abc import Callable
from typing import TypedDict

class Sale(TypedDict):
    sale_id: int
    customer: str
    amount: float

class Report(TypedDict):
    count: int
    total: float
    average: float

def valid_sales(sales: list[Sale]) -> list[Sale]:
    return [sale for sale in sales if sale["amount"] > 0]

def transform_sales(
    sales: list[Sale],
    transform: Callable[[float], float],
) -> list[Sale]:
    return [
        {
            **sale,
            "amount": transform(sale["amount"]),
        }
        for sale in sales
    ]

def build_report(sales: list[Sale]) -> Report:
    if not sales:
        return {"count": 0, "total": 0.0, "average": 0.0}

    total = sum(sale["amount"] for sale in sales)

    return {
        "count": len(sales),
        "total": total,
        "average": total / len(sales),
    }

sales: list[Sale] = [
    {"sale_id": 1, "customer": "Kaif", "amount": 65000.0},
    {"sale_id": 2, "customer": "Sara", "amount": 18000.0},
    {"sale_id": 3, "customer": "Ali", "amount": -500.0},
    {"sale_id": 4, "customer": "Maya", "amount": 42000.0},
]

cleaned = valid_sales(sales)
with_tax = transform_sales(cleaned, lambda amount: amount * 1.18)
report = build_report(with_tax)

print("Report:", report)
````

```python
scores: list[int] = [90, 95, 88]
```

### Dictionary

```python
scores: dict[str, int] = {"Kaif": 90}
```

### Optional

```python
def find_name() -> str | None:
    ...
```

### Type Alias — Compatible Form

```python
from typing import TypeAlias

CustomerID: TypeAlias = int
```

### Typed Dictionary

```python
from typing import TypedDict

class Customer(TypedDict):
    name: str
    age: int
```

```python
from typing import TypeVar

    return values[0]
```
