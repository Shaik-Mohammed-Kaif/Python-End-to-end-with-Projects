# 1. What Is Git?

**Git** is a distributed version control system.

It records changes to files so you can:

- track history
- compare changes
- restore earlier versions
- create branches
- collaborate
- manage project development

Think of Git as the version-history system for your project.

# 2. What Is GitHub?

**GitHub** is a platform for hosting Git repositories and collaborating around
them.

```text
LOCAL COMPUTER
     ↓
    GIT
     ↓
LOCAL REPOSITORY
     ↓
   PUSH
     ↓
  GITHUB
     ↓
COLLABORATION
```

Git and GitHub are related, but they are not the same thing.

- Git = version control tool
- GitHub = online platform/service for Git repositories

# 3. What Is a Repository?

A repository, or **repo**, contains a project's files plus Git's version
history.

A Python repository may look like:

```text
my-python-project/
├── src/
├── tests/
├── .gitignore
├── README.md
├── pyproject.toml
└── requirements.txt
```

# 7. `git status`

Check the current state:

```bash
git status
```

It helps you see:

- untracked files

# 8. Working Tree → Staging → Commit

The basic Git model:

```text
WORKING TREE
     ↓
 git add
     ↓
STAGING AREA
     ↓
 git commit
     ↓
REPOSITORY HISTORY
```

This is one of the most important Git concepts to understand.

# 9. `git add`

Stage a file:

```bash
git add README.md
```

Stage multiple selected files:

```bash
git add src/app.py tests/test_app.py
```

Stage all current changes:

```bash
git add .
```

Review `git status` before committing, especially when using `git add .`.

# 10. Commit

A commit records a snapshot of staged changes.

```bash
git commit -m "Add sales calculation"
```

Good commit messages communicate the change clearly.

Prefer:

```text
Add customer validation
Fix revenue calculation
Add sales tests
```

over:

```text
update
changes
final
```

# 11. Git Log

View commit history:

```bash
git log
```

A compact view:

```bash
git log --oneline
```

This helps you understand how the project evolved.

# 12. Git Diff

See unstaged changes:

```bash
git diff
```

See staged changes:

```bash
git diff --staged
```

Use diffs before committing to review what will actually be recorded.

# 13. GitHub Remote

A remote connects the local repository to an online repository.

Common commands:

```bash

```

Typical architecture:

```text

       │
       │ push
       ▼
GITHUB REMOTE
       │

       ▼

```

# 14. Push

Send local commits to a remote repository:

```bash
git push
```

For the first push on a new branch, you may use:

```bash
git push -u origin main
```

The exact default branch name depends on your repository configuration.

# 15. Pull

Get remote changes and integrate them into your current branch:

```bash
git pull
```

Before pulling, understand whether you have local uncommitted changes that
could complicate the operation.

# 16. Clone

To copy an existing GitHub repository:

```bash

```

Then:

```bash
cd project-name
```

This is commonly how you begin working with an existing project.

# 17. Branches

A branch is an independent line of development.

```text
main
  │
  ├──── feature-login
  │
  └──── feature-report
```

Create a branch:

```bash

```

List branches:

```bash
git branch
```

Switch branches:

```bash
git switch main
```

# 18. Why Use Branches?

Branches allow you to work on a feature without changing the main development
line immediately.


# 19. Merge

Merge combines changes from one branch into another.

Example:

```bash
git switch main
git merge feature-login
```

The exact result depends on the branch history.

# 20. Merge Conflicts

A conflict can occur when Git cannot automatically combine changes.

```text
BRANCH A
       CONFLICT
   /
BRANCH B
```

General workflow:

```text
MERGE
 ↓
CONFLICT
 ↓
OPEN FILE
 ↓
CHOOSE / EDIT CORRECT CONTENT
 ↓
REMOVE CONFLICT MARKERS
 ↓
TEST
 ↓
git add
 ↓
COMPLETE MERGE
```

Never blindly choose a side without understanding the code.

# 22. `.gitignore`

`.gitignore` tells Git which files/directories should normally not be tracked.

For Python, common entries include:

```gitignore

__pycache__/
*.py[cod]
.ipynb_checkpoints/
.pytest_cache/

.env
```

Do not commit secrets such as API keys or passwords.

# 23. Why `.venv` Should Usually Be Ignored

A virtual environment contains installed packages and environment-specific
files.

Do not normally upload the entire `.venv/` directory to GitHub.

Instead:

```text
PROJECT
 ├── source code
 ├── requirements / project metadata
 └── .venv/  ← local environment, ignored
```

# 24. Python Virtual Environment

Create:

```bash
python -m venv .venv
```

Activate on Windows PowerShell:

```powershell
.\.venv\Scripts\Activate.ps1
```

Activate on Windows Command Prompt:

```cmd
.venv\Scripts\activate
```

Activate on macOS/Linux:

```bash
source .venv/bin/activate
```

The activation command depends on your operating system and shell.

# 25. `requirements.txt`

A `requirements.txt` file records Python package dependencies.

Example:

```text

```

```bash
python -m pip install -r requirements.txt
```

Generate a snapshot from an environment:

```bash
python -m pip freeze > requirements.txt
```

For modern Python packaging, `pyproject.toml` can also be used to declare
project metadata and dependencies.

# 33. Secrets and Environment Variables

Never commit secrets such as:

```text
API_KEY
PASSWORD
DATABASE_PASSWORD
ACCESS_TOKEN
```

Use environment variables or an appropriate secrets-management solution.

Example local `.env` entry:

```text
API_KEY=your-local-secret
```

Then ignore `.env`:

```gitignore
.env
```

A `.gitignore` rule does not remove a secret that was already committed.
If a secret is exposed, rotate/revoke it immediately.

A clean repository makes the project easier to review and use.

# 35. Common Git Mistakes

### Mistake 1 — Committing `.venv`

Ignore it.

### Mistake 2 — Committing secrets

Use environment variables/secrets management.

### Mistake 3 — Huge unclear commits

Keep commits focused.

### Mistake 4 — Working directly on main for every feature

Use branches when the workflow calls for isolated development.

### Mistake 5 — Never checking `git diff`

Review changes before committing.

### Mistake 6 — Ignoring failed tests

Fix or investigate them before merging.

### Mistake 7 — Force-pushing carelessly

Understand the consequences before rewriting shared history.

# 🚀 Project 1 — 01 — Initialize a Python Git Repository

## Business Problem

Create a clean Python project and initialize Git.

## Architecture

```text
PYTHON CODE
 ↓
TESTS
 ↓
GIT TRACKING
 ↓
COMMIT
 ↓
BRANCH
 ↓
GITHUB
 ↓
PULL REQUEST
 ↓
REVIEW
 ↓
MERGE
```

## Implementation

````python
from pathlib import Path

project = Path("git_python_project")

directories = [
    project / "src",
    project / "tests",
]

for directory in directories:
    directory.mkdir(parents=True, exist_ok=True)

(project / "README.md").write_text(
    "# Git + Python Demo\n",
    encoding="utf-8",
)

(project / ".gitignore").write_text(
    ".venv/\n"
    "__pycache__/\n"
    ".pytest_cache/\n"
    ".env\n",
    encoding="utf-8",
)

print("Project created:", project.resolve())
print("Folders:", [str(path) for path in directories])
````

# 🚀 Project 9 — 09 — Clean GitHub-Ready Analytics Repository

## Business Problem

Build the project files for a small analytics repository with source, tests, documentation, and Git hygiene.

## Architecture

```text
PYTHON CODE
 ↓
TESTS
 ↓
GIT TRACKING
 ↓
COMMIT
 ↓
BRANCH
 ↓
GITHUB
 ↓
PULL REQUEST
 ↓
REVIEW
 ↓
MERGE
```

## Implementation

````python
from pathlib import Path

root = Path("analytics_github_project")

files = {
    root / "src" / "analytics" / "__init__.py": "",
    root / "src" / "analytics" / "metrics.py":
        'def total_sales(values: list[float]) -> float:\n'
        '    """Return total sales."""\n'
        '    return sum(values)\n',
    root / "tests" / "test_metrics.py":
        'from src.analytics.metrics import total_sales\n\n'
        'def test_total_sales():\n'
        '    assert total_sales([100.0, 200.0]) == 300.0\n',
    root / "README.md":
        '# Analytics GitHub Project\n\n'
        'A small Python analytics project.\n',
    root / ".gitignore":
        '.venv/\n__pycache__/\n.pytest_cache/\n.env\n',
    root / "requirements.txt":
        '# Add project dependencies here\n',
}

for path, content in files.items():
    path.parent.mkdir(parents=True, exist_ok=True)
    path.write_text(content, encoding="utf-8")

print("Repository prepared:", root.resolve())

for path in sorted(root.rglob("*")):
    if path.is_file():
        print(path.relative_to(root))
````
