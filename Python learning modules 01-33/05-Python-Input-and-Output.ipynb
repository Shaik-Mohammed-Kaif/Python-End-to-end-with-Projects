## 1. What is Input & Output?

**Input** means taking data from the user.

**Output** means displaying information to the user.

Python provides two important built-in functions for basic input and output:

```python
input()
```

and

```python
print()
```

### Basic Flow

```text
User Input
    ↓
input()
    ↓
Python Program
    ↓
Processing
    ↓
print()
    ↓
Output
```

---

# 2. `print()` Function

The `print()` function is used to display information on the screen.

### Syntax

```python
print(value)
```

### Example

```python
print("Hello, Python!")
```

### Output

```text
Hello, Python!
```

---

## Printing Numbers

```python
print(100)
print(25.5)
print(10 + 20)
```

### Output

```text
100
25.5
30
```

---

## Printing Multiple Values

You can provide multiple values to `print()`.

```python
name = "Kaif"
age = 22

print(name, age)
```

### Output

```text
Kaif 22
```

---

# 3. `input()` Function

The `input()` function is used to receive information from the user.

### Syntax

```python
input("message")
```

### Example

```python
name = input("Enter your name: ")
print(name)
```

### Example Interaction

```text
Enter your name: Mohammed Kaif
Mohammed Kaif
```

---

# 4. Important Point About `input()`

The `input()` function returns the user's input as a **string**.

For example:

```python
age = input("Enter your age: ")

print(type(age))
```

If the user enters:

```text
22
```

The output will be:

```text
<class 'str'>
```

Even though `22` looks like a number, Python receives it as text.

# 7. Output Formatting

Output formatting means presenting information in a clear and readable way.

Instead of:

```python
name = "Kaif"
age = 22

print(name, age)
```

You can create a more meaningful output:

```python
name = "Kaif"
age = 22

print("Name:", name)
print("Age:", age)
```

### Output

```text
Name: Kaif
Age: 22
```

# 8. Escape Characters

Escape characters are special sequences that allow you to control how text is displayed.

They usually begin with a backslash:

```text
\
```

Common escape characters include:

| Escape Character | Meaning      |
| ---------------- | ------------ |
| `\n`             | New line     |
| `\t`             | Tab          |
| `\\`             | Backslash    |
| `\'`             | Single quote |
| `\"`             | Double quote |

---

# 9. New Line — `\n`

The `\n` escape character moves the following text to a new line.

### Example

```python
print("Python\nProgramming")
```

### Output

```text
Python
Programming
```

---

## Multiple New Lines

```python
print("Name: Kaif\nAge: 22\nRole: Data Scientist")
```

### Output

```text
Name: Kaif
Age: 22
Role: Data Scientist
```

---

# 10. Tab — `\t`

The `\t` escape character inserts a tab space.

### Example

```python
print("Python\tProgramming")
```

### Output

```text
Python  Programming
```

---

## Practical Example

```python
print("Name:\tMohammed Kaif")
print("Role:\tData Scientist")
print("Skill:\tPython")
```

### Output

```text
Name:   Mohammed Kaif
Role:   Data Scientist
Skill:  Python
```

---

# 11. Backslash — `\\`

To display a backslash character, use `\\`.

```python
print("C:\\Users\\Kaif\\Documents")
```

### Output

```text
C:\Users\Kaif\Documents
```

---

# 12. Quotes Inside Strings

You can use escape characters when quotes need to appear inside a string.

```python

```

### Output

```text

```

You can also use single quotes:

```python
print('He said, "Python is powerful."')
```

### Output

```text

```

# 13. String Formatting

String formatting allows you to insert values into text.

Consider:

```python
name = "Kaif"
age = 22

print("My name is", name, "and I am", age, "years old.")
```

This works, but modern Python provides cleaner approaches.

One of the most useful approaches is the **f-string**.

---

# 14. f-Strings

An f-string allows you to insert variables and expressions directly inside a string.

### Syntax

```python
f"text {variable}"
```

### Example

```python
name = "Kaif"

print(f"Hello {name}")
```

### Output

```text
Hello Kaif
```

---

# 15. f-String With Multiple Variables

```python
name = "Mohammed Kaif"
age = 22
role = "Data Scientist"

print(f"My name is {name}.")
print(f"I am {age} years old.")
print(f"I am learning {role}.")
```

### Output

```text
My name is Mohammed Kaif.
I am 22 years old.
I am learning Data Scientist.
```

---

# 16. f-Strings With Expressions

You can also place expressions inside `{}`.

```python
num1 = 100
num2 = 200

print(f"Total: {num1 + num2}")
```

### Output

```text
Total: 300
```

---

# 17. f-Strings With Calculations


# 18. Formatting Decimal Values

f-strings can format numerical values.

For example:

```python
price = 499.98765

print(f"Price: {price:.2f}")
```

### Output

```text
Price: 499.99
```

Here:

```text
.2f
```

means displaying the number with **2 digits after the decimal point**.

# 25. Common Mistakes

## Mistake 1 — Forgetting Type Conversion

Incorrect:

```python
age = input("Enter age: ")

print(age + 5)
```

This causes a type error because `input()` returns a string.

Correct:

```python
age = int(input("Enter age: "))

print(age + 5)
```

---

## Mistake 2 — Forgetting the `f` in f-String

Incorrect:

```python
name = "Kaif"

print("Hello {name}")
```

Output:

```text
Hello {name}
```

Correct:

```python
print(f"Hello {name}")
```

Output:

```text
Hello Kaif
```

---

## Mistake 3 — Incorrect Escape Character

Incorrect:

```python
print("Python\nProgramming")
```

This intentionally creates a new line.

If you want the literal characters `\n`, use:

```python
print("Python\\nProgramming")
```

Output:

```text
Python\nProgramming
```

---

# 26. Best Practices

Follow these practices when working with Python input and output:

* Use meaningful prompts with `input()`.
* Convert user input to the correct data type.
* Use f-strings for readable string formatting.
* Use `\n` when separating sections of output.
* Use `\t` when simple tabular alignment is useful.
* Format decimal values when displaying financial or measurement data.
* Keep output clear and easy to understand.
* Validate user input in real applications.
* Avoid unnecessarily complicated string formatting.

# 28. Quick Reference

| Feature   | Purpose            | Example                |     |
| --------- | ------------------ | ---------------------- | --- |
| `print()` | Display output     | `print("Hello")`       |     |
| `input()` | Receive user input | `input("Name: ")`      |     |
| `int()`   | Convert to integer | `int("10")`            |     |
| `float()` | Convert to float   | `float("10.5")`        |     |
| `\n`      | New line           | `"Hello\nWorld"`       |     |
| `\t`      | Tab                | `"Name:\tKaif"`        |     |
| `\\`      | Backslash          | `"C:\\Users"`          |     |
| `f""`     | f-string           | `f"Hello {name}"`      |     |
| `sep`     | Change separator   | `print(a, b, sep="     | ")` |
| `end`     | Change line ending | `print("Hi", end=" ")` |     |

* Use `int()` for integer conversion.

* `print()` displays information.
* `\n` creates a new line.
* `\t` creates a tab.
* f-strings provide clean and modern string formatting.
* `sep` controls separators.
* `end` controls what happens after `print()`.
