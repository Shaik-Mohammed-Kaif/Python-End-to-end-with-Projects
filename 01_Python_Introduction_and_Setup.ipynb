# 🐍 WHAT IS PYTHON?

</div>

Python is a **high-level, general-purpose programming language** known for its readable syntax and broad ecosystem.

Python is used to build everything from small automation scripts to large data platforms, web applications, machine learning systems, and scientific applications.

## 🔑 Key Characteristics

| Characteristic | Meaning |
|---|---|
| 🧑‍💻 High-level | Easier for humans to read and write |
| 🎯 General-purpose | Used across many domains |
| 📖 Readable | Clean and expressive syntax |
| 🧩 Interpreted | Python code is executed by the Python runtime |
| 🏗️ Object-oriented | Supports object-oriented programming |
| 🔌 Extensible | Can work with other technologies and languages |
| 📦 Rich ecosystem | Thousands of libraries and packages |
| 🌍 Open source | Python is freely available |

# 📦 WHAT IS pip?

</div>

`pip` is Python's package installer.

It allows developers to install and manage Python packages from the Python Package Index ecosystem.

Official documentation:

[📚 pip Documentation](https://pip.pypa.io/en/stable/)

---

## Basic Commands

### Check pip

Windows:

```bash
python -m pip --version
```

macOS/Linux:

```bash
python3 -m pip --version
```

---

### Install a Package

```bash
python -m pip install pandas
```

---

### Install Multiple Packages

```bash
python -m pip install numpy pandas matplotlib
```

---

### Upgrade a Package

```bash
python -m pip install --upgrade pandas
```

---

### Uninstall a Package

```bash
python -m pip uninstall pandas
```

---

### View Installed Packages

```bash
python -m pip list
```

---

### Show Package Information

```bash
python -m pip show pandas
```

---

### Export Dependencies

```bash
python -m pip freeze > requirements.txt
```

---

### Install from requirements.txt

```bash
python -m pip install -r requirements.txt
```

---

## Why Use `python -m pip`?

It explicitly associates pip with the Python interpreter you are invoking.

This is useful when multiple Python installations or environments exist.

# 🧪 VIRTUAL ENVIRONMENTS

</div>

A virtual environment provides an isolated Python environment for a project.

## Why?

Imagine:

```text
Project A
 └── Pandas 1.x

Project B
 └── Pandas 2.x
```

Installing everything globally can create dependency conflicts.

A virtual environment helps keep project dependencies separate.

---

## Windows

Create:

```bash
python -m venv .venv
```

Activate:

```bash
.venv\Scripts\activate
```

Deactivate:

```bash
deactivate
```

---

## macOS

Create:

```bash
python3 -m venv .venv
```

Activate:

```bash
source .venv/bin/activate
```

Deactivate:

```bash
deactivate
```

# 🧩 MODULES vs PACKAGES vs LIBRARIES

</div>

These terms are related but should not be treated as identical.

---

## Module

A Python module is typically a Python file containing reusable code.

Example:

```text
calculator.py
```

It might contain:

```python
def add(a, b):
    return a + b
```

Then another file can import it.

```python
import calculator
```

---

## Package

A package organizes related Python modules into a directory structure.

Conceptually:

```text
analytics/
│
├── cleaning.py
├── statistics.py
└── visualization.py
```

---

## Library

"Library" is a broader term for reusable software functionality.

For example, Pandas is commonly described as a Python library.

---

## Simple Mental Model

```text
Module
   ↓
Reusable Python Code

Package
   ↓
Collection / Organization of Modules

Library
   ↓
Reusable Software Functionality
```

The exact terminology can vary by context, so focus on understanding the structure and how imports work.

# 🔑 PYTHON KEYWORDS

</div>

Python has reserved words called **keywords**.

Keywords have special meanings in the Python language and generally cannot be used as ordinary variable names.

Examples include:

```text
if
else
elif
for
while
in
is
not
and
or
def
return
class
import
from
try
except
finally
with
as
lambda
yield
True
False
None
```

---

## View Keywords Programmatically

Python provides the `keyword` module.

```python
import keyword

print(keyword.kwlist)
```

---

## Why Learn Keywords?

Keywords form the building blocks of Python syntax.

For example:

```python
if
```

is used for conditions.

```python
for
```

is used for iteration.

```python
def
```

is used to define functions.

```python
class
```

is used to define classes.

```python
import
```

is used to import modules.

````python
import keyword

print("Python Keywords:")
print(keyword.kwlist)

print("\nNumber of keywords:", len(keyword.kwlist))
````

# 🔌 IMPORTS IN PYTHON

</div>

Python uses `import` to access functionality from modules and packages.

---

## Import a Module

```python
import math

print(math.sqrt(25))
```

---

## Import Specific Functionality

```python
from math import sqrt

print(sqrt(25))
```

---

## Use an Alias

```python
import pandas as pd
```

Now:

```python
pd.read_csv(...)
```

instead of:

```python
pandas.read_csv(...)
```

---

## Common Industry Convention

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

These aliases are conventions widely encountered in data-related Python code.

````python
import math
import datetime

number = 25

print("Square root:", math.sqrt(number))
print("Current date:", datetime.date.today())
````
