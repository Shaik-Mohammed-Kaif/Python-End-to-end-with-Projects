Python's `if` statement selects code based on a condition; in an `if`/`elif` chain, conditions are evaluated in order and the first true branch is executed. Indentation defines each controlled block.

# 1. What Is a Conditional Statement?

A conditional statement allows a program to make decisions.

### Basic Syntax

```python
if condition:
    statement
```

The condition is evaluated as true or false.

### Example

```python
age = 20

if age >= 18:
    print("Adult")
```

**Important:** Python uses indentation to define the block controlled by `if`, and the condition ends with `:`.

````python
age = 20

if age >= 18:
    print("Adult")
````

# 2. `if` Statement

Use `if` when an action should happen only when a condition is true.

```python
marks = 85

if marks >= 40:
    print("Student passed")
```

````python
marks = 85

if marks >= 40:
    print("Student passed")
````

# 3. `if` + `else`

Use `else` for the alternative path.

```python
if condition:
    ...
else:
    ...
```

Example:

````python
marks = 32

if marks >= 40:
    print("Pass")
else:
    print("Fail")
````

# 4. `if` + `elif` + `else`

Use `elif` when several conditions are possible.

Python checks the conditions from top to bottom and executes the first true branch.

```python
if condition_1:
    ...
elif condition_2:
    ...
else:
    ...
```

````python
marks = 78

if marks >= 90:
    grade = "A+"
elif marks >= 80:
    grade = "A"
elif marks >= 70:
    grade = "B"
elif marks >= 60:
    grade = "C"
else:
    grade = "F"

print("Grade:", grade)
````

# 5. Comparison Operators

| Operator | Meaning |
|---|---|
| `==` | Equal |
| `!=` | Not equal |
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal |
| `<=` | Less than or equal |

These operators commonly produce Boolean results used by conditions.

````python
age = 25

print(age == 25)
print(age > 18)
print(age < 18)
print(age != 30)
````

# 6. Logical Operators

### `and`
Both conditions must be true.

### `or`
At least one condition must be true.

### `not`
Reverses a condition.

Example:

```python
if age >= 18 and has_id:
    ...
```

Python's Boolean operations are defined by `and`, `or`, and `not`. Note that `and` and `or` can return operand values, not only literal Boolean values.

````python
age = 22
has_id = True

if age >= 18 and has_id:
    print("Entry allowed")

is_blocked = False

if not is_blocked:
    print("Account is active")
````

# 7. Multiple Conditions

Conditions can be combined to represent business rules.

Example: a customer receives a discount when they are at least 18 and have spent at least ₹5,000.

````python
age = 25
purchase = 7000

if age >= 18 and purchase >= 5000:
    print("Discount eligible")
else:
    print("Not eligible")
````

# 8. Nested Conditions

A nested condition is an `if` statement inside another conditional block.

### Syntax

```python
if condition:
    if another_condition:
        ...
```

Use nested conditions when a second decision depends on the first.

````python
age = 22
has_id = True

if age >= 18:
    if has_id:
        print("Entry allowed")
    else:
        print("ID required")
else:
    print("Underage")
````

# 9. Truthy and Falsy Values

Python conditions can use values directly.

Common false values include:
- `False`
- `None`
- numeric zero
- empty strings
- empty containers

Most other values are treated as true.

Example:

````python
name = "Kaif"

if name:
    print("Name exists")

empty_name = ""

if not empty_name:
    print("Name is empty")
````

# 10. Conditional Expression

Python supports a compact conditional expression:

```python
value_if_true if condition else value_if_false
```

Example:

````python
age = 20

status = "Adult" if age >= 18 else "Minor"

print(status)
````

# 11. Common Mistakes

### Missing colon

Wrong:
```python
if age >= 18
```

Correct:
```python
if age >= 18:
    print("Adult")
```

### Wrong indentation

Wrong:
```python
if age >= 18:
print("Adult")
```

Correct:
```python
if age >= 18:
    print("Adult")
```

### Assignment instead of comparison

Wrong:
```python
if age = 18:
```

Correct:
```python
if age == 18:
```

### Best Practice

Keep conditions readable. Avoid unnecessary deep nesting and order overlapping conditions carefully.

# 🚀 Project 01 — Grade Calculator

## Problem

Accept marks and determine the grade and performance.

### Rules

```text
90–100 → A+
80–89  → A
70–79  → B
60–69  → C
40–59  → D
Below 40 → F
```

Reject values outside `0–100`.

````python
# Project 01 — Grade Calculator

marks = float(input("Enter your marks (0-100): "))

if marks < 0 or marks > 100:
    print("Invalid marks")
elif marks >= 90:
    print("Grade: A+")
    print("Performance: Outstanding")
elif marks >= 80:
    print("Grade: A")
    print("Performance: Excellent")
elif marks >= 70:
    print("Grade: B")
    print("Performance: Good")
elif marks >= 60:
    print("Grade: C")
    print("Performance: Satisfactory")
elif marks >= 40:
    print("Grade: D")
    print("Performance: Needs Improvement")
else:
    print("Grade: F")
    print("Performance: Fail")
````

# 🚀 Project 02 — Age Checker

## Rules

```text
0–12   → Child
13–17  → Teenager
18–59  → Adult
60+    → Senior
```

Reject negative ages.

````python
# Project 02 — Age Checker

age = int(input("Enter your age: "))

if age < 0:
    print("Invalid age")
elif age <= 12:
    print("Category: Child")
elif age <= 17:
    print("Category: Teenager")
elif age <= 59:
    print("Category: Adult")
else:
    print("Category: Senior")
````

Create a basic educational login simulator.

Demo credentials:

Username: admin

Real applications should not store plaintext passwords.

# 🚀 Project 04 — ATM Condition Simulator

Requirements:
- Start with an account balance.
- Ask for withdrawal amount.
- Reject zero/negative withdrawals.
- Reject withdrawals above the balance.
- Approve valid withdrawals.
- Display the remaining balance.

````python
# Project 04 — ATM Condition Simulator

balance = 10000.0

print(f"Available Balance: ₹{balance:.2f}")

withdrawal = float(input("Enter withdrawal amount: ₹"))

if withdrawal <= 0:
    print("Invalid withdrawal amount")
elif withdrawal > balance:
    print("Insufficient balance")
else:
    balance -= withdrawal
    print("Withdrawal successful")
    print(f"Remaining Balance: ₹{balance:.2f}")
````

# 🚀 Project 05 — Eligibility Checker

A student is eligible for a fictional training scholarship when:
- Age is between 18 and 30
- Score is at least 70
- The prerequisite is completed

Use logical conditions and clear decision branches.

````python
# Project 05 — Eligibility Checker

age = int(input("Enter age: "))
score = float(input("Enter score: "))
prerequisite = input("Prerequisite completed? (yes/no): ").strip().lower()

if age < 18 or age > 30:
    print("Not eligible: age requirement not satisfied")
elif score < 70:
    print("Not eligible: minimum score is 70")
elif prerequisite == "yes":
    print("Eligible for the scholarship")
else:
    print("Not eligible: prerequisite is required")
````

# ⚡ Bonus Project — Electricity Bill Calculator

Educational tariff:

```text
0–100 units    → ₹5 per unit
101–200 units  → ₹7 per unit
201+ units     → ₹10 per unit
```

The calculation applies each rate progressively by slab.

````python
# Bonus — Electricity Bill Calculator

units = float(input("Enter electricity units: "))

if units < 0:
    print("Invalid units")
elif units <= 100:
    bill = units * 5
elif units <= 200:
    bill = (100 * 5) + ((units - 100) * 7)
else:
    bill = (100 * 5) + (100 * 7) + ((units - 200) * 10)

if units >= 0:
    print(f"Electricity Bill: ₹{bill:.2f}")
````

# 🧠 Conditional Statements — Quick Reference

| Structure | Use |
|---|---|
| `if` | Run code when a condition is true |
| `if / else` | Choose between two paths |
| `if / elif / else` | Choose among multiple paths |
| Nested `if` | Make a decision inside another decision |
| `and` | All combined conditions must be true |
| `or` | At least one condition must be true |
| `not` | Negate a condition |
| Conditional expression | Compact `if/else` expression |

### Core Pattern

```python
if condition:
    action
elif another_condition:
    another_action
else:
    default_action
```
