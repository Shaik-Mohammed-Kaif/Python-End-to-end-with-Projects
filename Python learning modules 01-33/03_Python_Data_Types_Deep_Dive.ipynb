# 📚 What Is a Data Type?

A **data type** tells Python what kind of value an object represents.

For example:

| Value | Data Type |
|---|---|
| `25` | `int` |
| `25.5` | `float` |
| `"Python"` | `str` |
| `True` | `bool` |
| `3 + 4j` | `complex` |
| `None` | `NoneType` |

Python is **dynamically typed**, which means we do not have to explicitly declare the type of a variable before assigning a value.

### Example

```python
age = 25
name = "Alex"
salary = 45000.50

````python
age = 25
name = "Alex"
salary = 45000.50
is_active = True

print(age)
print(name)
print(salary)
print(is_active)
````

# 🔍 Checking Data Types with `type()`

Python provides the built-in `type()` function to identify the type of an object.

### Syntax

```python
type(value)

````python
age = 25

type(age)
````

````python
age = 25
name = "Alex"
salary = 45000.50
is_active = True

print(type(age))
print(type(name))
print(type(salary))
print(type(is_active))
````

# 🔢 1. Integer — `int`

An integer represents **whole numbers** without a decimal component.

### Examples

```python
10
25
-100
0

Integers are commonly used for:

- Age
- Quantity
- Number of employees
- Product stock
- Years
- Counts

````python
# Example :-
employee_count = 250

print("The value is an integer :", employee_count)
````

````python
age = 25
employee_count = 250
product_stock = 1000
temperature = -5

print(age)
print(employee_count)
print(product_stock)
print(temperature)

print(type(age))
print(type(employee_count))
print(type(product_stock))
print(type(temperature))
````

# 💰 2. Float — `float`

A floating-point number represents a number containing a decimal value.

### Examples

```python
10.5
99.99
45000.75
-12.5

### Floats are commonly used for:

- Salary
- Price
- Percentage
- Weight
- Measurements
- Ratings

## Real-World Example

````python
product_price = 1499.99

print("Here Product price is :", product_price)
````

````python
product_price = 1499.99
salary = 55000.75
rating = 4.8
weight = 72.5

print(product_price)
print(salary)
print(rating)
print(weight)

print(type(product_price))
print(type(salary))
print(type(rating))
print(type(weight))
````

# 🧮 3. Complex — `complex`

Python supports complex numbers.

A complex number contains:

- A real component
- An imaginary component

### Example

```python
z = 3 + 4j

````python
z = 3 + 4j 

type(z)
````

```python
z = 3 + 4j

print(z)
print(type(z))

print("Real part:", z.real)
print("Imaginary part:", z.imag)

# 🔤 4. String — `str`

A string represents **textual data**.

Strings can be created using:

- Single quotes `'...'`
- Double quotes `"..."`
- Triple quotes `'''...'''`
- Triple double quotes `"""..."""`

### Examples

```python
"Python"
'Data Analytics'
"Google"

````python
customer_name = "Alex"

print("The value is text, so Python identifies it as ", type(customer_name))
````

````python
customer_name = "Alex"
city = "Hyderabad"
course = "Python Data Analytics"
email = "alex@example.com"

print(customer_name)
print(city)
print(course)
print(email)

print(type(customer_name))
print(type(city))
print(type(course))
print(type(email))
````

# ✅ 5. Boolean — `bool`

A Boolean represents one of two logical states:

```text
True
False

````python
is_employee_active = True

print("The value is Boolean, so its type is ", type(is_employee_active))
````

````python
is_active = True
is_verified = False
has_subscription = True
is_deleted = False

print(is_active)
print(is_verified)
print(has_subscription)
print(is_deleted)

print(type(is_active))
print(type(is_verified))
print(type(has_subscription))
print(type(is_deleted))
````

# ⚪ 6. None — `NoneType`

Python has a special value called `None`.

`None` represents the **absence of a value** or "no value currently available."

Example:

```python
middle_name = None

````python
middle_name = None
phone_number = None
delivery_date = None

print(middle_name)
print(phone_number)
print(delivery_date)

print(type(middle_name))
print(type(phone_number))
print(type(delivery_date))
````

# 📊 Python Basic Data Types — Quick Reference

| Data Type | Python Type | Example | Typical Usage |
|---|---|---|---|
| Integer | `int` | `100` | Counts, age, quantity |
| Float | `float` | `99.99` | Price, salary, measurement |
| Complex | `complex` | `3 + 4j` | Scientific / mathematical work |
| String | `str` | `"Python"` | Text |
| Boolean | `bool` | `True` | Logical state |
| None | `NoneType` | `None` | Missing / unavailable value |

---

## 🧠 Remember

```text
25              → int
25.5            → float
3 + 4j          → complex
"Python"        → str
True / False     → bool
None             → NoneType

````python
# Data Type Practice

integer_value = 100
float_value = 99.99
complex_value = 3 + 4j
string_value = "Python"
boolean_value = True
none_value = None

values = [
    integer_value,
    float_value,
    complex_value,
    string_value,
    boolean_value,
    none_value
]

for value in values:
    print(value, "→", type(value).__name__)
````
