# 1. What Is OOP?

**Object-Oriented Programming (OOP)** is a programming style where data and
the behavior that works with that data are organized into objects.

### Why use OOP?

OOP becomes useful when an application contains many related entities such as:

- Customers
- Products
- Employees
- Bank accounts
- Orders
- Vehicles
- Students

Instead of keeping everything as unrelated variables and functions, we can
model these entities with classes and objects.

### Core vocabulary

```text
Class
  ↓
Blueprint

Object
  ↓
Actual instance of the blueprint

Attribute
  ↓
Data belonging to an object/class

Method
  ↓
Function defined inside a class
```

# 2. Classes

A class defines the structure and behavior of objects.

Basic syntax:

```python
class Student:
    pass
```

`pass` means that the class currently has no implementation.

````python
class Student:
    pass

student = Student()

print(type(student))
print(student.__class__.__name__)
````

# 3. Objects

An **object** is an instance created from a class.

```python
student1 = Student()
student2 = Student()
```

Both objects belong to the same class but can hold different state.

````python
class Student:
    pass

student1 = Student()
student2 = Student()

print(student1 is student2)
print(type(student1).__name__)
````

# 4. Attributes

Attributes store information associated with an object.

Commonly, instance attributes are created inside `__init__()` using `self`.

````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

student = Student("Kaif", 92)

print(student.name)
print(student.marks)
````

# 5. Methods

A method is a function defined inside a class.

Instance methods normally receive `self` as their first parameter.

```python
class Student:
    def display(self):
        ...
```

````python
class Student:
    def __init__(self, name):
        self.name = name

    def display(self):
        return f"Student: {self.name}"

student = Student("Kaif")
print(student.display())
````

# 6. The Constructor — `__init__()`

`__init__()` is an initializer that runs after an object is created.

It is commonly used to initialize instance attributes.

```python
class Student:
    def __init__(self, name):
        self.name = name
```

````python
class Employee:
    def __init__(self, name, department):
        self.name = name
        self.department = department

employee = Employee("Kaif", "Data Science")

print(employee.name)
print(employee.department)
````

# 7. Understanding `self`

`self` refers to the current instance.

It lets an instance method access that object's attributes and other methods.

```python
self.name
self.calculate_salary()
```

`self` is a conventional name and should be used for instance methods.

````python
class Customer:
    def __init__(self, name):
        self.name = name

    def greet(self):
        return f"Hello, {self.name}"

customer = Customer("Kaif")
print(customer.greet())
````

# 8. Instance Variables

Instance variables belong to individual objects.

Two objects can have different values for the same attribute.

````python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

laptop = Product("Laptop", 65000)
mouse = Product("Mouse", 1200)

print(laptop.name, laptop.price)
print(mouse.name, mouse.price)
````

# 9. Class Variables

A class variable is shared at the class level and is normally defined directly
inside the class body.

Use class variables for data that conceptually belongs to the class rather
than one specific instance.

````python
class Employee:
    company = "DataTech"

    def __init__(self, name):
        self.name = name

employee1 = Employee("Kaif")
employee2 = Employee("Sara")

print(employee1.company)
print(employee2.company)
print(Employee.company)
````

# 10. Instance vs Class Variables

```text
Class variable
    ↓
shared class-level value

Instance variable
    ↓

```

Be careful with mutable class variables: a shared list/dictionary may
intentionally or unintentionally be shared by every instance.

````python
class TeamMember:
    company = "DataTech"

    def __init__(self, name):
        self.name = name

a = TeamMember("Kaif")
b = TeamMember("Sara")

print(a.company)
print(b.company)
print(a.name)
print(b.name)
````

# 11. Encapsulation

Encapsulation means keeping related data and behavior together while controlling
how implementation details are accessed.

Python does not enforce traditional private fields in the same way as some
languages.

Common conventions:

```text
name       → public
_name      → internal-use convention
__name     → name mangling
```

````python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self._balance = balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit must be positive")
        self._balance += amount

    def get_balance(self):
        return self._balance

account = BankAccount("Kaif", 5000)
account.deposit(1500)

print(account.get_balance())
````

# 12. Name Mangling — `__name`

A double underscore at the beginning triggers name mangling for attributes.

This is mainly useful to avoid accidental name collisions in subclasses.
It is **not** a security mechanism.

````python
class Example:
    def __init__(self):
        self.__value = 100

    def get_value(self):
        return self.__value

obj = Example()
print(obj.get_value())
````

# 13. Inheritance

Inheritance allows a class to derive behavior from another class.

```text
Parent / Base Class
        ↓
Child / Derived Class
```

Use inheritance when the child truly represents a specialized form of the
parent.

````python
class Employee:
    def work(self):
        return "Employee is working"

class DataScientist(Employee):
    def analyze(self):
        return "Analyzing data"

person = DataScientist()

print(person.work())
print(person.analyze())
````

# 14. `super()`

`super()` provides access to methods of a parent class.

A common use is calling the parent's `__init__()` from a subclass.

````python
class Employee:
    def __init__(self, name):
        self.name = name

class DataScientist(Employee):
    def __init__(self, name, skill):
        super().__init__(name)
        self.skill = skill

person = DataScientist("Kaif", "Machine Learning")

print(person.name)
print(person.skill)
````

# 15. Method Overriding

A child class can provide its own implementation of an inherited method.

This is commonly used to specialize behavior.

````python
class Employee:
    def role(self):
        return "Employee"

class DataScientist(Employee):
    def role(self):
        return "Data Scientist"

print(Employee().role())
print(DataScientist().role())
````

# 16. Polymorphism

Polymorphism means the same operation/interface can work with different
object types.

Python's dynamic typing makes this especially natural.

### Duck typing

> If an object provides the behavior your code needs, it can often be used
> without requiring a particular inheritance relationship.

````python
class PDFReport:
    def export(self):
        return "Exporting PDF"

class ExcelReport:
    def export(self):
        return "Exporting Excel"

def export_report(report):
    print(report.export())

export_report(PDFReport())
export_report(ExcelReport())
````

# 17. Abstraction

Abstraction focuses on **what an object should do** while hiding unnecessary
implementation details.

Python provides formal abstract base classes through the `abc` module.

Important tools:

```python
ABC
abstractmethod
```

````python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

class CardPayment(Payment):
    def pay(self, amount):
        return f"Paid ₹{amount} by card"

payment = CardPayment()
print(payment.pay(2500))
````

# 18. Abstract Classes Cannot Be Used Like Normal Concrete Classes

An abstract class with unimplemented abstract methods cannot normally be
instantiated until those methods are implemented by a concrete subclass.

````python
from abc import ABC, abstractmethod

class Report(ABC):
    @abstractmethod
    def generate(self):
        pass

class SalesReport(Report):
    def generate(self):
        return "Sales report generated"

report = SalesReport()
print(report.generate())
````

# 19. Static Methods

A `@staticmethod` does not receive an automatic `self` or `cls` argument.

Use it when the operation logically belongs to the class but does not need
instance or class state.

````python
class Temperature:
    @staticmethod
    def celsius_to_fahrenheit(celsius):
        return (celsius * 9 / 5) + 32

print(Temperature.celsius_to_fahrenheit(25))
````

# 20. Class Methods

A `@classmethod` receives the class as `cls`.

Common use cases include alternative constructors and class-level behavior.

````python
class Employee:
    company = "DataTech"

    def __init__(self, name):
        self.name = name

    @classmethod
    def from_name(cls, name):
        return cls(name)

employee = Employee.from_name("Kaif")

print(employee.name)
print(employee.company)
````

# 21. Properties

`@property` lets a method be accessed like an attribute.

This is useful when you want controlled access or computed values without
changing the public interface.

````python
class Product:
    def __init__(self, price, tax_rate):
        self.price = price
        self.tax_rate = tax_rate

    @property
    def final_price(self):
        return self.price * (1 + self.tax_rate)

product = Product(1000, 0.18)

print(product.final_price)
````

# 22. Property Setter

A setter can validate or transform values when an attribute-like property is
assigned.

```python
@property
def value(self):
    ...

@value.setter
def value(self, new_value):
    ...
```

````python
class Employee:
    def __init__(self, salary):
        self.salary = salary

    @property
    def salary(self):
        return self._salary

    @salary.setter
    def salary(self, value):
        if value < 0:
            raise ValueError("Salary cannot be negative")
        self._salary = value

employee = Employee(50000)
employee.salary = 60000

print(employee.salary)
````

# 23. `__str__()`

`__str__()` defines a human-readable string representation of an object.

It is used by `str()` and `print()`.

````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def __str__(self):
        return f"{self.name} — {self.marks}%"

student = Student("Kaif", 92)

print(student)
````

# 24. `__repr__()`

`__repr__()` provides a developer-oriented representation.

A useful `repr` should make the object's state clear and unambiguous.

````python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __repr__(self):
        return f"Product(name={self.name!r}, price={self.price!r})"

product = Product("Laptop", 65000)

print(repr(product))
````

# 25. `__len__()`

Implement `__len__()` when an object has a meaningful length.

Then `len(object)` can use that implementation.

````python
class Playlist:
    def __init__(self, songs):
        self.songs = songs

    def __len__(self):
        return len(self.songs)

playlist = Playlist(["Song A", "Song B", "Song C"])

print(len(playlist))
````

# 26. `__eq__()`

`__eq__()` controls equality comparisons using `==`.

Define equality according to the logical identity of your domain object.

````python
class Product:
    def __init__(self, product_id, name):
        self.product_id = product_id
        self.name = name

    def __eq__(self, other):
        if not isinstance(other, Product):
            return NotImplemented
        return self.product_id == other.product_id

a = Product(101, "Laptop")
b = Product(101, "Laptop")
c = Product(102, "Mouse")

print(a == b)
print(a == c)
````

# 27. Other Useful Magic Methods

Python has many data-model methods.

Examples:

```text
__init__()  → initialize an instance
__str__()   → user-friendly string
__repr__()  → developer representation
__len__()   → len(object)
__eq__()    → object == object
__lt__()    → object < object
__iter__()  → iteration
__contains__() → membership testing
__enter__() / __exit__() → context managers
```

Do not implement magic methods just for decoration. Add them when the object
has a meaningful corresponding behavior.

````python
class Score:
    def __init__(self, value):
        self.value = value

    def __lt__(self, other):
        if not isinstance(other, Score):
            return NotImplemented
        return self.value < other.value

scores = [Score(80), Score(60), Score(90)]
scores.sort(key=lambda score: score.value)

print([score.value for score in scores])
````

# 28. Multiple Inheritance

Python supports multiple inheritance:

```python
class Child(ParentA, ParentB):
    ...
```

It can be useful, but it increases design complexity.

Prefer simple composition or single inheritance when they communicate the
design more clearly.

````python
class Logger:
    def log(self):
        return "Logged"

class Exporter:
    def export(self):
        return "Exported"

class ReportService(Logger, Exporter):
    pass

service = ReportService()

print(service.log())
print(service.export())
````

# 29. Method Resolution Order — MRO

Python uses an MRO to determine the order in which classes are searched for
attributes and methods.

Use:

```python
ClassName.mro()
```

or:

```python
ClassName.__mro__
```

````python
class A:
    pass

class B(A):
    pass

class C(B):
    pass

print(C.mro())
````

# 30. Composition

Composition means building an object from other objects instead of inheriting
from them.

This often models **has-a** relationships.

```text
Car HAS-A Engine
Order HAS-A Customer
Report HAS-A Formatter
```

````python
class Engine:
    def start(self):
        return "Engine started"

class Car:
    def __init__(self, engine):
        self.engine = engine

    def start(self):
        return self.engine.start()

car = Car(Engine())

print(car.start())
````

# 31. Inheritance vs Composition

| Question | Prefer |
|---|---|
| Is X genuinely a specialized Y? | Inheritance |
| Does X contain/use Y? | Composition |
| Need flexible components? | Composition |
| Need shared polymorphic interface? | Either, based on design |
| Deep inheritance hierarchy? | Usually simplify |

### Rule

**Prefer the simplest design that accurately represents the domain.**

# 32. Dataclasses and OOP

`dataclasses` can reduce boilerplate for classes primarily used to store data.

They automatically provide useful methods depending on configuration.

This is still part of Python OOP, but use it when the model is mainly data.

````python
from dataclasses import dataclass

@dataclass
class Employee:
    name: str
    department: str
    salary: float

employee = Employee("Kaif", "Data Science", 65000)

print(employee)
print(employee.name)
````

# 33. Type Hints in Classes

Type hints improve readability, tooling, and maintainability.

They do not automatically enforce types at runtime.

````python
class Product:
    def __init__(self, name: str, price: float):
        self.name: str = name
        self.price: float = price

    def discount(self, rate: float) -> float:
        return self.price * (1 - rate)

product = Product("Laptop", 65000.0)

print(product.discount(0.10))
````

# 34. Class Design Checklist

When creating a class, ask:

1. What real concept does this class represent?
2. What state should an object store?
3. Which behaviors belong to the object?
4. Which values are class-wide?
5. Should the class be inherited?
6. Would composition be simpler?
7. Does the class need validation?
8. Does it need a property?
9. Does it need meaningful magic methods?
10. Can another developer understand the class easily?

# 35. Common OOP Mistakes

### Mistake 1 — Forgetting `self`

```python
def display():
    ...
```

Instance methods normally need `self`.

### Mistake 2 — Confusing class and instance variables

A class variable is not automatically a separate copy for every object.

### Mistake 3 — Overusing inheritance

Inheritance should represent a meaningful relationship.

### Mistake 4 — Treating `_name` as a security boundary

A leading underscore is mainly a convention.

### Mistake 5 — Writing huge classes

A class should have a clear responsibility.

### Mistake 6 — Creating getters/setters for everything

Python properties and direct attributes can often be cleaner.

### Mistake 7 — Ignoring `super()`

When extending parent initialization, make sure the parent behavior is handled
appropriately.

### Mistake 8 — Implementing magic methods incorrectly

Return the expected type and respect Python's data-model conventions.

# 36. Best Practices

- Give classes clear, meaningful names.
- Keep each class focused.
- Keep methods small and understandable.
- Prefer composition when inheritance does not express the domain well.
- Use properties for controlled attribute access.
- Use `@classmethod` for alternative constructors.
- Use `@staticmethod` only when instance/class state is unnecessary.
- Use type hints where they improve clarity.
- Use dataclasses for suitable data-centric models.
- Implement magic methods only when they have useful semantics.
- Avoid unnecessary deep inheritance.
- Validate important inputs.
- Write tests for important class behavior.
- Keep public interfaces stable and easy to understand.

# 🚀 Project 1 — 01 — Student Management System

## Business Problem

Model students, calculate average marks, and determine grades.

## Architecture

```text
DOMAIN OBJECTS
      ↓
ATTRIBUTES
      ↓
METHODS
      ↓
VALIDATION
      ↓
OOP RELATIONSHIPS
      ↓
BUSINESS RESULT
```

## Implementation

````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    @property
    def average(self):
        return sum(self.marks) / len(self.marks)

    def grade(self):
        score = self.average
        if score >= 90:
            return "A"
        if score >= 75:
            return "B"
        if score >= 60:
            return "C"
        return "D"

    def __str__(self):
        return f"{self.name}: {self.average:.2f} ({self.grade()})"

students = [
    Student("Kaif", [90, 95, 92]),
    Student("Sara", [82, 88, 85]),
]

for student in students:
    print(student)
````

# 🚀 Project 2 — 02 — Bank Account System

## Business Problem

Encapsulate account balance and expose controlled deposit/withdraw operations.

## Architecture

```text
DOMAIN OBJECTS
      ↓
ATTRIBUTES
      ↓
METHODS
      ↓
VALIDATION
      ↓
OOP RELATIONSHIPS
      ↓
BUSINESS RESULT
```

## Implementation

````python
class BankAccount:
    def __init__(self, owner, balance=0.0):
        self.owner = owner
        self._balance = balance

    @property
    def balance(self):
        return self._balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit must be positive")
        self._balance += amount

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal must be positive")
        if amount > self._balance:
            raise ValueError("Insufficient balance")
        self._balance -= amount

account = BankAccount("Kaif", 10000)
account.deposit(2500)
account.withdraw(1500)

print("Owner:", account.owner)
print("Balance:", account.balance)
````

# 🚀 Project 3 — 03 — E-Commerce Product System

## Business Problem

Create products, calculate discounts, and display product information.

## Architecture

```text
DOMAIN OBJECTS
      ↓
ATTRIBUTES
      ↓
METHODS
      ↓
VALIDATION
      ↓
OOP RELATIONSHIPS
      ↓
BUSINESS RESULT
```

## Implementation

````python
class Product:
    tax_rate = 0.18

    def __init__(self, product_id, name, price):
        self.product_id = product_id
        self.name = name
        self.price = price

    def discounted_price(self, rate):
        if not 0 <= rate <= 1:
            raise ValueError("Rate must be between 0 and 1")
        return self.price * (1 - rate)

    @property
    def price_with_tax(self):
        return self.price * (1 + self.tax_rate)

    def __str__(self):
        return f"{self.product_id} | {self.name} | ₹{self.price:.2f}"

products = [
    Product(101, "Laptop", 65000),
    Product(102, "Mouse", 1200),
]

for product in products:
    print(product)
    print("10% discount:", product.discounted_price(0.10))
    print("With tax:", product.price_with_tax)
````

# 🚀 Project 4 — 04 — Employee Payroll System

## Business Problem

Use inheritance and polymorphism for different employee payment rules.

## Architecture

```text
DOMAIN OBJECTS
      ↓
ATTRIBUTES
      ↓
METHODS
      ↓
VALIDATION
      ↓
OOP RELATIONSHIPS
      ↓
BUSINESS RESULT
```

## Implementation

````python
class Employee:
    def __init__(self, name):
        self.name = name

    def calculate_pay(self):
        raise NotImplementedError

class SalariedEmployee(Employee):
    def __init__(self, name, monthly_salary):
        super().__init__(name)
        self.monthly_salary = monthly_salary

    def calculate_pay(self):
        return self.monthly_salary

class HourlyEmployee(Employee):
    def __init__(self, name, hours, rate):
        super().__init__(name)
        self.hours = hours
        self.rate = rate

    def calculate_pay(self):
        return self.hours * self.rate

employees = [
    SalariedEmployee("Kaif", 65000),
    HourlyEmployee("Sara", 160, 400),
]

for employee in employees:
    print(employee.name, "→", employee.calculate_pay())
````

# 🚀 Project 5 — 05 — Payment Processing System

## Business Problem

Use abstraction and polymorphism to create interchangeable payment implementations.

## Architecture

```text
DOMAIN OBJECTS
      ↓
ATTRIBUTES
      ↓
METHODS
      ↓
VALIDATION
      ↓
OOP RELATIONSHIPS
      ↓
BUSINESS RESULT
```

## Implementation

````python
from abc import ABC, abstractmethod

class PaymentMethod(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

class CardPayment(PaymentMethod):
    def pay(self, amount):
        return f"Card payment of ₹{amount:.2f}"

class UpiPayment(PaymentMethod):
    def pay(self, amount):
        return f"UPI payment of ₹{amount:.2f}"

def checkout(payment_method, amount):
    print(payment_method.pay(amount))

checkout(CardPayment(), 2500)
checkout(UpiPayment(), 1800)
````

# 🚀 Project 6 — 06 — Library Management System

## Business Problem

Use classes and composition to manage books and members.

## Architecture

```text
DOMAIN OBJECTS
      ↓
ATTRIBUTES
      ↓
METHODS
      ↓
VALIDATION
      ↓
OOP RELATIONSHIPS
      ↓
BUSINESS RESULT
```

## Implementation

````python
class Book:
    def __init__(self, title):
        self.title = title
        self.available = True

    def borrow(self):
        if not self.available:
            raise ValueError("Book is already borrowed")
        self.available = False

    def return_book(self):
        self.available = True

class Member:
    def __init__(self, name):
        self.name = name
        self.borrowed_books = []

    def borrow(self, book):
        book.borrow()
        self.borrowed_books.append(book)

    def return_book(self, book):
        book.return_book()
        self.borrowed_books.remove(book)

book = Book("Python Fundamentals")
member = Member("Kaif")

member.borrow(book)

print(member.name)
print([item.title for item in member.borrowed_books])
print(book.available)

member.return_book(book)
print(book.available)
````

# 🚀 Project 7 — 07 — Inventory Management System

## Business Problem

Use objects to manage stock quantities and inventory value.

## Architecture

```text
DOMAIN OBJECTS
      ↓
ATTRIBUTES
      ↓
METHODS
      ↓
VALIDATION
      ↓
OOP RELATIONSHIPS
      ↓
BUSINESS RESULT
```

## Implementation

````python
class InventoryItem:
    def __init__(self, sku, name, price, quantity):
        self.sku = sku
        self.name = name
        self.price = price
        self.quantity = quantity

    @property
    def stock_value(self):
        return self.price * self.quantity

    def add_stock(self, quantity):
        if quantity <= 0:
            raise ValueError("Quantity must be positive")
        self.quantity += quantity

    def sell(self, quantity):
        if quantity <= 0 or quantity > self.quantity:
            raise ValueError("Invalid sale quantity")
        self.quantity -= quantity

inventory = [
    InventoryItem("P101", "Laptop", 65000, 5),
    InventoryItem("P102", "Mouse", 1200, 20),
]

inventory[0].sell(2)
inventory[1].add_stock(10)

for item in inventory:
    print(item.sku, item.name, item.quantity, item.stock_value)
````

# 🚀 Project 8 — 08 — Data Analytics Report System

## Business Problem

Use polymorphism to allow different report formats to implement the same export interface.

## Architecture

```text
DOMAIN OBJECTS
      ↓
ATTRIBUTES
      ↓
METHODS
      ↓
VALIDATION
      ↓
OOP RELATIONSHIPS
      ↓
BUSINESS RESULT
```

## Implementation

````python
class Report:
    def __init__(self, title, rows):
        self.title = title
        self.rows = rows

    def export(self):
        raise NotImplementedError

class TextReport(Report):
    def export(self):
        return "\n".join(
            [self.title] + [f"{row['name']}: {row['sales']}" for row in self.rows]
        )

class CSVReport(Report):
    def export(self):
        lines = ["name,sales"]
        lines.extend(f"{row['name']},{row['sales']}" for row in self.rows)
        return "\n".join(lines)

rows = [
    {"name": "Kaif", "sales": 65000},
    {"name": "Sara", "sales": 48000},
]

for report in [TextReport("Sales Report", rows), CSVReport("Sales Report", rows)]:
    print(report.export())
    print("---")
````


## Implementation

````python
class Ticket:
    VALID_STATUSES = {"Open", "In Progress", "Closed"}

    def __init__(self, ticket_id, customer, issue):
        self.ticket_id = ticket_id
        self.customer = customer
        self.issue = issue
        self.status = "Open"

    def update_status(self, status):
        if status not in self.VALID_STATUSES:
            raise ValueError("Invalid ticket status")
        self.status = status

    def __str__(self):
        return f"#{self.ticket_id} | {self.customer} | {self.status} | {self.issue}"

ticket = Ticket(1001, "Kaif", "Login problem")

print(ticket)
ticket.update_status("In Progress")
print(ticket)
ticket.update_status("Closed")
print(ticket)
````

# 🚀 Project 10 — 10 — Mini Business Management System

## Business Problem

Combine customers, products, orders, and polymorphic payment behavior into a small domain model.

## Architecture

```text
DOMAIN OBJECTS
      ↓
ATTRIBUTES
      ↓
METHODS
      ↓
VALIDATION
      ↓
OOP RELATIONSHIPS
      ↓
BUSINESS RESULT
```

## Implementation

````python
from dataclasses import dataclass

@dataclass
class Customer:
    customer_id: int
    name: str

@dataclass
class Product:
    product_id: int
    name: str
    price: float

class Order:
    def __init__(self, customer):
        self.customer = customer
        self.items = []

    def add_product(self, product, quantity):
        if quantity <= 0:
            raise ValueError("Quantity must be positive")
        self.items.append((product, quantity))

    @property
    def total(self):
        return sum(product.price * quantity for product, quantity in self.items)

    def summary(self):
        return {
            "customer": self.customer.name,
            "items": [
                {
                    "product": product.name,
                    "quantity": quantity,
                    "amount": product.price * quantity
                }
                for product, quantity in self.items
            ],
            "total": self.total,
        }

customer = Customer(1, "Kaif")
laptop = Product(101, "Laptop", 65000)
mouse = Product(102, "Mouse", 1200)

order = Order(customer)
order.add_product(laptop, 1)
order.add_product(mouse, 2)

print(order.summary())
````
