# 1. What Is Type Conversion?

**Type conversion** means changing a value from one data type into another data type.

For example:

```python
age = "22"
```

Here, `age` is a string.

We can convert it into an integer:

```python
age = int(age)
```

Now `age` is an integer.

```text
String
  ↓
int()
  ↓
Integer
```

Type conversion is especially important when working with user input, files, APIs, databases, and calculations.

---

# 2. Why Is Type Conversion Important?

Python variables can contain different types of values.

For example:

```python
name = "Kaif"
age = 22
price = 499.99
```

These values have different data types.

Sometimes we need to convert one type into another so that Python can perform the required operation.

A common example is `input()`:

```python
age = input("Enter your age: ")
```

Even if the user enters:

```text
22
```

Python receives it as:

```python
str
```

If we want to perform numerical calculations, we need:

```python
age = int(age)
```

# 4. `int()`

The `int()` function converts a compatible value into an integer.

## Syntax

```python
int(value)
```

## Example

```python
number = int("100")

print(number)
print(type(number))
```

Output:

```text
100
<class 'int'>
```

---

## Converting Float to Integer

```python
number = int(25.9)

print(number)
```

Output:

```text
25
```

When converting a positive or negative floating-point number to `int`, the fractional portion is discarded; it does not round to the nearest integer.

---

## Converting Boolean to Integer

```python
print(int(True))
print(int(False))
```

Output:

```text
1
0
```

---

## `int()` With User Input

```python
age = int(input("Enter your age: "))

print(f"You are {age} years old.")
```

# 12. Implicit Type Conversion

**Implicit conversion** happens automatically in certain Python expressions when Python converts a value to a compatible type.

A common example is arithmetic involving integers and floating-point numbers.

```python
number = 10
price = 2.5

result = number + price

print(result)
print(type(result))
```

Output:

```text
12.5
<class 'float'>
```

Here:

```text
int + float
    ↓
float
```

Python automatically produces a floating-point result.

---

# 13. Another Implicit Conversion Example

```python

b = 2.0

result = a * b

print(result)
print(type(result))
```

Output:

```text
10.0
<class 'float'>
```

# 14. Explicit Type Conversion

**Explicit conversion** means that the programmer intentionally converts a value using a conversion function.

Example:

```python
age = "22"

age = int(age)

print(age)
print(type(age))
```

Output:

```text
22
<class 'int'>
```

Flow:

```text
"22"
  ↓
int()
  ↓
22
```

---

# 15. Implicit vs Explicit Conversion

| Feature      | Implicit            | Explicit                         |
| ------------ | ------------------- | -------------------------------- |
| Performed by | Python              | Programmer                       |
| Automatic    | Yes                 | No                               |
| Example      | `5 + 2.5`           | `int("10")`                      |
| Control      | Less                | More                             |
| Common Use   | Numeric expressions | User input / data transformation |

# 19. `type()` vs `isinstance()`

| Function       | Purpose                                                    |
| -------------- | ---------------------------------------------------------- |

# 20. Type Errors

A **TypeError** occurs when an operation or function is used with an inappropriate type.

Example:

```python
age = "22"

result = age + 5
```

This raises a `TypeError` because Python cannot add a string and an integer.

Correct version:

```python
age = "22"

result = int(age) + 5

print(result)
```

Output:

```text
27
```

---

# 21. Common `input()` Type Error

Incorrect:

```python
num1 = input("Enter first number: ")
num2 = input("Enter second number: ")

print(num1 + num2)
```

If the user enters:

```text
10
20
```

```text

```

Why?

Because both values are strings.

Correct:

```python
num1 = int(input("Enter first number: "))

print(num1 + num2)
```

Output:

```text
30
```

# 27. Collection Conversion

Python collections can often be converted between compatible iterable types.

```python
numbers = [10, 20, 30]

as_tuple = tuple(numbers)
as_set = set(numbers)

print(as_tuple)
print(as_set)
```

Output:

```text
(10, 20, 30)
{10, 20, 30}
```

Remember that converting to a set removes duplicates and does not preserve a meaningful positional order.

# 28. Common Mistakes

### Mistake 1 — Treating `input()` as a number

Incorrect:

```python
age = input("Enter age: ")

print(age + 1)
```

Correct:

```python
age = int(input("Enter age: "))

print(age + 1)
```

---

### Mistake 2 — Assuming `bool("False")` Is False

Incorrect assumption:

```python
print(bool("False"))
```

Output:

```text
True
```

Why?

Because `"False"` is a non-empty string.

---

### Mistake 3 — Expecting `int()` to Round

```python
number = int(9.99)

print(number)
```

Output:

```text
9
```

`int()` does not round the value to the nearest integer.

---

### Mistake 4 — Invalid Numeric Conversion

```python
number = int("Python")
```

This raises:

```text
ValueError
```

---

# 29. Best Practices

* Always know the current data type before converting values.
* Use `type()` or `isinstance()` when debugging or validating types.
* Convert user input explicitly when numerical data is required.
* Do not assume that numeric-looking strings are numbers.
* Validate external data before conversion in production applications.
* Use `float()` for decimal values when appropriate.
* Remember that `int()` does not round floating-point values.
* Remember that `set()` removes duplicates and does not provide positional ordering.
* Handle possible `ValueError` and `TypeError` when converting untrusted input.

### 10. Real-World Sales Data Pipeline

```python
raw_sales_data = [
    ["Laptop", "2", "55000.50"],
    ["Mouse", "5", "799.99"],
    ["Keyboard", "3", "1499.50"]
]

sales_data = []

for record in raw_sales_data:
    product = str(record[0])
    quantity = int(record[1])
    price = float(record[2])

    total = quantity * price

    sales_data.append({
        "product": product,
        "quantity": quantity,
        "price": price,
        "total": total
    })

print("Sales Report")
print("=============")

grand_total = 0

for sale in sales_data:
    print(f"Product: {sale['product']}")
    print(f"Quantity: {sale['quantity']}")
    print(f"Price: ₹{sale['price']:,.2f}")
    print(f"Total: ₹{sale['total']:,.2f}")
    print("-" * 30)

    grand_total += sale["total"]

print(f"Grand Total: ₹{grand_total:,.2f}")
```

````python
raw_sales_data = [
    ["Laptop", "2", "55000.50"],
    ["Mouse", "5", "799.99"],
    ["Keyboard", "3", "1499.50"]
]

sales_data = []

for record in raw_sales_data:
    product = str(record[0])
    quantity = int(record[1])
    price = float(record[2])

    total = quantity * price

    sales_data.append({
        "product": product,
        "quantity": quantity,
        "price": price,
        "total": total
    })

print("Sales Report")
print("=============")

grand_total = 0

for sale in sales_data:
    print(f"Product: {sale['product']}")
    print(f"Quantity: {sale['quantity']}")
    print(f"Price: ₹{sale['price']:,.2f}")
    print(f"Total: ₹{sale['total']:,.2f}")
    print("-" * 30)

    grand_total += sale["total"]

print(f"Grand Total: ₹{grand_total:,.2f}")
````
