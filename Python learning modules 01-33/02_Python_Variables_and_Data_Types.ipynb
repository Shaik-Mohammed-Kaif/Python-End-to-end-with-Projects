# 🧩 03 — PYTHON DATA TYPES

</div>

A **data type** describes the kind of value an object represents.

For the beginning of our Python journey, focus on these common types:

| Data Type | Meaning | Example |
|---|---|---|
| `int` | Whole number | `25` |
| `float` | Decimal number | `99.99` |
| `str` | Text | `"Python"` |
| `bool` | True/False value | `True` |
| `complex` | Complex number | `2 + 3j` |

### Visual Model

```text
                 PYTHON VALUES
                      │
       ┌──────────────┼──────────────┐
       ↓              ↓              ↓
     Numbers         Text         Boolean
       │              │              │
   ┌───┴───┐          │          True / False
   ↓       ↓          ↓
  int    float       str

````python
# Example :-
age = 25
price = 99.99
name = "Python"
is_active = True
complex_number = 2 + 3j
````

````python
# 🟦 — Check Data Types with `type()`

### Code Cell

age = 25
price = 99.99
name = "Python"
is_active = True
complex_number = 2 + 3j

print("age:", type(age))
print("price:", type(price))
print("name:", type(name))
print("is_active:", type(is_active))
print("complex_number:", type(complex_number))
````
