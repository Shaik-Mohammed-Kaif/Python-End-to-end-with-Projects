# 1. What Is a Dictionary?

A Python `dict` stores data as **key-value pairs**.

### Example

```python
student = {
    "name": "Kaif",
    "age": 22,
    "course": "Data Science"
}
```

Conceptually:

```text
Key       → Value
-------------------------
name      → Kaif
age       → 22
course    → Data Science
```

Dictionaries are ideal when you want to identify data by a meaningful
key instead of a numeric position.

### Important Characteristics

- Mutable
- Keys are unique
- Values may be duplicated
- Keys must be hashable
- Dictionaries preserve insertion order in modern Python
- Access is based on keys, not sequence indexes

````python
student = {
    "name": "Kaif",
    "age": 22,
    "course": "Data Science"
}

print(student)
print(type(student))
````

# 3. Keys and Values

Every dictionary entry contains:

```text
KEY → VALUE
```

Example:

```python
product = {
    "id": 101,
    "name": "Laptop",
    "price": 65000
}
```

Here:

```text
Keys   → id, name, price
Values → 101, Laptop, 65000
```

Keys are unique within a dictionary.

````python
product = {
    "id": 101,
    "name": "Laptop",
    "price": 65000
}

print("Keys:", product.keys())
print("Values:", product.values())
````

# 4. Accessing Values With `[]`

Use the key inside square brackets:

```python
student["name"]
```

If the key exists, its value is returned.

If the key does not exist, `KeyError` is raised.

````python
student = {
    "name": "Kaif",
    "age": 22,
    "course": "Python"
}

print(student["name"])
print(student["age"])
print(student["course"])
````

# 5. Safe Access With `get()`

`get()` is useful when a key may be missing.

```python
student.get("name")
```

If the key does not exist, `get()` returns `None` by default.

You can also provide a default:

```python
student.get("phone", "Not Available")
```

### Rule

```text
[]     → strict access
get()  → safe access
```

````python
student = {
    "name": "Kaif",
    "age": 22
}

print(student.get("name"))
print(student.get("phone"))
print(student.get("phone", "Not Available"))
````

# 9. `setdefault()`

`setdefault()` returns the value for a key.

If the key is missing, it inserts the key with the supplied default.

```python
data.setdefault("country", "India")
```

If `"country"` already exists, its existing value is not replaced.

````python
profile = {
    "name": "Kaif",
    "city": "Mumbai"
}

existing = profile.setdefault("city", "Delhi")
added = profile.setdefault("country", "India")

print("Existing value:", existing)
print("Added value:", added)
print(profile)
````

# 13. `items()`

This is one of the most important dictionary methods.

```python

```

````python
student = {
    "name": "Kaif",
    "age": 22,
    "course": "Python"
}

for key, value in student.items():
    print(f"{key}: {value}")
````

# 14. `pop()`

`pop(key)` removes the specified key and returns its value.

```python
age = student.pop("age")
```

You can optionally provide a default value when using `pop()`.

If the key is missing and no default is supplied, `KeyError` is raised.

````python
student = {
    "name": "Kaif",
    "age": 22,
    "course": "Python"
}

age = student.pop("age")

print("Removed age:", age)
print(student)

phone = student.pop("phone", "Not Available")
print("Phone:", phone)
````

# 15. `popitem()`

`popitem()` removes and returns the **last inserted key-value pair**.

It returns a two-item tuple:

```python
(key, value)
```

Calling it on an empty dictionary raises `KeyError`.

````python
data = {
    "a": 10,
    "b": 20,
    "c": 30
}

item = data.popitem()

print("Removed:", item)
print("Remaining:", data)
````

# 16. `clear()`

````python
data = {
    "name": "Kaif",
    "age": 22
}

data.clear()

print(data)
````

# 17. `copy()`

`copy()` creates a shallow copy.

```python
backup = data.copy()
```

For a flat dictionary, modifying the copied dictionary's top-level entries
does not modify the original.

````python
original = {
    "name": "Kaif",
    "skills": 3
}

backup = original.copy()

backup["skills"] = 4

print("Original:", original)
print("Backup:", backup)
````

# 20. Iterating Through a Dictionary

### Keys

```python
for key in data:
    print(key)
```

### Values

```python
for value in data.values():
    print(value)
```

### Key + Value

```python
for key, value in data.items():
    print(key, value)
```

`items()` is generally the clearest approach when both are needed.

````python
employee = {
    "id": 101,
    "name": "Sara",
    "department": "Data"
}

print("Keys:")
for key in employee:
    print(key)

print("\nValues:")
for value in employee.values():
    print(value)

print("\nPairs:")
for key, value in employee.items():
    print(key, "→", value)
````

# 21. Dictionary Comprehension

A dictionary comprehension creates dictionaries concisely.

### Syntax

```python
{key_expression: value_expression for item in iterable}
```

Example:

````python
numbers = [1, 2, 3, 4, 5]

squares = {number: number ** 2 for number in numbers}

print(squares)
````

# 22. Conditional Dictionary Comprehension

You can include a condition.

```python
even_squares = {

    if n % 2 == 0
}
```

````python
numbers = range(1, 11)

even_squares = {
    n: n ** 2
    for n in numbers
    if n % 2 == 0
}

print(even_squares)
````

# 25. Sorting Dictionaries

`sorted()` can sort dictionary keys or a sequence of dictionary records.

For a list of dictionaries:

```python
sorted(records, key=lambda record: record["salary"])
```

Use `reverse=True` for descending order.

````python
employees = [
    {"name": "Kaif", "salary": 60000},
    {"name": "Ali", "salary": 75000},
    {"name": "Sara", "salary": 68000}
]

highest_first = sorted(
    employees,
    key=lambda employee: employee["salary"],
    reverse=True
)

for employee in highest_first:
    print(employee["name"], employee["salary"])
````

# 27. Dictionary Key Rules

Keys must be hashable.

Common examples of valid keys:

```python
{
    "name": "Kaif",
    101: "Employee",
    (10, 20): "Coordinate"
}
```

A mutable list cannot be a dictionary key:

```python
{[1, 2]: "value"}
```

because lists are unhashable.

Keys should also be meaningful and unique within the dictionary.

````python
valid = {
    "name": "Kaif",
    101: "Employee",
    (10, 20): "Coordinate"
}

print(valid)

try:
    invalid = {[1, 2]: "value"}
except TypeError as error:
    print("Error:", error)
````

# 28. Duplicate Keys

A dictionary cannot retain two separate values for the same key.

If a key is repeated in a literal, the later value replaces the earlier
one.

```python
data = {
    "score": 80,
    "score": 95
}
```

The resulting dictionary contains `"score": 95`.

````python
data = {
    "score": 80,
    "score": 95
}

print(data)
````

# 30. Merging Dictionaries

Modern Python provides multiple ways to combine dictionaries.

### `update()`

```python
a.update(b)
```

This changes `a`.

### Merge operator `|`

```python
merged = a | b
```

This creates a new dictionary.

### In-place merge `|=`

```python
a |= b
```

This updates `a`.

If the same key exists in both dictionaries, the right-hand value wins.

````python
profile = {
    "name": "Kaif",
    "city": "Mumbai"
}

contact = {
    "email": "kaif@example.com",
    "city": "Delhi"
}

merged = profile | contact

print("Original:", profile)
print("Merged:", merged)

profile |= contact

print("Updated:", profile)
````

# 31. Dictionary Comparison

Dictionaries can be compared for equality.

Two dictionaries are equal when they contain the same key-value mappings,
regardless of insertion order.

```python
{"a": 1, "b": 2} == {"b": 2, "a": 1}
```

returns `True`.

````python
first = {"a": 1, "b": 2}
second = {"b": 2, "a": 1}

print(first == second)
````

# 32. Common Dictionary Mistakes

### Mistake 1 — Using a missing key with `[]`

```python

```

can raise `KeyError`.

Use `get()` when missing keys are expected.

### Mistake 2 — Thinking `in` checks values

```python
"Kaif" in data
```

checks keys, not values.

### Mistake 3 — Repeating keys

Duplicate keys overwrite earlier values.

### Mistake 4 — Using mutable keys

Lists cannot be dictionary keys.

### Mistake 5 — Modifying structure while iterating

Avoid adding/removing dictionary keys while directly iterating over that
same dictionary. Build a separate result or iterate over a snapshot when
appropriate.

# 🚀 Project 01 — Student Marks Management System

## Business Problem

Build a student record system using dictionaries.

Each student should have:

```text
ID
Name
Department
Marks
```

Calculate:

- Student count
- Total marks
- Average marks
- Highest scorer
- Lowest scorer
- Pass/fail status
- Search by student ID

````python
students = {
    101: {
        "name": "Kaif",
        "department": "Data Science",
        "marks": 88
    },
    102: {
        "name": "Ali",
        "department": "Computer Science",
        "marks": 76
    },
    103: {
        "name": "Sara",
        "department": "Data Science",
        "marks": 94
    },
    104: {
        "name": "Omar",
        "department": "AI",
        "marks": 67
    }
}

marks = [student["marks"] for student in students.values()]

print("Student Count:", len(students))
print("Total Marks:", sum(marks))
print(f"Average Marks: {sum(marks) / len(marks):.2f}")

top_id, top_student = max(
    students.items(),
    key=lambda item: item[1]["marks"]
)

low_id, low_student = min(
    students.items(),
    key=lambda item: item[1]["marks"]
)

print("Top Student:", top_student["name"], top_student["marks"])
print("Lowest Student:", low_student["name"], low_student["marks"])

print("\nResults:")

for student_id, student in students.items():
    result = "Pass" if student["marks"] >= 40 else "Fail"
    print(student_id, student["name"], result)

search_id = 103

if search_id in students:
    print("\nSearch Result:", students[search_id])
else:
    print("\nStudent not found.")
````

# 🚀 Project 02 — Inventory Management System

## Business Problem

Create an inventory using product IDs.

Track:

```text
Product Name
Category
Price
Stock
```

Perform:

- Product search
- Stock update
- New product insertion
- Low-stock identification
- Inventory value calculation
- Product sorting

````python
inventory = {
    1001: {
        "name": "Laptop",
        "category": "Electronics",
        "price": 65000,
        "stock": 12
    },
    1002: {
        "name": "Mouse",
        "category": "Accessories",
        "price": 1200,
        "stock": 35
    },
    1003: {
        "name": "Monitor",
        "category": "Electronics",
        "price": 18000,
        "stock": 8
    }
}

inventory[1004] = {
    "name": "Keyboard",
    "category": "Accessories",
    "price": 2500,
    "stock": 20
}

inventory[1002]["stock"] -= 5

low_stock_threshold = 10

print("LOW STOCK PRODUCTS")

for product_id, product in inventory.items():
    if product["stock"] <= low_stock_threshold:
        print(
            product_id,
            product["name"],
            "Stock:",
            product["stock"]
        )

total_inventory_value = sum(
    product["price"] * product["stock"]
    for product in inventory.values()
)

print("\nTotal Inventory Value: ₹", total_inventory_value)

sorted_inventory = sorted(
    inventory.items(),
    key=lambda item: item[1]["price"],
    reverse=True
)

print("\nProducts by Price:")

for product_id, product in sorted_inventory:
    print(product["name"], "₹", product["price"])
````

# 🚀 Project 03 — Expense Tracker

## Business Problem

Build an expense tracker where each expense has:

```text
ID
Category
Description
Amount
```

Calculate:

- Total expense
- Average expense
- Category totals
- Highest expense
- Lowest expense
- Expenses above a threshold

````python
expenses = {
    1: {
        "category": "Food",
        "description": "Lunch",
        "amount": 250
    },
    2: {
        "category": "Transport",
        "description": "Cab",
        "amount": 450
    },
    3: {
        "category": "Shopping",
        "description": "Keyboard",
        "amount": 2500
    },
    4: {
        "category": "Food",
        "description": "Dinner",
        "amount": 500
    }
}

amounts = [
    expense["amount"]
    for expense in expenses.values()
]

print("Total Expense: ₹", sum(amounts))
print(f"Average Expense: ₹{sum(amounts) / len(amounts):.2f}")

highest = max(
    expenses.items(),
    key=lambda item: item[1]["amount"]
)

lowest = min(
    expenses.items(),
    key=lambda item: item[1]["amount"]
)

print(
    "Highest:",
    highest[1]["description"],
    "₹",
    highest[1]["amount"]
)

print(
    "Lowest:",
    lowest[1]["description"],
    "₹",
    lowest[1]["amount"]
)

category_totals = {}

for expense in expenses.values():
    category = expense["category"]
    amount = expense["amount"]

    category_totals[category] = (
        category_totals.get(category, 0) + amount
    )

print("\nCategory Totals:")

for category, total in category_totals.items():
    print(category, "₹", total)
````

# 🚀 Project 04 — Employee Payroll Analyzer

## Business Problem

Store employee information using dictionaries.

Track:

```text
Employee ID
Name
Department
Salary
Experience
```

Analyze:

- Total payroll
- Average salary
- Highest-paid employee
- Department payroll
- Employees above average salary
- Senior employees

````python
employees = {
    101: {
        "name": "Kaif",
        "department": "Data",
        "salary": 65000,
        "experience": 2
    },
    102: {
        "name": "Ali",
        "department": "Engineering",
        "salary": 85000,
        "experience": 5
    },
    103: {
        "name": "Sara",
        "department": "Data",
        "salary": 78000,
        "experience": 4
    },
    104: {
        "name": "Omar",
        "department": "HR",
        "salary": 55000,
        "experience": 6
    }
}

salaries = [employee["salary"] for employee in employees.values()]

total_payroll = sum(salaries)
average_salary = total_payroll / len(salaries)

highest = max(
    employees.items(),
    key=lambda item: item[1]["salary"]
)

print("Employee Count:", len(employees))
print("Total Payroll: ₹", total_payroll)
print(f"Average Salary: ₹{average_salary:.2f}")
print(
    "Highest Paid:",
    highest[1]["name"],
    "₹",
    highest[1]["salary"]
)

department_payroll = {}

for employee in employees.values():
    department = employee["department"]
    salary = employee["salary"]

    department_payroll[department] = (
        department_payroll.get(department, 0) + salary
    )

print("\nDepartment Payroll:")

for department, payroll in department_payroll.items():
    print(department, "₹", payroll)

print("\nAbove Average:")

for employee in employees.values():
    if employee["salary"] > average_salary:
        print(employee["name"])

print("\nSenior Employees:")

for employee in employees.values():
    if employee["experience"] >= 5:
        print(employee["name"])
````

# 🚀 Project 05 — Customer Analytics System

## Business Problem

Build a customer dataset using dictionaries.

Track:

```text
Customer ID
Name
City
Plan
Monthly Spend
Status
```

Analyze:

- Active customers
- Churned customers
- Total revenue
- Average monthly spend
- Customers by plan
- Highest-value customer
- Customers above average spend

````python
customers = {
    1001: {
        "name": "Kaif",
        "city": "Mumbai",
        "plan": "Premium",
        "monthly_spend": 2499,
        "status": "Active"
    },
    1002: {
        "name": "Ali",
        "city": "Delhi",
        "plan": "Basic",
        "monthly_spend": 799,
        "status": "Active"
    },
    1003: {
        "name": "Sara",
        "city": "Pune",
        "plan": "Premium",
        "monthly_spend": 2999,
        "status": "Churned"
    },
    1004: {
        "name": "Omar",
        "city": "Bengaluru",
        "plan": "Standard",
        "monthly_spend": 1499,
        "status": "Active"
    }
}

spends = [
    customer["monthly_spend"]
    for customer in customers.values()
]

total_revenue = sum(spends)
average_spend = total_revenue / len(spends)

active = sum(
    customer["status"] == "Active"
    for customer in customers.values()
)

churned = sum(
    customer["status"] == "Churned"
    for customer in customers.values()
)

print("Customers:", len(customers))
print("Active:", active)
print("Churned:", churned)
print("Monthly Revenue: ₹", total_revenue)
print(f"Average Spend: ₹{average_spend:.2f}")

highest_value = max(
    customers.items(),
    key=lambda item: item[1]["monthly_spend"]
)

print(
    "Highest Value Customer:",
    highest_value[1]["name"],
    "₹",
    highest_value[1]["monthly_spend"]
)

plan_counts = {}

for customer in customers.values():
    plan = customer["plan"]
    plan_counts[plan] = plan_counts.get(plan, 0) + 1

print("\nCustomers by Plan:")

for plan, count in plan_counts.items():
    print(plan, count)

print("\nAbove Average Spend:")

for customer in customers.values():
    if customer["monthly_spend"] > average_spend:
        print(customer["name"])
````

`get()`, loops, comprehensions, `max()`, `sorted()`, and aggregation
patterns.

````python
sales = [
    {
        "id": 1001,
        "product": "Laptop",
        "category": "Electronics",
        "region": "West",
        "quantity": 2,
        "unit_price": 65000
    },
    {
        "id": 1002,
        "product": "Mouse",
        "category": "Accessories",
        "region": "North",
        "quantity": 10,
        "unit_price": 1200
    },
    {
        "id": 1003,
        "product": "Monitor",
        "category": "Electronics",
        "region": "West",
        "quantity": 4,
        "unit_price": 18000
    },
    {
        "id": 1004,
        "product": "Keyboard",
        "category": "Accessories",
        "region": "South",
        "quantity": 7,
        "unit_price": 2500
    },
    {
        "id": 1005,
        "product": "Laptop",
        "category": "Electronics",
        "region": "North",
        "quantity": 1,
        "unit_price": 65000
    }
]

for transaction in sales:
    transaction["revenue"] = (
        transaction["quantity"]
        * transaction["unit_price"]
    )

total_revenue = sum(
    transaction["revenue"]
    for transaction in sales
)

total_quantity = sum(
    transaction["quantity"]
    for transaction in sales
)

average_transaction = total_revenue / len(sales)

print("=" * 60)
print("             SALES ANALYTICS REPORT")
print("=" * 60)

print("Transactions:", len(sales))
print("Total Quantity:", total_quantity)
print(f"Total Revenue: ₹{total_revenue:,.2f}")
print(f"Average Transaction: ₹{average_transaction:,.2f}")

product_revenue = {}
category_revenue = {}
region_revenue = {}
product_quantity = {}

for transaction in sales:
    product = transaction["product"]
    category = transaction["category"]
    region = transaction["region"]
    revenue = transaction["revenue"]
    quantity = transaction["quantity"]

    product_revenue[product] = (
        product_revenue.get(product, 0) + revenue
    )

    category_revenue[category] = (
        category_revenue.get(category, 0) + revenue
    )

    region_revenue[region] = (
        region_revenue.get(region, 0) + revenue
    )

    product_quantity[product] = (
        product_quantity.get(product, 0) + quantity
    )

best_product = max(
    product_quantity.items(),
    key=lambda item: item[1]
)

highest_transaction = max(
    sales,
    key=lambda transaction: transaction["revenue"]
)

print("\nProduct Revenue:")

for product, revenue in sorted(
    product_revenue.items(),
    key=lambda item: item[1],
    reverse=True
):
    print(f"{product}: ₹{revenue:,.2f}")

print("\nCategory Revenue:")

for category, revenue in category_revenue.items():
    print(f"{category}: ₹{revenue:,.2f}")

print("\nRegion Revenue:")

for region, revenue in region_revenue.items():
    print(f"{region}: ₹{revenue:,.2f}")

print(
    "\nBest-Selling Product:",
    best_product[0],
    "| Quantity:",
    best_product[1]
)

print(
    "Highest-Revenue Transaction:",
    highest_transaction["id"],
    "|",
    highest_transaction["product"],
    "| ₹",
    f"{highest_transaction['revenue']:,.2f}"
)

print("\nSorted Transaction Report:")

for transaction in sorted(
    sales,
    key=lambda item: item["revenue"],
    reverse=True
):
    print(
        transaction["id"],
        "|",
        transaction["product"],
        "|",
        transaction["region"],
        "|",
        f"₹{transaction['revenue']:,.2f}"
    )
````

Key may be missing?

 data[key] = value

Add/update many?
        ↓
     update()

Get keys + values?
        ↓

Remove a known key?
        ↓

        ↓

Create from keys?

 dictionary comprehension

Combine dictionaries?

# 🎯 Dictionary vs Other Python Collections

| Requirement | Best Choice |
|---|---|
| Ordered changeable sequence | List |
| Fixed ordered values | Tuple |
| Unique values / set operations | Set |
| Named key-value data | Dictionary |

student_list = ["Kaif", 22, "Data Science"]

For structured records where each value has a meaningful name, a
